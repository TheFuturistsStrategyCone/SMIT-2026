## ⚖️ 2. BMI Health Checker

### Goal:
    Help a fitness app tell users which weight category they belong to.

### Steps:

    Ask for weight in kg and height in metres.
    
    Calculate BMI = weight / (height * height).
    
    BMI below 18.5 → "Underweight".
    
    18.5 to 24.9 → "Normal".
    
    25 to 29.9 → "Overweight".
    
    30 or more → "Obese".
    
    Print the BMI rounded to 2 decimals.

### Example:

    Enter weight: 70
    Enter height: 1.75
    BMI = 22.86 → Normal

```python
# Test 1 Input: 70, 1.75 -> Expected Output: BMI = 22.86 -> Normal
# Test 2 Input: 95, 1.80 -> Expected Output: BMI = 29.32 -> Overweight

# Step 1: Gather user weight and height
weight_input = input("Enter weight in kg: ").strip()
height_input = input("Enter height in metres: ").strip()

# Safely proceed if both inputs are provided (truthy check)
if not weight_input or not height_input:
    print("Invalid Input: Weight and height cannot be empty.")
else:
    weight = float(weight_input)
    height = float(height_input)

    # Step 2: Prevent Division by Zero if height is 0
    if height <= 0:
        print("Invalid Input: Height must be greater than zero.")
    else:
        # Step 3: Compute and round the BMI
        bmi = weight / (height * height)
        rounded_bmi = round(bmi, 2)

        # Step 4: Map BMI value to weight categories
        if bmi < 18.5:
            category = "Underweight"
        elif bmi <= 24.9:
            category = "Normal"
        elif bmi <= 29.9:
            category = "Overweight"
        else:
            category = "Obese"

        # Step 5: Print final evaluation
        print(f"BMI = {rounded_bmi} -> {category}")
```

## 🌡️ 8. Temperature Unit Converter

### Goal:
    Build a small converter for a weather app.

### Steps:

    Show a menu: 1 = Celsius to Fahrenheit, 2 = Fahrenheit to Celsius.
    
    Ask for the choice and then the temperature.
    
    Choice 1 → F = C * 9/5 + 32.
    
    Choice 2 → C = (F - 32) * 5/9.
    
    Any other choice → "Invalid choice".

### Example:

    Enter choice: 1
    Enter temperature: 100
    212.0 F

```python
# Test 1 Input: 1, 100 -> Expected Output: 212.0 F
# Test 2 Input: 2, 212 -> Expected Output: 100.0 C

# Step 1: Render configuration menu options and grab choice
print("Menu: 1 = Celsius to Fahrenheit, 2 = Fahrenheit to Celsius")
choice_input = input("Enter choice: ").strip()

if choice_input != "1" and choice_input != "2":
    print("Invalid choice")
else:
    # Step 2: Grab the operational baseline temperature value
    temp_input = input("Enter temperature: ").strip()
    
    if not temp_input:
        print("Invalid Input: Temperature value cannot be empty.")
    else:
        temperature = float(temp_input)
        
        # Step 3: Run target mathematical matrix routing
        if choice_input == "1":
            fahrenheit = temperature * 9 / 5 + 32
            print(f"{round(fahrenheit, 2)} F")
        else:
            celsius = (temperature - 32) * 5 / 9
            print(f"{round(celsius, 2)} C")
```

```python
# Step 1: Create visitors list
visitors = ["Ali", "Sara", "Ali", "Ahmed", "Sara"]

# Initialize a new list to store unique names
unique_visitors = []

# Step 2: Loop through the list
for name in visitors:
    # Step 3: Add a name to a new list only if it is not already in it
    if name not in unique_visitors:
        unique_visitors.append(name)

# Step 4: Print the unique visitors and their count
print(f"Unique: {unique_visitors}")
print(f"Count = {len(unique_visitors)}")
```

## 🎫 49. Roll Number Order Check

### Goal:
    A teacher checks whether roll numbers were entered in ascending order.

### Steps:

    Create a list of roll numbers.
    
    Loop through the list and compare each number with the next one.
    
    Use a flag variable that becomes False if any number is out of order.
    
    Print "Sorted" or "Not sorted".

### Example:

    [101, 102, 105, 110] → Sorted
    [101, 105, 103] → Not sorted

```python
# Step 1: Create a list of roll numbers
roll_numbers = [101, 102, 105, 110]  # Change this to [101, 105, 103] to test the other case

# Step 3: Initialize a flag variable
is_sorted = True

# Step 2: Loop through the list and compare each number with the next one
# We stop at len(roll_numbers) - 1 so we don't go out of bounds
for i in range(len(roll_numbers) - 1):
    if roll_numbers[i] > roll_numbers[i + 1]:
        is_sorted = False
        break  # No need to check further if one pair is out of order

# Step 4: Print the final result based on the flag
if is_sorted:
    print("Sorted")
else:
    print("Not sorted")
```

## 🌡️ 50. Weekly Temperature Report

### Goal:
    A weather station wants to know how many days were warmer than average.

### Steps:

    Create temps = [30, 32, 28, 35, 31, 29, 36].
    
    First loop: add all the temperatures and calculate the average.
    
    Second loop: count the days above the average.
    
    Print the average (2 decimals) and the count.

### Example:

    Average = 31.57
    Days above average = 3

```python
# Step 1: Create temps list
temps = [30, 32, 28, 35, 31, 29, 36]

# Initialize sum variables
total_sum = 0
above_average_count = 0

# Step 2: First loop - add all the temperatures
for temp in temps:
    total_sum += temp

# Calculate the average
average = total_sum / len(temps)

# Step 3: Second loop - count the days above the average
for temp in temps:
    if temp > average:
        above_average_count += 1

# Step 4: Print the average (2 decimals) and the count
print(f"Average = {average:.2f}")
print(f"Days above average = {above_average_count}")
```
