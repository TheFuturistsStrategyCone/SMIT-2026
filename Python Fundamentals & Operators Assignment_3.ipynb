## A student is buying a laptop.

    Laptop price: Rs. 85,000
    Student has: Rs. 100,000
    Student gets a discount of Rs. 5,000
    A laptop bag costs Rs. 2,500
    The store has a promotion code "SAVE10".

### Create variables and:

    Calculate the laptop price after discount.
    Add the bag price to get the final bill.
    Calculate the remaining money.
    Check whether the student has enough money.
    Check whether "SAVE10" is present in the promotion code.
    Check whether the final bill is greater than Rs. 80,000.
    Store the result of whether the student has enough money in a variable and check whether it is truthy.

```python
# 1. Define initial variables
laptop_price = 85000
student_money = 100000
discount = 5000
bag_price = 2500
promo_code = "SAVE10"

# 2. Perform financial calculations
discounted_laptop_price = laptop_price - discount
final_bill = discounted_laptop_price + bag_price
remaining_money = student_money - final_bill

# 3. Perform comparison and membership checks
has_enough_money = student_money >= final_bill
is_promo_valid = "SAVE10" in promo_code
is_bill_above_80k = final_bill > 80000

# 4. Check the truthiness of the balance validation variable
is_enough_truthy = bool(has_enough_money)

# 5. Output all values and checks
print(f"--- Calculations ---")
print(f"Laptop Price after Discount: Rs. {discounted_laptop_price}")
print(f"Final Bill (with bag): Rs. {final_bill}")
print(f"Remaining Money: Rs. {remaining_money}\n")

print(f"--- Verification Checks ---")
print(f"Does the student have enough money?: {has_enough_money}")
print(f"Is 'SAVE10' present in promo code?: {is_promo_valid}")
print(f"Is the final bill greater than Rs. 80,000?: {is_bill_above_80k}")
print(f"Is the validation result truthy?: {is_enough_truthy}")
```

### You are creating a student result calculator.

A student has:

    Mathematics: 85
    English: 72
    Python: 90
    Total possible marks per subject: 100
    Student name: "Ahmed"
    Required passing marks per subject: 50

### Create variables and:

    Calculate the student's total marks.
    Calculate the average marks.
    Calculate the percentage.
    Check whether the student passed Mathematics.
    Check whether the student passed all three subjects.
    Check whether the name contains the letter "A".
    Check whether the average marks are greater than or equal to 80.
    Store the final percentage and check whether it is truthy.

Print all results.

```python
# 1. Define initial variables
student_name = "Ahmed"
math_score = 85
english_score = 72
python_score = 90

marks_per_subject = 100
passing_marks = 50
total_subjects = 3

# 2. Perform academic calculations
total_marks = math_score + english_score + python_score
average_marks = total_marks / total_subjects
max_possible_marks = marks_per_subject * total_subjects
percentage = (total_marks / max_possible_marks) * 100

# 3. Perform comparison and membership checks
passed_math = math_score >= passing_marks
passed_all = (math_score >= passing_marks) and (english_score >= passing_marks) and (python_score >= passing_marks)
name_contains_A = "A" in student_name  # Note: This checks for capital 'A'
is_average_distinction = average_marks >= 80

# 4. Check truthiness of the percentage variable
is_percentage_truthy = bool(percentage)

# 5. Output all calculations and check reports
print(f"--- Academic Performance for {student_name} ---")
print(f"Total Marks Scored: {total_marks} / {max_possible_marks}")
print(f"Average Marks: {average_marks:.2f}")
print(f"Final Percentage: {percentage:.2f}%\n")

print(f"--- Verification Status ---")
print(f"Did the student pass Mathematics?: {passed_math}")
print(f"Did the student pass all three subjects?: {passed_all}")
print(f"Does the student's name contain the letter 'A'?: {name_contains_A}")
print(f"Are average marks greater than or equal to 80?: {is_average_distinction}")
print(f"Is the final percentage value truthy?: {is_percentage_truthy}")
```
