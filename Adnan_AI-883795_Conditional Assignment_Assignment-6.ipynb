## Write a program that takes a number from the user and:

    Prints "Positive" if the number is greater than 0

    Prints "Negative" if the number is less than 0
    
    Prints "Zero" if the number is 0

```python
def check_number_sign(number: float) -> str:
    """Determine if a number is positive, negative, or zero."""
    if number > 0:
        return "Positive"
    elif number < 0:
        return "Negative"
    else:
        return "Zero"


def main() -> None:
    """Get user input and evaluate the number."""
    try:
        # Prompt the user for input and convert to a float to support decimals
        user_input = input("Enter a number: ")
        number = float(user_input)

        # Get the evaluation result and print it
        result = check_number_sign(number)
        print(result)

    except ValueError:
        print("Error: Invalid input. Please enter a valid number.")


if __name__ == "__main__":
    main()
```

## Ask the user for:

    Two numbers
    
    An operator (+, -, *, /)
    
    if operator + perform addition
    if operator - perform subtraction
    if operator * perform multiplication
    if operator / perform division

```python
def calculate(num1: float, num2: float, operator: str) -> float | str:
    """Perform mathematical operations based on the provided operator string."""
    if operator == "+":
        return num1 + num2
    elif operator == "-":
        return num1 - num2
    elif operator == "*":
        return num1 * num2
    elif operator == "/":
        # Handle structural zero division constraints safely
        if num2 == 0:
            return "Error: Division by zero is undefined."
        return num1 / num2
    else:
        return "Error: Invalid operator chosen."


def main() -> None:
    """Gather user inputs, execute calculation logic, and display output."""
    try:
        # Prompt user for sequential numerical values
        input1 = input("Enter first number: ")
        num1 = float(input1)

        input2 = input("Enter second number: ")
        num2 = float(input2)

        # Prompt for target operator context
        operator = input("Enter an operator (+, -, *, /): ").strip()

        # Execute processing structure
        result = calculate(num1, num2, operator)
        print(f"Result: {result}")

    except ValueError:
        print("Error: Invalid numeric input. Please enter valid numbers.")


if __name__ == "__main__":
    main()
```

## Online Order System

    Ask:
    
    Order amount
    
    User type (regular / premium)
    
### Conditions:

    If order ≥ 1000:
    
    Premium → 20% discount
    
    Regular → 10% discount
    
    Else → "No discount available"

```python
def calculate_discount(order_amount: float, user_type: str) -> str:
    """Calculate the discount tier and return the final pricing message."""
    if order_amount >= 1000:
        if user_type == "premium":
            discount = 0.20  # 20%
            discount_pct = "20%"
        else:  # regular
            discount = 0.10  # 10%
            discount_pct = "10%"
        
        saved_amount = order_amount * discount
        final_price = order_amount - saved_amount
        return f"{user_type.capitalize()} discount applied: {discount_pct}. You save: ${saved_amount:,.2f}. Final price: ${final_price:,.2f}."
    else:
        return f"No discount available. Total price: ${order_amount:,.2f}."


def main() -> None:
    """Get e-commerce order parameters from the user and calculate total price."""
    try:
        # Prompt for the monetary value of the order
        amount_input = input("Enter your order amount: ")
        order_amount = float(amount_input)

        if order_amount <= 0:
            print("Error: Order amount must be greater than 0.")
            return

        # Prompt for user classification and clean up inputs
        user_type = input("Enter user type (regular/premium): ").strip().lower()

        if user_type not in ["regular", "premium"]:
            print("Error: Invalid user type. Please enter exactly 'regular' or 'premium'.")
            return

        # Process conditions and print summary
        result = calculate_discount(order_amount, user_type)
        print(result)

    except ValueError:
        print("Error: Invalid numeric input. Please enter a valid number for the order amount.")


if __name__ == "__main__":
    main()
```

## A traffic system calculates fines:

Speed limit = 60 km/h
* If speed ≤ 60 → No fine
* If 61–80 → Fine = 500
* If 81–100 → Fine = 2000
* If >100 → License suspended
#### BUT:
*If emergency vehicle → no fine
*If rainy weather AND speed > 60 → fine increases by 50%

```python
def calculate_traffic_fine(speed: float, is_emergency: str, is_rainy: str) -> str:
    """Calculate traffic violation penalties based on environmental factors and vehicle type."""
    # Emergency vehicles are exempt from all fines
    if is_emergency == "yes":
        return "No fine (Emergency Vehicle Exemption)"

    # Base fine tier assessment
    if speed <= 60:
        return "No fine"
    elif speed <= 80:
        base_fine = 500
        action = f"Fine = {base_fine}"
    elif speed <= 100:
        base_fine = 2000
        action = f"Fine = {base_fine}"
    else:
        # Extreme speeding results in immediate suspension, regardless of weather modifiers
        return "License suspended"

    # Environmental fine modifier evaluation
    if is_rainy == "yes":
        adjusted_fine = int(base_fine * 1.5)
        return f"{action} + 50% Rainy Weather Surcharge = Total Fine: {adjusted_fine}"
    
    return action


def main() -> None:
    """Collect vehicle telemetry data, sanitize inputs, and print the legal traffic action."""
    try:
        # Prompt for numerical vehicle speed telemetry
        speed_input = input("Enter vehicle speed (km/h): ")
        speed = float(speed_input)

        if speed < 0:
            print("Error: Speed cannot be a negative value.")
            return

        # Prompt for operational exceptions and clean up trailing spaces/casing
        is_emergency = input("Is it an emergency vehicle? (yes/no): ").strip().lower()
        if is_emergency not in ["yes", "no"]:
            print("Error: Please enter exactly 'yes' or 'no' for emergency vehicle status.")
            return

        is_rainy = input("Is the weather rainy? (yes/no): ").strip().lower()
        if is_rainy not in ["yes", "no"]:
            print("Error: Please enter exactly 'yes' or 'no' for weather status.")
            return

        # Process parameters and output the result
        result = calculate_traffic_fine(speed, is_emergency, is_rainy)
        print(result)

    except ValueError:
        print("Error: Invalid numeric input. Please enter a valid number for speed.")


if __name__ == "__main__":
    main()
```

## A company filters candidates:
* Experience ≥ 5 years AND skill score ≥ 80 → shortlisted
* Experience 2–4 years AND skill score ≥ 70 → interview
* Freshers → only if test score ≥ 85
* If criminal record = True → reject immediately
* If multiple job switches (>3 in 2 years) → flag as unstable

```python
def screen_candidate(
    experience: int, 
    skill_score: float, 
    test_score: float, 
    has_criminal_record: str, 
    job_switches: int
) -> str:
    """Evaluate and filter candidate application records based on professional metrics."""
    # Absolute veto condition: Reject immediately if criminal record exists
    if has_criminal_record == "yes":
        return "Rejected Immediately (Criminal Record)"

    # Base application evaluation
    if experience >= 5 and skill_score >= 80:
        status = "Shortlisted"
    elif 2 <= experience <= 4 and skill_score >= 70:
        status = "Interview"
    elif experience == 0 and test_score >= 85:
        status = "Shortlisted (Fresher Special Entry)"
    else:
        status = "Rejected (Does not meet minimum criteria)"

    # Operational modifier condition: Flag instability if job hopping is detected
    if job_switches > 3 and status in ["Shortlisted", "Interview", "Shortlisted (Fresher Special Entry)"]:
        return f"{status} (Flagged as Unstable due to >3 job switches in 2 years)"
        
    return status


def main() -> None:
    """Gather applicant profiles, sanitize metrics, and output the hiring status."""
    # Priority validation: Check safety background first
    has_criminal_record = input("Does the candidate have a criminal record? (yes/no): ").strip().lower()
    if has_criminal_record not in ["yes", "no"]:
        print("Error: Please enter exactly 'yes' or 'no'.")
        return

    try:
        # Collect and parse experience metrics
        exp_input = input("Enter years of experience (0 for freshers): ")
        experience = int(exp_input)
        if experience < 0:
            print("Error: Experience cannot be negative.")
            return

        # Collect performance indicators
        skill_score = float(input("Enter skill score (0–100): "))
        test_score = float(input("Enter test score (0–100): "))
        
        if not (0 <= skill_score <= 100) or not (0 <= test_score <= 100):
            print("Error: Scores must be strictly between 0 and 100.")
            return

        # Collect stability tracking metrics
        job_switches = int(input("Enter number of job switches in the last 2 years: "))
        if job_switches < 0:
            print("Error: Job switches cannot be a negative value.")
            return

        # Process screening pipeline and show result
        result = screen_candidate(experience, skill_score, test_score, has_criminal_record, job_switches)
        print(f"Candidate Status: {result}")

    except ValueError:
        print("Error: Invalid numeric input. Ensure scores and experience are formatted properly.")


if __name__ == "__main__":
    main()
```

## Write a program to decide passenger clearance status.
Take input:
* baggage_weight
* passport_valid (Yes/No)
* visa_status (Valid/Invalid)
* criminal_check (Clear/Flagged)
#### Rules:
* baggage > 30kg → extra charge
* passport invalid → reject
* visa invalid → reject
* criminal flagged → security hold
* otherwise → cleared for boarding

```python
def check_passenger_clearance(
    baggage_weight: float,
    passport_valid: str,
    visa_status: str,
    criminal_check: str
) -> str:
    """Evaluate airport border controls and security rules to determine clearance status."""
    # Priority 1: Check for severe security constraints
    if criminal_check == "flagged":
        return "Security Hold"

    # Priority 2: Check for essential international documentation validity
    if passport_valid == "no":
        return "Rejected (Invalid Passport)"
        
    if visa_status == "invalid":
        return "Rejected (Invalid Visa)"

    # Priority 3: Evaluate operational logistical modifiers if documents are clear
    if baggage_weight > 30:
        return "Cleared for Boarding (Extra Charge for Baggage)"
        
    return "Cleared for Boarding"


def main() -> None:
    """Gather airport border parameters from the user and print passenger status."""
    # Gather operational documentation statuses and sanitize casing/whitespace
    passport_valid = input("Is the passport valid? (yes/no): ").strip().lower()
    if passport_valid not in ["yes", "no"]:
        print("Error: Please enter exactly 'yes' or 'no' for passport validation.")
        return

    visa_status = input("Enter visa status (valid/invalid): ").strip().lower()
    if visa_status not in ["valid", "invalid"]:
        print("Error: Please enter exactly 'valid' or 'invalid'.")
        return

    criminal_check = input("Enter criminal check status (clear/flagged): ").strip().lower()
    if criminal_check not in ["clear", "flagged"]:
        print("Error: Please enter exactly 'clear' or 'flagged'.")
        return

    try:
        # Collect and parse logistical baggage data
        weight_input = input("Enter baggage weight (in kg): ")
        baggage_weight = float(weight_input)

        if baggage_weight < 0:
            print("Error: Baggage weight cannot be negative.")
            return

        # Execute rules workflow evaluation
        status = check_passenger_clearance(baggage_weight, passport_valid, visa_status, criminal_check)
        print(f"Passenger Clearance Status: {status}")

    except ValueError:
        print("Error: Invalid numeric input. Please enter a valid number for weight.")


if __name__ == "__main__":
    main()
```

## Write a program to detect suspicious banking transactions.
Take input:
* transaction_amount
* location_match (Yes/No)
* time_of_transaction(Day/Night)
* previous_transactions_count
* device_verified (Yes/No)
#### Rules:
* amount > 100,000 → suspicious
* location mismatch + high amount → fraud alert
* night transaction + unverified device → high risk
* many transactions (>10 in 1 hour) → freeze account
* otherwise → normal transaction

```python
def analyze_transaction_risk(
    amount: float,
    location_match: str,
    time_of_day: str,
    prev_count: int,
    device_verified: str
) -> str:
    """Evaluate banking transaction telemetry to detect risk anomalies and fraud vectors."""
    # Rule 4: Structural velocity check (High frequency spam prevention takes top priority)
    if prev_count > 10:
        return "Freeze Account (Velocity Alert: >10 transactions within 1 hour)"

    # Rule 2: Compound geological anomaly check
    if location_match == "no" and amount > 50000:  # Contextually matching "high amount" under the 100k threshold
        return "Fraud Alert (Location Mismatch with High Volume Capital)"

    # Rule 1: Absolute volume threshold check
    if amount > 100000:
        return "Suspicious Transaction (Volume Exceeds Standard Threshold)"

    # Rule 3: Environmental security verification check
    if time_of_day == "night" and device_verified == "no":
        return "High Risk (Unverified Device accessing portal during non-standard hours)"

    # Default fallback when all validation flags are completely green
    return "Normal Transaction"


def main() -> None:
    """Collect banking transaction parameters and run anti-fraud heuristics."""
    # Collect string telemetry metrics and normalize casing/whitespace formatting
    location_match = input("Is the transaction location a match? (yes/no): ").strip().lower()
    if location_match not in ["yes", "no"]:
        print("Error: Input must be exactly 'yes' or 'no'.")
        return

    time_of_day = input("Enter time of transaction (day/night): ").strip().lower()
    if time_of_day not in ["day", "night"]:
        print("Error: Input must be exactly 'day' or 'night'.")
        return

    device_verified = input("Is the access device verified? (yes/no): ").strip().lower()
    if device_verified not in ["yes", "no"]:
        print("Error: Input must be exactly 'yes' or 'no'.")
        return

    try:
        # Collect and parse quantitative transaction fields
        amount_input = input("Enter transaction amount: ")
        amount = float(amount_input)
        if amount <= 0:
            print("Error: Transaction amount must be greater than 0.")
            return

        count_input = input("Enter count of transactions in the last hour: ")
        prev_count = int(count_input)
        if prev_count < 0:
            print("Error: Transaction count cannot be a negative value.")
            return

        # Execute security analytics pipeline
        alert_status = analyze_transaction_risk(
            amount, location_match, time_of_day, prev_count, device_verified
        )
        print(f"Risk Assessment Engine Result: {alert_status}")

    except ValueError:
        print("Error: Invalid numeric input. Ensure amount is currency-scaled and count is a whole number.")


if __name__ == "__main__":
    main()
```

## Write a program to classify a person’s risk level in a police database.
Take input:
* criminal_record (Yes/No)
* number_of_cases
* travel_history_flagged (Yes/No)
* financial_activity_score
* last_activity_days
#### Rules:
* criminal_record = Yes AND cases > 3 → High Risk
* flagged travel + suspicious financial score → High Risk
* no activity for long time (>365 days) → Monitor
* clean record + normal activity → Low Risk
* mixed signals → Medium Risk

```python
def classify_police_risk_level(
    criminal_record: str,
    number_of_cases: int,
    travel_history_flagged: str,
    financial_activity_score: float,
    last_activity_days: int
) -> str:
    """Classify a person's risk profile within the law enforcement threat metrics database."""
    # Context definition: A financial activity score above 75 is treated as structurally suspicious
    is_suspicious_finance = financial_activity_score > 75

    # Rule 1: High Risk Case - Known repeat offender
    if criminal_record == "yes" and number_of_cases > 3:
        return "High Risk (Repeat Offender Profile)"

    # Rule 2: High Risk Case - International movement and financial tracking anomalies
    if travel_history_flagged == "yes" and is_suspicious_finance:
        return "High Risk (Cross-Border Tracking / Financial Anomaly)"

    # Rule 3: Monitor Status - Dormant profiles that require periodic tracking
    if last_activity_days > 365:
        return "Monitor (Dormant profile exceeds 365 days)"

    # Rule 4: Low Risk Case - Fully verified baseline profiles
    if criminal_record == "no" and number_of_cases == 0 and travel_history_flagged == "no" and not is_suspicious_finance:
        return "Low Risk"

    # Rule 5: Fallback state for overlapping or unclassified patterns
    return "Medium Risk (Mixed Signals Evaluated)"


def main() -> None:
    """Collect law enforcement database metrics, sanitize data integrity constraints, and print threat class."""
    # Prompt for profile categorical statuses and clean up whitespace variations
    criminal_record = input("Does the subject have a criminal record? (yes/no): ").strip().lower()
    if criminal_record not in ["yes", "no"]:
        print("Error: Input must be exactly 'yes' or 'no'.")
        return

    travel_history_flagged = input("Is the travel history flagged as suspicious? (yes/no): ").strip().lower()
    if travel_history_flagged not in ["yes", "no"]:
        print("Error: Input must be exactly 'yes' or 'no'.")
        return

    try:
        # Prompt for numerical indicators and whole metrics
        cases_input = input("Enter total number of cases: ")
        number_of_cases = int(cases_input)

        if number_of_cases < 0:
            print("Error: Number of cases cannot be a negative value.")
            return
            
        if criminal_record == "no" and number_of_cases > 0:
            print("Logical Anomaly: A subject cannot have active cases if they have a clean record.")
            return

        financial_input = input("Enter financial activity score (0–100): ")
        financial_activity_score = float(financial_input)

        if not (0 <= financial_activity_score <= 100):
            print("Error: Financial tracking scores must fall within the 0 to 100 spectrum.")
            return

        activity_input = input("Enter days since last activity recorded: ")
        last_activity_days = int(activity_input)

        if last_activity_days < 0:
            print("Error: Last activity timeline cannot be negative.")
            return

        # Process the evaluation metrics tree and output results
        risk_profile = classify_police_risk_level(
            criminal_record, number_of_cases, travel_history_flagged, financial_activity_score, last_activity_days
        )
        print(f"Database Classification Result: {risk_profile}")

    except ValueError:
        print("Error: Invalid numeric formatting. Ensure cases/days are whole integers and scores are numbers.")


if __name__ == "__main__":
    main()
```
