# 7) Compound Interest:
## Write a Python program to calculate compound interest. Use the formula:
### A = P * (1 + r/n)^(n-t)
### Where:
#### A= total amount
#### P = principal amount
#### r = annual interest rate (decimal)
#### n = number of times interest is compounded per year
#### t = time in years
#### Take P, r, n, and t as inputs from the user.

```python
# 1. Get user inputs
p = float(input("Enter principal amount (P): "))
r = float(input("Enter annual interest rate in % (r): "))
n = int(input("Enter number of compounding periods per year (n): "))
t = float(input("Enter time in years (t): "))

# 2. Convert percentage rate to a decimal
r_decimal = r / 100

# 3. Calculate total amount (A) using the formula
# Note: The standard formula uses multiplication for the exponent: n * t
amount = p * (1 + r_decimal / n) ** (n * t)

# 4. Print the result
print("Total Amount (A):", round(amount, 2))
```

# 11) Quadratic Equation Roots:
## Write a Python program to find the roots of a quadratic equation. Use the formula:
### x = (-b ± sqrt(b^2 - 4ac)) / (2a)
### Take a, b, and c as inputs from the user.

```python
import cmath

# 1. Get user inputs for coefficients
a = float(input("Enter coefficient a: "))
b = float(input("Enter coefficient b: "))
c = float(input("Enter coefficient c: "))

# 2. Calculate the discriminant (b^2 - 4ac)
discriminant = (b**2) - (4 * a * c)

# 3. Apply the quadratic formula
root1 = (-b - cmath.sqrt(discriminant)) / (2 * a)
root2 = (-b + cmath.sqrt(discriminant)) / (2 * a)

# 4. Print the results
print("Root 1:", root1)
print("Root 2:", root2)
```

# 19) Work Done by a Force:
## Write a Python program to calculate the work done by a force. Use the formula:
### W = f * d * cos(theta)
### where:
#### W = work done
#### f= force
#### d = distance
#### theta = angle between force and direction of movement (in degrees)
### Take f, d, and θ as inputs from the user.

```python
import math

# 1. Get user inputs
f = float(input("Enter force (in Newtons): "))
d = float(input("Enter distance (in meters): "))
theta = float(input("Enter angle in degrees (theta): "))

# 2. Convert degrees to radians (Python's cos function requires radians)
theta_radians = math.radians(theta)

# 3. Calculate work done
w = f * d * math.cos(theta_radians)

# 4. Print the result
print("Work Done:", round(w, 2), "J")
```

# 20) Heat Transfer:
## Write a Python program to calculate the amount of heat transferred. Use the formula:
### Q = m * c * deltaT
### where:
#### Q= heat transfer
#### m = mass
#### c = specific heat capacity
#### ΔT
### Take m, c, T as inputs from the user.

```python
# 1. Get user inputs
m = float(input("Enter mass (m) in kg: "))
c = float(input("Enter specific heat capacity (c): "))
delta_t = float(input("Enter temperature change (delta T) in °C: "))

# 2. Calculate heat transfer (Q)
q = m * c * delta_t

# 3. Print the result
print("Heat Transferred (Q):", round(q, 2), "Joules")
```
