## Start with an empty to-do list:

## Create a guest list:

    guests = ["Omar", "Ayesha", "Bilal", "Hina", "Usman"]

    Remove "Bilal" using remove()
    
    Remove the last guest using pop() and store it in a variable called removed_guest
    
    Remove the guest at index 1 using pop()
    
    Clear the entire list using clear()

```python
# Initial guest list
guests = ["Omar", "Ayesha", "Bilal", "Hina", "Usman"]

# Remove "Bilal" using remove()
guests.remove("Bilal")

# Remove the last guest using pop() and store it in removed_guest
removed_guest = guests.pop()

# Print the remaining list and the removed guest
print("Updated list:", guests)
print("Removed guest:", removed_guest)

# Remove the guest at index 1 ("Ayesha") using pop()
popped_guest = guests.pop(1)

# Print the remaining list and the popped guest
print("Updated list:", guests)
print("Popped guest at index 1:", popped_guest)

# Clear the entire list using clear()
guests.clear()

# Print the final list to confirm it is empty
print("Final guest list:", guests)
```

## Create a dictionary that manages records of two patients.

Each patient should have:

    Patient ID
    Name
    Age
    Disease
    Doctor Assigned
    Medicines (a nested dictionary of medicine names and their doses)

### Example:

    Patient 1: ID 501, Name Bilal, Age 45, Disease Diabetes, Doctor Dr. Ayesha
    Medicines: Metformin 500mg, Insulin 10 units
    Patient 2: ID 502, Name Hina, Age 32, Disease Flu, Doctor Dr. Usman
    Medicines: Paracetamol 500mg, Cough Syrup 10ml

Print the complete patient records.

Print only the medicines of Patient 1.

Add a new medicine "Vitamin D" with dose "1000 IU" for Patient 2.

## Create a nested dictionary for a company that stores three employees.

Each employee should have:

    Employee ID
    Name
    Department
    Salary
    Skills (a list of skills inside the dictionary)

Employees:

    ID 201: Name Ali, Department IT, Salary 75000, Skills: Python, SQL, HTML
    ID 202: Name Fatima, Department HR, Salary 60000, Skills: Communication, Recruitment
    ID 203: Name Zain, Department Finance, Salary 80000, Skills: Excel, Accounting

### Tasks:

    Print the complete employee database.
    Print only the skills of employee 201.
    Add a new skill "JavaScript" for Ali.
    Increase Fatima’s salary to 65000.
    Add a new employee: ID 204, Name Hina, Department Marketing, Salary 55000, Skills: SEO, Content Writing.
    Remove employee 203.
    Print the final database.

```python
# Create a nested dictionary for a company storing three employees
company_db = {
    201: {
        "Name": "Ali",
        "Department": "IT",
        "Salary": 75000,
        "Skills": ["Python", "SQL", "HTML"]
    },
    202: {
        "Name": "Fatima",
        "Department": "HR",
        "Salary": 60000,
        "Skills": ["Communication", "Recruitment"]
    },
    203: {
        "Name": "Zain",
        "Department": "Finance",
        "Salary": 80000,
        "Skills": ["Excel", "Accounting"]
    }
}

# 1. Print the complete employee database
print("--- Complete Employee Database ---")
print(company_db)
print()

# 2. Print only the skills of employee 201
print("--- Skills of Employee 201 ---")
print(company_db[201]["Skills"])
print()

# 3. Add a new skill "JavaScript" for Ali
company_db[201]["Skills"].append("JavaScript")

# 4. Increase Fatima’s salary to 65000
company_db[202]["Salary"] = 65000

# 5. Add a new employee: ID 204
company_db[204] = {
    "Name": "Hina",
    "Department": "Marketing",
    "Salary": 55000,
    "Skills": ["SEO", "Content Writing"]
}

# 6. Remove employee 203
del company_db[203]

# 7. Print the final database
print("--- Final Updated Database ---")
print(company_db)
```

## Common Subjects Between Two Student
Create a set :

    Ali’s subjects: Math, Science, English, Computer, Urdu

    Sara’s subjects: Science, English, Islamiat, Computer, Arts
## Find and print:

    Subjects that both students study (common subjects)
    All subjects taken by either Ali or Sara (combined unique subjects)
    Subjects that only Ali studies
    Subjects that only Sara studies

```python
# Create sets for both students
ali_subjects = {"Math", "Science", "English", "Computer", "Urdu"}
sara_subjects = {"Science", "English", "Islamiat", "Computer", "Arts"}

# 1. Subjects that both students study (Intersection)
common_subjects = ali_subjects.intersection(sara_subjects)
print("Common subjects both study:", common_subjects)

# 2. All subjects taken by either Ali or Sara (Union)
combined_subjects = ali_subjects.union(sara_subjects)
print("All combined unique subjects:", combined_subjects)

# 3. Subjects that only Ali studies (Difference)
only_ali = ali_subjects.difference(sara_subjects)
print("Subjects that only Ali studies:", only_ali)

# 4. Subjects that only Sara studies (Difference)
only_sara = sara_subjects.difference(ali_subjects)
print("Subjects that only Sara studies:", only_sara)
```
