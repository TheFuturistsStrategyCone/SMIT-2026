## A temperature monitoring system receives values as text. Convert the input into a number and determine if the temperature is above normal (greater than 37°C).

```python
# Take the temperature input as text
temperature_text = input("Enter the current temperature (°C): ")

# Convert the text into a float (numeric type to handle decimal values)
temperature = float(temperature_text)

# Determine if the temperature is above normal
if temperature > 37.0:
    print(f"Alert: Temperature ({temperature}°C) is above normal!")
else:
    print(f"Normal: Temperature ({temperature}°C) is within the safe range.")
```

## A store offers a 10% discount on a product that costs 1500. Write a Python expression to calculate the discounted price using the appropriate arithmetic operator. Ask the user to enter the product price and discount rate.

```python
# Ask the user for the product price and discount rate
price = float(input("Enter the product price: "))
discount_rate = float(input("Enter the discount rate (in %): "))

# Python expression using arithmetic operators to calculate the discounted price
discounted_price = price - (price * (discount_rate / 100))

# Display the final price
print(f"The discounted price is: {discounted_price:.2f}")
```

## You have some apples. Ask the user to enter the total number of apples they have and the number of apples they give to a friend. Write a Python program to calculate how many apples are left.
### Which operator will you use to find the remaining apples?

```python
# Ask the user for the total number of apples and apples given away
total_apples = int(input("Enter the total number of apples you have: "))
apples_given = int(input("Enter the number of apples you give to your friend: "))

# Calculate how many apples are left using the subtraction operator (-)
apples_left = total_apples - apples_given

# Print the final result
print(f"You have {apples_left} apple(s) left.")
```

## Ask the user to enter the ages of two friends. Write a Python program to check who is older.
### Which comparison operator will you use to find out who is older?

```python
# Ask the user to enter the ages of two friends
friend1_age = int(input("Enter the age of the first friend: "))
friend2_age = int(input("Enter the age of the second friend: "))

# Check who is older using comparison operators
if friend1_age > friend2_age:
    print("The first friend is older.")
elif friend2_age > friend1_age:
    print("The second friend is older.")
else:
    print("Both friends are the exact same age.")
```
