## Write a program to calculate how much each person should pay for a restaurant bill. Handle invalid input and the situation where the number of people is zero.

```python
def get_bill_amount(prompt: str = "Enter the total restaurant bill: ") -> float:
    """Safely prompts for a valid, non-negative bill amount."""
    while True:
        raw_input = input(prompt).strip()
        try:
            amount = float(raw_input)
            if amount < 0:
                print("Bill amount cannot be negative. Please enter a valid amount.")
                continue
            return amount
        except ValueError:
            print(f"Invalid input: '{raw_input}' is not a valid number. Please try again.")


def get_people_count(prompt: str = "Enter the number of people splitting the bill: ") -> int:
    """Safely prompts for a positive integer count of people."""
    while True:
        raw_input = input(prompt).strip()
        try:
            count = int(raw_input)
            if count == 0:
                raise ZeroDivisionError("Number of people cannot be zero.")
            if count < 0:
                print("Number of people cannot be negative. Please enter a positive whole number.")
                continue
            return count
        except ValueError:
            print(f"Invalid input: '{raw_input}' is not a whole number. Please try again.")
        except ZeroDivisionError as err:
            print(f"Error: {err} At least 1 person is required to split the bill.")


def split_restaurant_bill() -> None:
    """Calculates and displays the exact share per person."""
    total_bill = get_bill_amount()
    num_people = get_people_count()

    try:
        share_per_person = total_bill / num_people
        print(f"\nTotal Bill: {total_bill:,.2f}")
        print(f"Split between {num_people} {'person' if num_people == 1 else 'people'}: {share_per_person:,.2f} each")
    except ZeroDivisionError:
        print("Calculation failed: Cannot divide the bill by zero people.")


if __name__ == "__main__":
    split_restaurant_bill()
```

## Write a program to simulate an ATM withdrawal. Use `finally` to display a message that the ATM transaction process has ended whether the transaction succeeds or fails.

```python
class InsufficientFundsError(Exception):
    """Raised when the requested withdrawal exceeds the available balance."""


def atm_withdrawal(account_balance: float) -> float:
    """Processes an ATM withdrawal and guarantees a session-end message via `finally`."""
    print(f"Current Balance: PKR {account_balance:,.2f}")
    raw_input = input("Enter amount to withdraw (in multiples of 500): ").strip()

    try:
        amount = float(raw_input)
        if amount <= 0:
            raise ValueError("Withdrawal amount must be greater than zero.")
        if amount % 500 != 0:
            raise ValueError("ATM only dispenses notes in multiples of PKR 500.")
        if amount > account_balance:
            raise InsufficientFundsError(
                f"Requested PKR {amount:,.2f}, but available balance is PKR {account_balance:,.2f}."
            )
    except ValueError as err:
        print(f"Transaction Failed (Invalid Input): {err}")
    except InsufficientFundsError as err:
        print(f"Transaction Failed (Insufficient Funds): {err}")
    else:
        account_balance -= amount
        print("\nTransaction Approved! Please collect your cash.")
        print(f"Withdrawn Amount : PKR {amount:,.2f}")
        print(f"Remaining Balance: PKR {account_balance:,.2f}")
    finally:
        print("\n[ATM Notice] The ATM transaction process has ended. Thank you for banking with us.")

    return account_balance


if __name__ == "__main__":
    starting_balance = 25_000.00
    atm_withdrawal(starting_balance)
```

## Write a program to calculate a shopping cart total using product names, prices, and quantities. Handle invalid product names, invalid quantities, and invalid numeric input using appropriate exceptions.

```python
from dataclasses import dataclass


class InvalidProductError(ValueError):
    """Raised when a product name is empty, numeric-only, or malformed."""


class InvalidQuantityError(ValueError):
    """Raised when a quantity is not a positive whole integer."""


@dataclass(frozen=True)
class CartItem:
    """Represents a validated item in the shopping cart."""
    name: str
    unit_price: float
    quantity: int

    @property
    def subtotal(self) -> float:
        return self.unit_price * self.quantity


def validate_product_name(raw_name: str) -> str:
    """Ensures the product name is non-empty and contains alphabetic characters."""
    name = raw_name.strip()
    if not name:
        raise InvalidProductError("Product name cannot be blank.")
    if not any(char.isalpha() for char in name):
        raise InvalidProductError(f"'{name}' is not a valid product name (must include letters).")
    return name


def validate_price(raw_price: str) -> float:
    """Ensures the price is a valid non-negative floating-point number."""
    cleaned = raw_price.strip()
    try:
        price = float(cleaned)
    except ValueError as err:
        raise ValueError(f"'{cleaned}' is not a valid numeric price.") from err

    if price < 0:
        raise ValueError(f"Price cannot be negative (received {price:,.2f}).")
    return price


def validate_quantity(raw_qty: str) -> int:
    """Ensures the quantity is a positive whole integer."""
    cleaned = raw_qty.strip()
    try:
        qty = int(cleaned)
    except ValueError as err:
        raise InvalidQuantityError(f"'{cleaned}' is not a valid whole number for quantity.") from err

    if qty <= 0:
        raise InvalidQuantityError(f"Quantity must be at least 1 (received {qty}).")
    return qty


def build_shopping_cart() -> list[CartItem]:
    """Interactively collects and validates shopping cart items."""
    cart: list[CartItem] = []
    print("Enter shopping cart items (type 'done' as the product name to finish):")

    while True:
        raw_name = input("\nProduct name: ")
        if raw_name.strip().lower() == "done":
            break

        try:
            name = validate_product_name(raw_name)
            price = validate_price(input("Unit price (PKR): "))
            quantity = validate_quantity(input("Quantity: "))
        except (InvalidProductError, InvalidQuantityError, ValueError) as err:
            print(f"Item Rejected — {err} Please re-enter this item.")
        else:
            item = CartItem(name=name, unit_price=price, quantity=quantity)
            cart.append(item)
            print(f"Added: {item.quantity}x {item.name} @ PKR {item.unit_price:,.2f} = PKR {item.subtotal:,.2f}")

    return cart


def display_receipt(cart: list[CartItem]) -> None:
    """Prints an itemized receipt and grand total."""
    if not cart:
        print("\nYour shopping cart is empty. Total: PKR 0.00")
        return

    print("\n" + "=" * 54)
    print(f"{'Product':<22} {'Qty':>5} {'Unit Price':>12} {'Subtotal':>12}")
    print("-" * 54)
    for item in cart:
        print(f"{item.name[:21]:<22} {item.quantity:>5} {item.unit_price:>12,.2f} {item.subtotal:>12,.2f}")
    print("=" * 54)

    grand_total = sum(item.subtotal for item in cart)
    print(f"{'GRAND TOTAL (PKR)':<41} {grand_total:>12,.2f}")


if __name__ == "__main__":
    shopping_cart = build_shopping_cart()
    display_receipt(shopping_cart)
```

## Write a program to create a student result system. Store multiple students and their marks. Handle invalid marks and missing student information while continuing to process valid students.

```python
from dataclasses import dataclass
from typing import Any


class StudentDataError(Exception):
    """Custom exception for invalid or missing student information."""


@dataclass
class StudentResult:
    student_id: str
    name: str
    marks: dict[str, float]
    total: float
    average: float
    grade: str


def calculate_grade(average: float) -> str:
    """Determine letter grade based on average percentage."""
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    return "F"


def validate_and_process_student(record: dict[str, Any]) -> StudentResult:
    """Validate a single student record and compute their final result."""
    # 1. Check for missing or empty core fields
    student_id = str(record.get("id", "")).strip()
    name = str(record.get("name", "")).strip()
    marks = record.get("marks")

    if not student_id or not name:
        raise StudentDataError("Missing student ID or name.")

    if not isinstance(marks, dict) or not marks:
        raise StudentDataError(f"Missing or invalid marks dictionary for {name} ({student_id}).")

    # 2. Validate individual subject marks
    validated_marks: dict[str, float] = {}
    for subject, score in marks.items():
        if score is None or isinstance(score, bool):
            raise StudentDataError(f"Missing score in {subject} for {name} ({student_id}).")
        
        try:
            numeric_score = float(score)
        except (ValueError, TypeError):
            raise StudentDataError(f"Non-numeric score '{score}' in {subject} for {name} ({student_id}).")

        if not (0 <= numeric_score <= 100):
            raise StudentDataError(f"Out-of-range score ({numeric_score}) in {subject} for {name} ({student_id}).")

        validated_marks[subject] = numeric_score

    # 3. Compute totals and grade
    total = sum(validated_marks.values())
    average = round(total / len(validated_marks), 2)
    grade = calculate_grade(average)

    return StudentResult(
        student_id=student_id,
        name=name,
        marks=validated_marks,
        total=total,
        average=average,
        grade=grade,
    )


def generate_class_results(student_records: list[dict[str, Any]]) -> tuple[list[StudentResult], list[str]]:
    """Process a batch of students, separating valid results from error logs."""
    valid_results: list[StudentResult] = []
    error_logs: list[str] = []

    for index, record in enumerate(student_records, start=1):
        try:
            result = validate_and_process_student(record)
            valid_results.append(result)
        except StudentDataError as err:
            error_logs.append(f"Record #{index} skipped -> {err}")

    return valid_results, error_logs


# --- Example Usage ---
if __name__ == "__main__":
    raw_student_data = [
        {"id": "S101", "name": "Ayesha Khan", "marks": {"Math": 92, "Science": 88, "English": 95}},
        {"id": "S102", "name": "", "marks": {"Math": 75, "Science": 80}},  # Missing name
        {"id": "S103", "name": "Bilal Ahmed", "marks": {"Math": 105, "Science": 70}},  # Invalid mark (>100)
        {"id": "S104", "name": "Zainab Raza", "marks": {"Math": "N/A", "Science": 82}},  # Non-numeric mark
        {"id": "S105", "name": "Omar Farooq", "marks": {}},  # Missing marks
        {"id": "S106", "name": "Hina Tariq", "marks": {"Math": 64, "Science": 71, "English": 68}},
    ]

    results, errors = generate_class_results(raw_student_data)

    print("=== VALID STUDENT RESULTS ===")
    for res in results:
        print(f"{res.student_id} | {res.name:<15} | Total: {res.total:>5.1f} | Avg: {res.average:>5.2f}% | Grade: {res.grade}")

    print("\n=== SKIPPED RECORDS (ERRORS) ===")
    for error in errors:
        print(error)
```

## Write a program to create a student result system. Store multiple students and their marks. Handle invalid marks and missing student information while continuing to process valid students.

```python
import math
from dataclasses import dataclass
from typing import Any


@dataclass(frozen=True)
class PayrollEntry:
    """Represents a successfully processed employee salary calculation."""
    emp_id: str
    name: str
    base_salary: float
    allowance: float
    net_salary: float


def parse_positive_amount(value: Any, field_name: str) -> float:
    """Converts a raw field to a valid, non-negative finite float."""
    if value is None or isinstance(value, bool):
        raise ValueError(f"'{field_name}' is missing or null.")

    if isinstance(value, str) and not value.strip():
        raise ValueError(f"'{field_name}' is blank.")

    try:
        amount = float(value)
    except (TypeError, ValueError) as err:
        raise ValueError(f"'{field_name}' has invalid numeric value: {value!r}.") from err

    if math.isnan(amount) or math.isinf(amount):
        raise ValueError(f"'{field_name}' cannot be NaN or infinity.")
    if amount < 0:
        raise ValueError(f"'{field_name}' cannot be negative ({amount:,.2f}).")

    return amount


def process_payroll(employees: list[dict[str, Any]]) -> tuple[list[PayrollEntry], list[str]]:
    """Processes all employee records, isolating bad entries so payroll finishes completely."""
    processed: list[PayrollEntry] = []
    errors: list[str] = []

    for idx, record in enumerate(employees, start=1):
        emp_id = str(record.get("id", f"ROW-{idx}"))
        name = str(record.get("name", "Unknown Employee")).strip() or "Unknown Employee"

        try:
            if "base_salary" not in record:
                raise KeyError("Missing required 'base_salary' field.")

            base_salary = parse_positive_amount(record["base_salary"], "base_salary")
            allowance = parse_positive_amount(record.get("allowance", 0.0), "allowance")

            if base_salary == 0:
                raise ValueError("'base_salary' must be greater than zero.")

            net_salary = base_salary + allowance
            processed.append(
                PayrollEntry(
                    emp_id=emp_id,
                    name=name,
                    base_salary=base_salary,
                    allowance=allowance,
                    net_salary=net_salary,
                )
            )
        except (KeyError, ValueError, TypeError) as err:
            errors.append(f"[{emp_id} - {name}] Skipped: {err}")

    return processed, errors


def print_payroll_report(processed: list[PayrollEntry], errors: list[str]) -> None:
    """Displays processed salaries, total payout, and an audit log of skipped records."""
    print("=" * 68)
    print(f"{'ID':<10} {'Employee Name':<20} {'Base':>11} {'Allowance':>11} {'Net Pay':>12}")
    print("-" * 68)

    for entry in processed:
        print(
            f"{entry.emp_id:<10} {entry.name[:19]:<20} "
            f"{entry.base_salary:>11,.2f} {entry.allowance:>11,.2f} {entry.net_salary:>12,.2f}"
        )

    print("=" * 68)
    total_payout = sum(entry.net_salary for entry in processed)
    print(f"Processed Employees : {len(processed)}")
    print(f"Total Payroll Payout: PKR {total_payout:,.2f}")

    if errors:
        print("\n--- Payroll Exceptions (Requires HR Review) ---")
        for issue in errors:
            print(f"  ! {issue}")


if __name__ == "__main__":
    employee_records: list[dict[str, Any]] = [
        {"id": "EMP-01", "name": "Sara Tariq", "base_salary": 120_000, "allowance": 15_000},
        {"id": "EMP-02", "name": "Usman Ghani"},  # Missing salary key
        {"id": "EMP-03", "name": "Hina Altaf", "base_salary": "95,000", "allowance": 10_000},  # Invalid string with comma
        {"id": "EMP-04", "name": "Kamran Akmal", "base_salary": -45_000, "allowance": 5_000},  # Negative salary
        {"id": "EMP-05", "name": "Nadia Jamil", "base_salary": None, "allowance": 8_000},  # Null salary
        {"id": "EMP-06", "name": "Tariq Mahmood", "base_salary": 150_000, "allowance": 20_000},
    ]

    valid_entries, skipped_logs = process_payroll(employee_records)
    print_payroll_report(valid_entries, skipped_logs)
```

## Write a program to create a restaurant billing system using a menu dictionary. Handle unavailable menu items, invalid quantities, and invalid numeric input.

```python
from dataclasses import dataclass


class MenuItemNotFoundError(KeyError):
    """Raised when a customer orders an item that is not on the menu."""


class InvalidQuantityError(ValueError):
    """Raised when an ordered quantity is non-numeric, zero, or negative."""


@dataclass(frozen=True)
class OrderLine:
    """Represents a validated line item on a restaurant bill."""
    item_name: str
    unit_price: float
    quantity: int

    @property
    def subtotal(self) -> float:
        return self.unit_price * self.quantity


def display_menu(menu: dict[str, float]) -> None:
    """Prints the restaurant menu neatly formatted."""
    print("\n" + "=" * 42)
    print(f"{'RESTAURANT MENU':^42}")
    print("=" * 42)
    for item, price in menu.items():
        print(f"  {item:<26} PKR {price:>8,.2f}")
    print("=" * 42)


def find_menu_item(raw_name: str, menu: dict[str, float]) -> tuple[str, float]:
    """Matches user input case-insensitively against the menu dictionary."""
    cleaned = raw_name.strip()
    if not cleaned:
        raise MenuItemNotFoundError("Item name cannot be blank.")

    lookup = {key.lower(): (key, price) for key, price in menu.items()}
    try:
        return lookup[cleaned.lower()]
    except KeyError as err:
        raise MenuItemNotFoundError(f"'{cleaned}' is not available on our menu.") from err


def parse_quantity(raw_qty: str) -> int:
    """Validates that the ordered quantity is a positive whole number."""
    cleaned = raw_qty.strip()
    try:
        qty = int(cleaned)
    except ValueError as err:
        raise InvalidQuantityError(f"'{cleaned}' is not a valid whole number.") from err

    if qty <= 0:
        raise InvalidQuantityError(f"Quantity must be at least 1 (received {qty}).")
    return qty


def take_order(menu: dict[str, float]) -> dict[str, OrderLine]:
    """Interactively collects customer orders and merges duplicate items."""
    order: dict[str, OrderLine] = {}
    print("\nEnter items to order (type 'done' when finished):")

    while True:
        raw_item = input("\nMenu item: ")
        if raw_item.strip().lower() == "done":
            break

        try:
            canonical_name, unit_price = find_menu_item(raw_item, menu)
            qty = parse_quantity(input(f"Quantity for {canonical_name}: "))
        except (MenuItemNotFoundError, InvalidQuantityError) as err:
            print(f"Order Error — {err} Please try again.")
        else:
            existing_qty = order[canonical_name].quantity if canonical_name in order else 0
            updated_line = OrderLine(
                item_name=canonical_name,
                unit_price=unit_price,
                quantity=existing_qty + qty,
            )
            order[canonical_name] = updated_line
            print(f"Added: {qty}x {canonical_name} (Total in order: {updated_line.quantity})")

    return order


def print_bill(order: dict[str, OrderLine], tax_rate: float = 0.13) -> None:
    """Calculates subtotal, GST, and grand total, then prints the final bill."""
    if not order:
        print("\nNo items ordered. Bill cancelled.")
        return

    subtotal = sum(line.subtotal for line in order.values())
    tax_amount = subtotal * tax_rate
    grand_total = subtotal + tax_amount

    print("\n" + "=" * 56)
    print(f"{'FINAL RESTAURANT BILL':^56}")
    print("=" * 56)
    print(f"{'Item':<24} {'Qty':>5} {'Price':>11} {'Total':>12}")
    print("-" * 56)
    for line in order.values():
        print(
            f"{line.item_name[:23]:<24} {line.quantity:>5} "
            f"{line.unit_price:>11,.2f} {line.subtotal:>12,.2f}"
        )
    print("-" * 56)
    print(f"{'Subtotal':<42} PKR {subtotal:>9,.2f}")
    print(f"{f'GST ({tax_rate * 100:.0f}%)':<42} PKR {tax_amount:>9,.2f}")
    print("=" * 56)
    print(f"{'GRAND TOTAL':<42} PKR {grand_total:>9,.2f}")
    print("=" * 56)


if __name__ == "__main__":
    restaurant_menu: dict[str, float] = {
        "Chicken Biryani": 550.00,
        "Beef Nihari": 850.00,
        "Chicken Karahi": 1_450.00,
        "Garlic Naan": 90.00,
        "Mint Raita": 120.00,
        "Kashmiri Chai": 220.00,
    }

    display_menu(restaurant_menu)
    customer_order = take_order(restaurant_menu)
    print_bill(customer_order)
```

## Write a program to read sales records from a text file and calculate total sales. Handle a missing file and invalid records separately.

```python
from pathlib import Path


class InvalidSalesRecordError(ValueError):
    """Raised when a line in the sales file is malformed or contains an invalid amount."""


def parse_sales_line(line: str, line_number: int) -> tuple[str, float]:
    """Parses a single 'item,amount' record and validates its fields."""
    parts = [part.strip() for part in line.split(",")]
    if len(parts) != 2 or not parts[0] or not parts[1]:
        raise InvalidSalesRecordError(
            f"Line {line_number}: Expected 'Item,Amount' format, got {line.strip()!r}."
        )

    item_name, raw_amount = parts
    try:
        amount = float(raw_amount)
    except ValueError as err:
        raise InvalidSalesRecordError(
            f"Line {line_number}: '{raw_amount}' is not a valid numeric amount."
        ) from err

    if amount < 0:
        raise InvalidSalesRecordError(
            f"Line {line_number}: Sale amount cannot be negative ({amount:,.2f})."
        )

    return item_name, amount


def calculate_total_sales(file_path: str) -> None:
    """Reads sales records from a text file, separating file errors from line-level data errors."""
    path = Path(file_path.strip())
    total_sales = 0.0
    valid_count = 0
    invalid_logs: list[str] = []

    try:
        with path.open(mode="r", encoding="utf-8") as file:
            for line_number, raw_line in enumerate(file, start=1):
                if not raw_line.strip() or raw_line.lstrip().startswith("#"):
                    continue

                try:
                    _, amount = parse_sales_line(raw_line, line_number)
                except InvalidSalesRecordError as err:
                    invalid_logs.append(str(err))
                else:
                    total_sales += amount
                    valid_count += 1

    except FileNotFoundError:
        print(f"File Error: '{path}' does not exist. Please check the file path and try again.")
        return
    except PermissionError:
        print(f"File Error: Permission denied while reading '{path}'.")
        return
    except OSError as err:
        print(f"File Error: Unable to read '{path}' ({err}).")
        return

    print("\n--- Sales Summary Report ---")
    print(f"File Processed : {path.name}")
    print(f"Valid Records  : {valid_count}")
    print(f"Total Sales    : PKR {total_sales:,.2f}")

    if invalid_logs:
        print("\n--- Skipped Invalid Records ---")
        for issue in invalid_logs:
            print(f"  ! {issue}")


requested_path = input("Enter sales file path (e.g., sales.txt): ")
calculate_total_sales(requested_path)
```

## Write a program using functions to validate a customer's registration information. Handle invalid age, invalid email information, and missing required data with appropriate exceptions.

```python
import re
from dataclasses import dataclass


class RegistrationError(ValueError):
    """Base exception for all customer registration validation failures."""


class MissingFieldError(RegistrationError):
    """Raised when a required registration field is empty or missing."""


class InvalidEmailError(RegistrationError):
    """Raised when an email address does not match a valid format."""


class InvalidAgeError(RegistrationError):
    """Raised when age is non-numeric or outside the permitted registration range."""


@dataclass(frozen=True)
class CustomerProfile:
    """Represents a validated customer registration record."""
    full_name: str
    email: str
    age: int
    city: str


def validate_required_text(value: str | None, field_name: str) -> str:
    """Ensures a required text field is present and not just whitespace."""
    if value is None or not str(value).strip():
        raise MissingFieldError(f"'{field_name}' is required and cannot be blank.")
    return str(value).strip()


def validate_email(raw_email: str | None) -> str:
    """Validates that the email is present and follows standard user@domain.tld syntax."""
    email = validate_required_text(raw_email, "Email").lower()
    email_pattern = r"^[a-z0-9._%+-]+@[a-z0-9.-]+\.[a-z]{2,}$"

    if not re.match(email_pattern, email):
        raise InvalidEmailError(f"'{email}' is not a valid email address (expected format: name@example.com).")
    return email


def validate_age(raw_age: str | int | None, min_age: int = 18, max_age: int = 120) -> int:
    """Validates that age is provided, is a whole integer, and falls within allowed bounds."""
    if raw_age is None or (isinstance(raw_age, str) and not raw_age.strip()):
        raise MissingFieldError("'Age' is required and cannot be blank.")

    if isinstance(raw_age, bool):
        raise InvalidAgeError("Age must be a valid integer, not a boolean.")

    try:
        age = int(str(raw_age).strip())
    except ValueError as err:
        raise InvalidAgeError(f"'{raw_age}' is not a valid whole number for age.") from err

    if not (min_age <= age <= max_age):
        raise InvalidAgeError(f"Age must be between {min_age} and {max_age} (received {age}).")
    return age


def register_customer(
    full_name: str | None,
    email: str | None,
    age: str | int | None,
    city: str | None,
) -> CustomerProfile:
    """Runs all field validators and returns a verified CustomerProfile."""
    clean_name = validate_required_text(full_name, "Full Name")
    clean_email = validate_email(email)
    clean_age = validate_age(age)
    clean_city = validate_required_text(city, "City")

    return CustomerProfile(
        full_name=clean_name,
        email=clean_email,
        age=clean_age,
        city=clean_city,
    )


def run_registration_form() -> None:
    """Interactive CLI flow that validates each field individually with helpful feedback."""
    print("=== Customer Registration ===")

    while True:
        try:
            name = validate_required_text(input("Full Name : "), "Full Name")
            break
        except MissingFieldError as err:
            print(f"  Error: {err}")

    while True:
        try:
            email = validate_email(input("Email     : "))
            break
        except (MissingFieldError, InvalidEmailError) as err:
            print(f"  Error: {err}")

    while True:
        try:
            age = validate_age(input("Age (18+) : "))
            break
        except (MissingFieldError, InvalidAgeError) as err:
            print(f"  Error: {err}")

    while True:
        try:
            city = validate_required_text(input("City      : "), "City")
            break
        except MissingFieldError as err:
            print(f"  Error: {err}")

    profile = CustomerProfile(full_name=name, email=email, age=age, city=city)
    print("\nRegistration Successful!")
    print(f"  Name  : {profile.full_name}")
    print(f"  Email : {profile.email}")
    print(f"  Age   : {profile.age}")
    print(f"  City  : {profile.city}")


if __name__ == "__main__":
    run_registration_form()
```

## Write a program that accepts a list of numbers and performs several calculations. Handle `ValueError`, `TypeError`, `ZeroDivisionError`, and other appropriate exceptions separately.

```python
import math
from typing import Iterable


def parse_number_list(raw_items: Iterable[object]) -> list[float]:
    """Converts an iterable of raw items into validated finite floats."""
    if isinstance(raw_items, (str, bytes)):
        raise TypeError("Expected a list or sequence of numbers, not a single string.")

    numbers: list[float] = []
    for idx, item in enumerate(raw_items):
        if item is None or isinstance(item, bool):
            raise TypeError(f"Element at index {idx} has unsupported type: {type(item).__name__}.")

        try:
            num = float(item)
        except TypeError as err:
            raise TypeError(f"Element at index {idx} ({item!r}) is not convertible to a number.") from err
        except ValueError as err:
            raise ValueError(f"Element at index {idx} ({item!r}) is not a valid numeric string.") from err

        if math.isnan(num) or math.isinf(num):
            raise ValueError(f"Element at index {idx} cannot be NaN or infinity.")

        numbers.append(num)

    return numbers


def analyze_numbers(raw_items: Iterable[object]) -> None:
    """Performs sum, average, range, geometric ratio, and reciprocal sum with separate error handling."""
    try:
        numbers = parse_number_list(raw_items)

        total = sum(numbers)
        count = len(numbers)
        average = total / count  # Raises ZeroDivisionError if list is empty
        spread = max(numbers) - min(numbers)

        # Sum of reciprocals (1 / x) — raises ZeroDivisionError if any element is 0
        reciprocal_sum = sum(1.0 / x for x in numbers)

        # Product of elements — raises OverflowError if magnitude exceeds float limits
        product = math.prod(numbers)
        if math.isinf(product):
            raise OverflowError("Product of numbers exceeded maximum floating-point range.")

    except TypeError as err:
        print(f"[TypeError] Invalid container or non-numeric object type: {err}")
    except ValueError as err:
        print(f"[ValueError] Invalid numeric value encountered: {err}")
    except ZeroDivisionError as err:
        print(f"[ZeroDivisionError] Division by zero occurred (empty list or zero element): {err}")
    except OverflowError as err:
        print(f"[OverflowError] Numeric overflow during calculation: {err}")
    except Exception as err:
        print(f"[Unexpected Error] An unhandled exception occurred: {err}")
    else:
        print("\n--- Calculation Results ---")
        print(f"Validated Numbers : {numbers}")
        print(f"Count             : {count}")
        print(f"Sum               : {total:,.4f}")
        print(f"Average (Mean)    : {average:,.4f}")
        print(f"Range (Max - Min) : {spread:,.4f}")
        print(f"Product           : {product:,.4f}")
        print(f"Sum of Reciprocals: {reciprocal_sum:,.4f}")


def get_numbers_from_user() -> list[str]:
    """Prompts the user for comma-separated numbers."""
    raw_input = input("Enter numbers separated by commas (e.g., 10, 25.5, 4): ").strip()
    if not raw_input:
        return []
    return [token.strip() for token in raw_input.split(",")]


if __name__ == "__main__":
    # 1. Interactive user input run
    user_entries = get_numbers_from_user()
    analyze_numbers(user_entries)

    # 2. Automated edge-case demonstrations
    print("\n--- Testing Exception Scenarios ---")
    analyze_numbers([])                   # Triggers ZeroDivisionError (empty list)
    analyze_numbers([10, 0, 5])           # Triggers ZeroDivisionError (reciprocal of 0)
    analyze_numbers([10, "abc", 30])      # Triggers ValueError (unparseable string)
    analyze_numbers([10, None, 30])       # Triggers TypeError (NoneType element)
    analyze_numbers([1e200, 1e200])       # Triggers OverflowError (float overflow)

```

## Write a program to process customer records stored in nested dictionaries and lists. Handle missing dictionary keys, invalid list indexes, and incorrect data types.

```python
from dataclasses import dataclass
from typing import Any


@dataclass(frozen=True)
class CustomerSummary:
    """Represents a validated, flattened summary of a nested customer record."""
    customer_id: str
    full_name: str
    primary_city: str
    latest_order_amount: float
    total_spent: float


def extract_customer_summary(record: Any, record_index: int) -> CustomerSummary:
    """Safely extracts and validates fields from a nested customer dictionary."""
    if not isinstance(record, dict):
        raise TypeError(f"Record #{record_index}: Expected a dict, got {type(record).__name__}.")

    # 1. Extract top-level ID and nested profile fields
    customer_id = str(record["id"])  # Raises KeyError if 'id' is missing

    profile = record["profile"]  # Raises KeyError if 'profile' is missing
    if not isinstance(profile, dict):
        raise TypeError(f"'profile' must be a dict, got {type(profile).__name__}.")

    full_name = profile["name"]
    if not isinstance(full_name, str) or not full_name.strip():
        raise ValueError("'profile.name' must be a non-empty string.")

    # 2. Navigate nested list of addresses (expects at least 1 primary address at index 0)
    addresses = profile["addresses"]
    if not isinstance(addresses, list):
        raise TypeError(f"'profile.addresses' must be a list, got {type(addresses).__name__}.")

    primary_address = addresses  # Raises IndexError if 'addresses' is empty
    if not isinstance(primary_address, dict):
        raise TypeError(f"'addresses' must be a dict, got {type(primary_address).__name__}.")

    primary_city = primary_address["city"]
    if not isinstance(primary_city, str) or not primary_city.strip():
        raise ValueError("'addresses.city' must be a non-empty string.")

    # 3. Navigate nested list of orders and sum valid numeric amounts
    orders = record["orders"]
    if not isinstance(orders, list):
        raise TypeError(f"'orders' must be a list, got {type(orders).__name__}.")

    latest_order = orders[-1]  # Raises IndexError if 'orders' is empty
    if not isinstance(latest_order, dict):
        raise TypeError(f"Latest order entry must be a dict, got {type(latest_order).__name__}.")

    total_spent = 0.0
    latest_amount = 0.0

    for idx, order in enumerate(orders):
        if not isinstance(order, dict):
            raise TypeError(f"Order at index {idx} must be a dict, got {type(order).__name__}.")

        raw_amount = order["amount"]  # Raises KeyError if 'amount' is missing
        if isinstance(raw_amount, bool) or not isinstance(raw_amount, (int, float)):
            raise TypeError(
                f"Order #{idx} 'amount' must be int or float, got {type(raw_amount).__name__} ({raw_amount!r})."
            )
        if raw_amount < 0:
            raise ValueError(f"Order #{idx} 'amount' cannot be negative ({raw_amount}).")

        total_spent += float(raw_amount)
        if idx == len(orders) - 1:
            latest_amount = float(raw_amount)

    return CustomerSummary(
        customer_id=customer_id,
        full_name=full_name.strip(),
        primary_city=primary_city.strip(),
        latest_order_amount=latest_amount,
        total_spent=total_spent,
    )


def process_customer_batch(records: Any) -> None:
    """Processes a list of nested customer records and reports successes and categorized errors."""
    if not isinstance(records, list):
        print(f"[Fatal TypeError] Expected a list of customer records, got {type(records).__name__}.")
        return

    summaries: list[CustomerSummary] = []
    error_log: list[str] = []

    for idx, record in enumerate(records, start=1):
        label = record.get("id", f"Record#{idx}") if isinstance(record, dict) else f"Record#{idx}"
        try:
            summary = extract_customer_summary(record, idx)
        except KeyError as err:
            error_log.append(f"[{label}] KeyError — Missing required dictionary key: {err}")
        except IndexError as err:
            error_log.append(f"[{label}] IndexError — Missing expected list element (empty list): {err}")
        except TypeError as err:
            error_log.append(f"[{label}] TypeError — Incorrect data structure or field type: {err}")
        except ValueError as err:
            error_log.append(f"[{label}] ValueError — Invalid field value: {err}")
        else:
            summaries.append(summary)

    print("=" * 74)
    print(f"{'ID':<10} {'Customer Name':<20} {'City':<14} {'Latest (PKR)':>13} {'Total (PKR)':>13}")
    print("-" * 74)
    for item in summaries:
        print(
            f"{item.customer_id:<10} {item.full_name[:19]:<20} {item.primary_city[:13]:<14} "
            f"{item.latest_order_amount:>13,.2f} {item.total_spent:>13,.2f}"
        )
    print("=" * 74)

    if error_log:
        print("\n--- Skipped Malformed Customer Records ---")
        for entry in error_log:
            print(f"  ! {entry}")


if __name__ == "__main__":
    nested_customers: list[Any] = [
        # 1. Valid nested customer record
        {
            "id": "CUST-01",
            "profile": {
                "name": "Hamza Siddiqui",
                "addresses": [{"city": "Karachi", "area": "PECHS"}],
            },
            "orders": [{"order_id": "ORD-1", "amount": 4500}, {"order_id": "ORD-2", "amount": 12800.50}],
        },
        # 2. Missing 'city' dictionary key -> KeyError
        {
            "id": "CUST-02",
            "profile": {
                "name": "Mahnoor Tariq",
                "addresses": [{"area": "Gulberg"}],
            },
            "orders": [{"order_id": "ORD-3", "amount": 3200}],
        },
        # 3. Empty 'orders' list -> IndexError on orders[-1]
        {
            "id": "CUST-03",
            "profile": {
                "name": "Bilal Qureshi",
                "addresses": [{"city": "Lahore"}],
            },
            "orders": [],
        },
        # 4. Wrong data type ('orders' is a dict instead of a list, or 'amount' is a string) -> TypeError
        {
            "id": "CUST-04",
            "profile": {
                "name": "Sana Malik",
                "addresses": [{"city": "Islamabad"}],
            },
            "orders": [{"order_id": "ORD-4", "amount": "five thousand"}],
        },
        # 5. Top-level item is not a dictionary -> TypeError
        "CorruptedRecordString",
        # 6. Another valid customer record
        {
            "id": "CUST-05",
            "profile": {
                "name": "Zeeshan Haider",
                "addresses": [{"city": "Karachi"}, {"city": "Hyderabad"}],
            },
            "orders": [{"order_id": "ORD-5", "amount": 9500}],
        },
    ]

    process_customer_batch(nested_customers)
```

## Write a program to create and save a sales report into a file. Use `else` to report successful file creation and `finally` to guarantee the required cleanup or final status message.

```python
from datetime import datetime
from pathlib import Path
from typing import IO


def format_sales_report(sales_data: list[tuple[str, int, float]]) -> str:
    """Formats itemized sales records into a clean text report."""
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    lines = [
        "=" * 62,
        f"{'DAILY SALES REPORT':^62}",
        f"{f'Generated on: {timestamp}':^62}",
        "=" * 62,
        f"{'Item':<24} {'Units':>8} {'Unit Price':>13} {'Revenue':>13}",
        "-" * 62,
    ]

    grand_total = 0.0
    total_units = 0

    for item, units, price in sales_data:
        if units < 0 or price < 0:
            raise ValueError(f"Invalid negative value in sales record for '{item}'.")
        revenue = units * price
        grand_total += revenue
        total_units += units
        lines.append(f"{item[:23]:<24} {units:>8} {price:>13,.2f} {revenue:>13,.2f}")

    lines.extend([
        "=" * 62,
        f"{'TOTAL UNITS SOLD':<47} {total_units:>13}",
        f"{'TOTAL REVENUE (PKR)':<47} {grand_total:>13,.2f}",
        "=" * 62,
    ])
    return "\n".join(lines) + "\n"


def save_sales_report(file_path: str, sales_data: list[tuple[str, int, float]]) -> None:
    """Writes the sales report to disk using try-except-else-finally."""
    target = Path(file_path.strip())
    file_handle: IO[str] | None = None

    try:
        if not target.name:
            raise ValueError("File path cannot be empty.")

        report_content = format_sales_report(sales_data)

        # Open explicitly so `finally` can demonstrate guaranteed resource cleanup
        file_handle = target.open(mode="w", encoding="utf-8")
        file_handle.write(report_content)

    except ValueError as err:
        print(f"[Data Error] Could not generate sales report: {err}")
    except PermissionError:
        print(f"[Permission Error] Access denied when writing to '{target}'.")
    except IsADirectoryError:
        print(f"[Path Error] '{target}' is a directory, not a file.")
    except OSError as err:
        print(f"[File System Error] Failed to write '{target}': {err}")
    else:
        # Runs ONLY if formatting and file writing succeeded with zero exceptions
        print(f"Success! Sales report created and saved to '{target.resolve()}'.")
    finally:
        # Guaranteed cleanup and closing status message regardless of success or failure
        if file_handle is not None and not file_handle.closed:
            file_handle.close()
            print("[Cleanup] File stream safely flushed and closed.")
        print("[Status] Sales report export process completed.")


if __name__ == "__main__":
    daily_sales: list[tuple[str, int, float]] = [
        ("Wireless Keyboard", 14, 4_500.00),
        ("Gaming Mouse", 22, 3_200.00),
        ("27-inch IPS Monitor", 5, 48_000.00),
        ("USB-C Docking Station", 9, 8_500.00),
    ]

    output_path = input("Enter destination filename (e.g., sales_report.txt): ")
    save_sales_report(output_path, daily_sales)
```

## Write a program that repeatedly asks the user to enter a valid product price. Invalid input should not terminate the program. Continue asking until a valid price is entered.

## Write a program to validate a loan application. Use `raise` to reject invalid business conditions such as an invalid age, invalid income, or an excessive loan amount.

```python
import math
from dataclasses import dataclass


class LoanApplicationError(ValueError):
    """Base exception raised when a loan application violates business rules."""


class IneligibleAgeError(LoanApplicationError):
    """Raised when the applicant's age falls outside the eligible lending range."""


class InsufficientIncomeError(LoanApplicationError):
    """Raised when the applicant's monthly income is below the minimum requirement."""


class ExcessiveLoanAmountError(LoanApplicationError):
    """Raised when the requested loan exceeds the maximum allowed income multiplier."""


@dataclass(frozen=True)
class LoanApplication:
    """Represents a validated loan application."""
    applicant_name: str
    age: int
    monthly_income: float
    requested_amount: float


def validate_loan_application(
    applicant_name: str,
    age: int,
    monthly_income: float,
    requested_amount: float,
    min_age: int = 21,
    max_age: int = 60,
    min_monthly_income: float = 50_000.0,
    max_income_multiplier: float = 12.0,
) -> LoanApplication:
    """Validates business rules for a loan application and raises domain exceptions on failure."""
    clean_name = applicant_name.strip()
    if not clean_name:
        raise LoanApplicationError("Applicant name cannot be blank.")

    if isinstance(age, bool) or not isinstance(age, int) or not (min_age <= age <= max_age):
        raise IneligibleAgeError(
            f"Applicant age must be between {min_age} and {max_age} years (received {age})."
        )

    if math.isnan(monthly_income) or math.isinf(monthly_income) or monthly_income < min_monthly_income:
        raise InsufficientIncomeError(
            f"Minimum monthly income of PKR {min_monthly_income:,.2f} is required "
            f"(received PKR {monthly_income:,.2f})."
        )

    if math.isnan(requested_amount) or math.isinf(requested_amount) or requested_amount <= 0:
        raise LoanApplicationError("Requested loan amount must be a positive number.")

    max_eligible_loan = monthly_income * max_income_multiplier
    if requested_amount > max_eligible_loan:
        raise ExcessiveLoanAmountError(
            f"Requested loan of PKR {requested_amount:,.2f} exceeds the maximum eligible limit of "
            f"PKR {max_eligible_loan:,.2f} ({max_income_multiplier:g}x monthly income)."
        )

    return LoanApplication(
        applicant_name=clean_name,
        age=age,
        monthly_income=monthly_income,
        requested_amount=requested_amount,
    )


def run_loan_portal() -> None:
    """Collects user inputs, parses numeric fields, and enforces loan business rules."""
    print("=== Personal Loan Eligibility Check ===")
    try:
        name = input("Applicant Full Name       : ")
        age = int(input("Age (years)               : ").strip())
        income = float(input("Monthly Net Income (PKR)  : ").strip())
        loan_amount = float(input("Requested Loan (PKR)      : ").strip())

        application = validate_loan_application(name, age, income, loan_amount)

    except IneligibleAgeError as err:
        print(f"\n[Rejected — Age Policy] {err}")
    except InsufficientIncomeError as err:
        print(f"\n[Rejected — Income Policy] {err}")
    except ExcessiveLoanAmountError as err:
        print(f"\n[Rejected — Credit Limit] {err}")
    except LoanApplicationError as err:
        print(f"\n[Rejected — Application Error] {err}")
    except ValueError:
        print("\n[Input Error] Please enter valid numeric values for age, income, and loan amount.")
    else:
        print("\n[Approved] Loan application passed all eligibility checks!")
        print(f"  Applicant        : {application.applicant_name}")
        print(f"  Age              : {application.age} years")
        print(f"  Monthly Income   : PKR {application.monthly_income:,.2f}")
        print(f"  Approved Request : PKR {application.requested_amount:,.2f}")


if __name__ == "__main__":
    run_loan_portal()
```

## Create multiple custom exception classes for an online payment system. Handle invalid payment amount, insufficient balance, and unsupported payment method as different business errors.

```python
import math
from dataclasses import dataclass


class PaymentError(Exception):
    """Base exception for all online payment processing errors."""


class InvalidPaymentAmountError(PaymentError):
    """Raised when a payment amount is zero, negative, non-finite, or exceeds limits."""

    def __init__(self, amount: float, reason: str | None = None) -> None:
        self.amount = amount
        default_msg = f"Invalid payment amount (PKR {amount:,.2f}): Amount must be greater than zero."
        super().__init__(reason or default_msg)


class InsufficientBalanceError(PaymentError):
    """Raised when the wallet balance cannot cover the requested payment."""

    def __init__(self, balance: float, required: float) -> None:
        self.balance = balance
        self.required = required
        self.shortfall = required - balance
        super().__init__(
            f"Insufficient balance: Available PKR {balance:,.2f}, "
            f"required PKR {required:,.2f} (shortfall: PKR {self.shortfall:,.2f})."
        )


class UnsupportedPaymentMethodError(PaymentError):
    """Raised when the chosen payment method is not supported by the gateway."""

    def __init__(self, method: str, supported_methods: set[str]) -> None:
        self.method = method
        self.supported_methods = supported_methods
        allowed = ", ".join(sorted(supported_methods))
        super().__init__(
            f"Unsupported payment method '{method}'. Supported options: {allowed}."
        )


@dataclass
class PaymentGateway:
    """Simulates a digital payment processor with wallet balance and supported payment rails."""
    account_holder: str
    balance: float
    supported_methods: frozenset[str] = frozenset({"card", "raast", "easypaisa", "jazzcash", "bank_transfer"})
    max_single_transaction: float = 500_000.00

    def process_payment(self, amount: float, method: str) -> str:
        """Validates business rules and processes a payment transaction."""
        normalized_method = method.strip().lower()

        # 1. Validate payment method
        if normalized_method not in self.supported_methods:
            raise UnsupportedPaymentMethodError(method.strip(), set(self.supported_methods))

        # 2. Validate payment amount
        if isinstance(amount, bool) or not isinstance(amount, (int, float)):
            raise InvalidPaymentAmountError(0.0, f"Invalid amount type: {type(amount).__name__}.")
        if math.isnan(amount) or math.isinf(amount) or amount <= 0:
            raise InvalidPaymentAmountError(amount)
        if amount > self.max_single_transaction:
            raise InvalidPaymentAmountError(
                amount,
                f"Amount PKR {amount:,.2f} exceeds single transaction limit of PKR {self.max_single_transaction:,.2f}.",
            )

        # 3. Validate wallet balance
        if amount > self.balance:
            raise InsufficientBalanceError(balance=self.balance, required=amount)

        # 4. Commit transaction
        self.balance -= float(amount)
        return (
            f"Payment of PKR {amount:,.2f} via '{normalized_method.upper()}' succeeded. "
            f"Remaining balance: PKR {self.balance:,.2f}"
        )


def run_payment_checkout() -> None:
    """Interactive checkout loop demonstrating distinct handling for each business exception."""
    gateway = PaymentGateway(account_holder="Ayesha Siddiqui", balance=15_000.00)
    print(f"=== Online Checkout ({gateway.account_holder}) ===")
    print(f"Wallet Balance    : PKR {gateway.balance:,.2f}")
    print(f"Supported Methods : {', '.join(sorted(gateway.supported_methods))}\n")

    raw_method = input("Select payment method : ")
    raw_amount = input("Enter payment amount  : ").strip()

    try:
        amount = float(raw_amount)
        receipt = gateway.process_payment(amount=amount, method=raw_method)
    except ValueError:
        print(f"\n[Input Error] '{raw_amount}' is not a valid numeric amount.")
    except UnsupportedPaymentMethodError as err:
        print(f"\n[Method Error] {err}")
    except InvalidPaymentAmountError as err:
        print(f"\n[Amount Error] {err}")
    except InsufficientBalanceError as err:
        print(f"\n[Balance Error] {err} Please top up at least PKR {err.shortfall:,.2f}.")
    except PaymentError as err:
        print(f"\n[Payment Failed] {err}")
    else:
        print(f"\n[Receipt] {receipt}")


if __name__ == "__main__":
    run_payment_checkout()
```

## Create a class-based bank account system. Handle deposits and withdrawals using custom exceptions for invalid transaction amounts and insufficient balance.

```python
import math
from dataclasses import dataclass
from datetime import datetime


class BankAccountError(Exception):
    """Base exception for all bank account errors."""


class InvalidTransactionAmountError(BankAccountError):
    """Raised when a deposit or withdrawal amount is zero, negative, or non-finite."""

    def __init__(self, amount: float, reason: str | None = None) -> None:
        self.amount = amount
        default_msg = f"Invalid transaction amount (PKR {amount:,.2f}): Amount must be greater than zero."
        super().__init__(reason or default_msg)


class InsufficientBalanceError(BankAccountError):
    """Raised when a withdrawal exceeds the available account balance."""

    def __init__(self, balance: float, requested: float) -> None:
        self.balance = balance
        self.requested = requested
        self.shortfall = requested - balance
        super().__init__(
            f"Insufficient balance: Available PKR {balance:,.2f}, "
            f"requested PKR {requested:,.2f} (shortfall: PKR {self.shortfall:,.2f})."
        )


@dataclass(frozen=True)
class TransactionRecord:
    """Immutable audit entry for a completed deposit or withdrawal."""
    timestamp: str
    tx_type: str
    amount: float
    balance_after: float


class BankAccount:
    """Represents a customer bank account with validated deposits, withdrawals, and statement history."""

    def __init__(self, account_number: str, holder_name: str, initial_balance: float = 0.0) -> None:
        if not account_number.strip() or not holder_name.strip():
            raise ValueError("Account number and holder name cannot be blank.")
        if isinstance(initial_balance, bool) or math.isnan(initial_balance) or math.isinf(initial_balance) or initial_balance < 0:
            raise InvalidTransactionAmountError(
                initial_balance, "Initial balance cannot be negative or non-finite."
            )

        self.account_number = account_number.strip()
        self.holder_name = holder_name.strip()
        self._balance = float(initial_balance)
        self._history: list[TransactionRecord] = []

        if self._balance > 0:
            self._record_transaction("OPENING_DEPOSIT", self._balance)

    @property
    def balance(self) -> float:
        """Read-only access to the current account balance."""
        return self._balance

    def _validate_amount(self, amount: float) -> float:
        """Ensures a transaction amount is a valid, finite positive number."""
        if isinstance(amount, bool) or not isinstance(amount, (int, float)):
            raise InvalidTransactionAmountError(0.0, f"Amount must be numeric, got {type(amount).__name__}.")
        if math.isnan(amount) or math.isinf(amount) or amount <= 0:
            raise InvalidTransactionAmountError(amount)
        return float(amount)

    def _record_transaction(self, tx_type: str, amount: float) -> None:
        self._history.append(
            TransactionRecord(
                timestamp=datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
                tx_type=tx_type,
                amount=amount,
                balance_after=self._balance,
            )
        )

    def deposit(self, amount: float) -> float:
        """Deposits a positive amount into the account and returns the updated balance."""
        valid_amount = self._validate_amount(amount)
        self._balance += valid_amount
        self._record_transaction("DEPOSIT", valid_amount)
        return self._balance

    def withdraw(self, amount: float) -> float:
        """Withdraws a positive amount if sufficient funds exist and returns the updated balance."""
        valid_amount = self._validate_amount(amount)
        if valid_amount > self._balance:
            raise InsufficientBalanceError(balance=self._balance, requested=valid_amount)

        self._balance -= valid_amount
        self._record_transaction("WITHDRAWAL", valid_amount)
        return self._balance

    def print_statement(self) -> None:
        """Displays a formatted mini-statement of all account transactions."""
        print("\n" + "=" * 68)
        print(f"ACCOUNT STATEMENT | {self.holder_name} ({self.account_number})")
        print("=" * 68)
        print(f"{'Timestamp':<21} {'Type':<17} {'Amount (PKR)':>13} {'Balance (PKR)':>14}")
        print("-" * 68)
        for tx in self._history:
            print(f"{tx.timestamp:<21} {tx.tx_type:<17} {tx.amount:>13,.2f} {tx.balance_after:>14,.2f}")
        print("=" * 68)


def run_banking_cli() -> None:
    """Interactive menu demonstrating deposits, withdrawals, and custom exception handling."""
    account = BankAccount(account_number="PK36-HABB-00192837", holder_name="Zainab Raza", initial_balance=20_000.00)
    print(f"Welcome, {account.holder_name}! Opening Balance: PKR {account.balance:,.2f}")

    while True:
        print("\n[1] Deposit   [2] Withdraw   [3] Mini-Statement   [4] Exit")
        choice = input("Choose an option (1-4): ").strip()

        if choice == "4":
            print("Session ended. Thank you for banking with us.")
            break
        if choice == "3":
            account.print_statement()
            continue
        if choice not in {"1", "2"}:
            print("Invalid option. Please choose 1, 2, 3, or 4.")
            continue

        raw_amount = input("Enter amount (PKR): ").strip()
        try:
            amount = float(raw_amount)
            if choice == "1":
                new_bal = account.deposit(amount)
                print(f"Deposited PKR {amount:,.2f}. New Balance: PKR {new_bal:,.2f}")
            else:
                new_bal = account.withdraw(amount)
                print(f"Withdrew PKR {amount:,.2f}. Remaining Balance: PKR {new_bal:,.2f}")
        except ValueError:
            print(f"Input Error: '{raw_amount}' is not a valid numeric amount.")
        except InvalidTransactionAmountError as err:
            print(f"Transaction Rejected (Invalid Amount): {err}")
        except InsufficientBalanceError as err:
            print(f"Transaction Rejected (Insufficient Funds): {err}")


if __name__ == "__main__":
    run_banking_cli()
```

## Create an e-commerce order system that handles invalid products, invalid quantities, insufficient stock, and invalid payments. Use custom exceptions for business-rule failures.

```python
import math
from dataclasses import dataclass, field


# --- Custom Business-Rule Exceptions ---
class OrderSystemError(Exception):
    """Base exception for all e-commerce order processing failures."""


class InvalidProductError(OrderSystemError):
    """Raised when a requested SKU or product does not exist in the catalog."""

    def __init__(self, sku: str) -> None:
        self.sku = sku
        super().__init__(f"Product SKU '{sku}' does not exist in the store catalog.")


class InvalidQuantityError(OrderSystemError):
    """Raised when an ordered quantity is not a positive whole integer."""

    def __init__(self, quantity: object) -> None:
        self.quantity = quantity
        super().__init__(f"Invalid quantity ({quantity!r}): Quantity must be a whole integer of at least 1.")


class InsufficientStockError(OrderSystemError):
    """Raised when requested quantity exceeds available inventory."""

    def __init__(self, product_name: str, sku: str, requested: int, available: int) -> None:
        self.product_name = product_name
        self.sku = sku
        self.requested = requested
        self.available = available
        super().__init__(
            f"Insufficient stock for '{product_name}' ({sku}): "
            f"Requested {requested}, but only {available} left in stock."
        )


class InvalidPaymentError(OrderSystemError):
    """Raised when payment method is unsupported or payment amount does not match order total."""

    def __init__(self, reason: str) -> None:
        super().__init__(f"Payment failed: {reason}")


# --- Domain Models ---
@dataclass
class Product:
    """Represents an item in the store inventory."""
    sku: str
    name: str
    price: float
    stock: int


@dataclass(frozen=True)
class OrderItem:
    """Represents a validated line item inside an order."""
    sku: str
    name: str
    unit_price: float
    quantity: int

    @property
    def subtotal(self) -> float:
        return self.unit_price * self.quantity


@dataclass
class StoreCatalog:
    """Manages store products, stock validation, and order checkout."""
    products: dict[str, Product] = field(default_factory=dict)
    supported_payment_methods: frozenset[str] = frozenset({"card", "raast", "easypaisa", "jazzcash", "cod"})

    def add_product(self, product: Product) -> None:
        self.products[product.sku.strip().upper()] = product

    def display_catalog(self) -> None:
        print("\n" + "=" * 64)
        print(f"{'SKU':<10} {'Product Name':<26} {'Price (PKR)':>13} {'In Stock':>10}")
        print("-" * 64)
        for item in self.products.values():
            print(f"{item.sku:<10} {item.name[:25]:<26} {item.price:>13,.2f} {item.stock:>10}")
        print("=" * 64)

    def place_order(
        self,
        cart_requests: list[tuple[str, int]],
        payment_method: str,
        payment_amount: float,
    ) -> list[OrderItem]:
        """
        Validates all items, stock levels, and payment before committing inventory deductions.
        Guarantees atomic execution: if any rule fails, no stock is deducted.
        """
        if not cart_requests:
            raise OrderSystemError("Cannot place an empty order.")

        # 1. Aggregate quantities per SKU so split lines don't bypass stock checks
        aggregated_qty: dict[str, int] = {}
        for raw_sku, qty in cart_requests:
            sku = str(raw_sku).strip().upper()
            if sku not in self.products:
                raise InvalidProductError(raw_sku)
            if isinstance(qty, bool) or not isinstance(qty, int) or qty <= 0:
                raise InvalidQuantityError(qty)
            aggregated_qty[sku] = aggregated_qty.get(sku, 0) + qty

        # 2. Validate stock availability for every item before mutating inventory
        order_lines: list[OrderItem] = []
        for sku, total_qty in aggregated_qty.items():
            product = self.products[sku]
            if total_qty > product.stock:
                raise InsufficientStockError(
                    product_name=product.name,
                    sku=sku,
                    requested=total_qty,
                    available=product.stock,
                )
            order_lines.append(
                OrderItem(
                    sku=sku,
                    name=product.name,
                    unit_price=product.price,
                    quantity=total_qty,
                )
            )

        # 3. Validate payment method and payment amount
        grand_total = sum(line.subtotal for line in order_lines)
        normalized_method = payment_method.strip().lower()

        if normalized_method not in self.supported_payment_methods:
            allowed = ", ".join(sorted(self.supported_payment_methods))
            raise InvalidPaymentError(f"Unsupported method '{payment_method}'. Allowed: {allowed}.")

        if isinstance(payment_amount, bool) or not isinstance(payment_amount, (int, float)):
            raise InvalidPaymentError("Payment amount must be a valid number.")
        if math.isnan(payment_amount) or math.isinf(payment_amount) or payment_amount <= 0:
            raise InvalidPaymentError(f"Payment amount must be greater than zero (received {payment_amount}).")
        if not math.isclose(payment_amount, grand_total, rel_tol=1e-9, abs_tol=0.01):
            raise InvalidPaymentError(
                f"Amount mismatch: Order total is PKR {grand_total:,.2f}, but received PKR {payment_amount:,.2f}."
            )

        # 4. Commit stock deductions only after all checks succeed
        for line in order_lines:
            self.products[line.sku].stock -= line.quantity

        return order_lines


def print_order_confirmation(order_lines: list[OrderItem], payment_method: str) -> None:
    """Displays an itemized invoice for a completed order."""
    print("\n--- Order Confirmed ---")
    for line in order_lines:
        print(f"  {line.quantity}x {line.name} ({line.sku}) @ PKR {line.unit_price:,.2f} = PKR {line.subtotal:,.2f}")
    total = sum(line.subtotal for line in order_lines)
    print(f"Payment Method : {payment_method.upper()}")
    print(f"Total Paid     : PKR {total:,.2f}")


if __name__ == "__main__":
    store = StoreCatalog()
    store.add_product(Product(sku="SKU-101", name="Mechanical Keyboard", price=12_500.00, stock=5))
    store.add_product(Product(sku="SKU-102", name="Wireless Mouse", price=4_500.00, stock=10))
    store.add_product(Product(sku="SKU-103", name="USB-C Hub", price=6_000.00, stock=2))

    store.display_catalog()

    test_scenarios = [
        # Scenario 1: Invalid SKU -> InvalidProductError
        ([("SKU-999", 1)], "card", 5_000.00),
        # Scenario 2: Invalid quantity -> InvalidQuantityError
        ([("SKU-101", 0)], "card", 12_500.00),
        # Scenario 3: Insufficient stock -> InsufficientStockError
        ([("SKU-103", 4)], "raast", 24_000.00),
        # Scenario 4: Wrong payment amount -> InvalidPaymentError (stock must remain untouched)
        ([("SKU-101", 1), ("SKU-102", 2)], "easypaisa", 10_000.00),
        # Scenario 5: Valid order -> Succeeds and deducts inventory
        ([("SKU-101", 1), ("SKU-102", 2)], "raast", 21_500.00),
    ]

    for idx, (items, method, paid) in enumerate(test_scenarios, start=1):
        print(f"\n[Test Order #{idx}] Items={items}, Method='{method}', Paid=PKR {paid:,.2f}")
        try:
            confirmed_items = store.place_order(items, method, paid)
        except InvalidProductError as err:
            print(f"  Rejected (Product Error)  : {err}")
        except InvalidQuantityError as err:
            print(f"  Rejected (Quantity Error) : {err}")
        except InsufficientStockError as err:
            print(f"  Rejected (Stock Error)    : {err}")
        except InvalidPaymentError as err:
            print(f"  Rejected (Payment Error)  : {err}")
        except OrderSystemError as err:
            print(f"  Rejected (Order Error)    : {err}")
        else:
            print_order_confirmation(confirmed_items, method)

    print("\nUpdated Inventory After All Orders:")
    store.display_catalog()
```

## ### University Course Registration System

### Goal:
Create a program that allows students to register for university courses.

### Steps:
- Store students and courses.
- Check whether the student exists.
- Check whether the course exists.
- Prevent duplicate registration.
- Check course capacity.
- Check prerequisites.
- Create and handle suitable custom exceptions.

→ Registration successful

```python
from dataclasses import dataclass, field


# --- Custom Business-Rule Exceptions ---
class RegistrationSystemError(Exception):
    """Base exception for all university course registration errors."""


class StudentNotFoundError(RegistrationSystemError):
    """Raised when the provided student ID does not exist in the university registry."""

    def __init__(self, student_id: int) -> None:
        self.student_id = student_id
        super().__init__(f"Student ID {student_id} was not found in the system.")


class CourseNotFoundError(RegistrationSystemError):
    """Raised when the requested course does not exist in the catalog."""

    def __init__(self, course_name: str) -> None:
        self.course_name = course_name
        super().__init__(f"Course '{course_name}' does not exist in the course catalog.")


class DuplicateRegistrationError(RegistrationSystemError):
    """Raised when a student attempts to register for a course they are already enrolled in."""

    def __init__(self, student_name: str, course_title: str) -> None:
        super().__init__(f"{student_name} is already registered for '{course_title}'.")


class CourseFullError(RegistrationSystemError):
    """Raised when a course has reached its maximum seat capacity."""

    def __init__(self, course_title: str, capacity: int) -> None:
        super().__init__(f"Course '{course_title}' is full (maximum capacity: {capacity} seats).")


class PrerequisiteNotMetError(RegistrationSystemError):
    """Raised when a student has not completed all required prerequisite courses."""

    def __init__(self, student_name: str, course_title: str, missing_prereqs: set[str]) -> None:
        self.missing_prereqs = missing_prereqs
        missing_list = ", ".join(sorted(missing_prereqs))
        super().__init__(
            f"{student_name} cannot register for '{course_title}'. "
            f"Missing prerequisite(s): {missing_list}."
        )


# --- Domain Models ---
@dataclass
class Student:
    """Represents a university student and their academic enrollment state."""
    student_id: int
    name: str
    completed_courses: set[str] = field(default_factory=set)
    enrolled_courses: set[str] = field(default_factory=set)


@dataclass
class Course:
    """Represents a university course with seat capacity and prerequisite requirements."""
    code: str
    title: str
    capacity: int
    prerequisites: set[str] = field(default_factory=set)
    enrolled_students: set[int] = field(default_factory=set)

    @property
    def available_seats(self) -> int:
        return self.capacity - len(self.enrolled_students)


class UniversityRegistry:
    """Manages students, courses, and rule-enforced course registration."""

    def __init__(self) -> None:
        self.students: dict[int, Student] = {}
        self.courses: dict[str, Course] = {}  # Keyed by lowercase title for case-insensitive lookup

    def add_student(self, student: Student) -> None:
        self.students[student.student_id] = student

    def add_course(self, course: Course) -> None:
        self.courses[course.title.strip().lower()] = course

    def display_catalog(self) -> None:
        print("\n" + "=" * 72)
        print(f"{'Code':<10} {'Course Title':<24} {'Seats':<10} {'Prerequisites'}")
        print("-" * 72)
        for course in self.courses.values():
            seats = f"{len(course.enrolled_students)}/{course.capacity}"
            prereqs = ", ".join(sorted(course.prerequisites)) if course.prerequisites else "None"
            print(f"{course.code:<10} {course.title:<24} {seats:<10} {prereqs}")
        print("=" * 72)

    def register(self, student_id: int, course_name: str) -> str:
        """Validates all registration rules and enrolls the student in the course."""
        # 1. Check whether the student exists
        if student_id not in self.students:
            raise StudentNotFoundError(student_id)
        student = self.students[student_id]

        # 2. Check whether the course exists
        lookup_key = course_name.strip().lower()
        if lookup_key not in self.courses:
            raise CourseNotFoundError(course_name.strip())
        course = self.courses[lookup_key]

        # 3. Prevent duplicate registration
        if course.title in student.enrolled_courses or student_id in course.enrolled_students:
            raise DuplicateRegistrationError(student.name, course.title)

        # 4. Check course capacity
        if course.available_seats <= 0:
            raise CourseFullError(course.title, course.capacity)

        # 5. Check prerequisites (case-insensitive comparison against completed courses)
        completed_normalized = {c.lower() for c in student.completed_courses}
        missing = {
            prereq for prereq in course.prerequisites
            if prereq.lower() not in completed_normalized
        }
        if missing:
            raise PrerequisiteNotMetError(student.name, course.title, missing)

        # 6. Commit enrollment atomically
        student.enrolled_courses.add(course.title)
        course.enrolled_students.add(student.student_id)

        return (
            f"Registration successful: {student.name} (ID: {student.student_id}) "
            f"enrolled in '{course.title}' ({course.code})."
        )


def attempt_registration(registry: UniversityRegistry, student_id: int, course_name: str) -> None:
    """Helper to execute a registration request and display clear status feedback."""
    print(f"\nStudent ID: {student_id} | Course: {course_name}")
    try:
        message = registry.register(student_id, course_name)
    except StudentNotFoundError as err:
        print(f"-> Failed [Student Error]      : {err}")
    except CourseNotFoundError as err:
        print(f"-> Failed [Course Error]       : {err}")
    except DuplicateRegistrationError as err:
        print(f"-> Failed [Duplicate Error]    : {err}")
    except CourseFullError as err:
        print(f"-> Failed [Capacity Error]     : {err}")
    except PrerequisiteNotMetError as err:
        print(f"-> Failed [Prerequisite Error] : {err}")
    except RegistrationSystemError as err:
        print(f"-> Failed [System Error]       : {err}")
    else:
        print(f"-> {message}")


if __name__ == "__main__":
    uni = UniversityRegistry()

    # Seed students
    uni.add_student(Student(student_id=101, name="Ali Raza", completed_courses={"Programming Fundamentals"}))
    uni.add_student(Student(student_id=102, name="Ayesha Khan", completed_courses={"Programming Fundamentals", "Python"}))
    uni.add_student(Student(student_id=103, name="Usman Tariq", completed_courses=set()))

    # Seed courses
    uni.add_course(Course(code="CS-101", title="Programming Fundamentals", capacity=30))
    uni.add_course(Course(code="CS-102", title="Python", capacity=2, prerequisites={"Programming Fundamentals"}))
    uni.add_course(Course(code="CS-201", title="Data Science", capacity=1, prerequisites={"Python"}))

    uni.display_catalog()

    # Run test scenarios
    attempt_registration(uni, 102, "Python")         # Success (matches prompt example)
    attempt_registration(uni, 102, "Python")         # DuplicateRegistrationError
    attempt_registration(uni, 103, "Python")         # PrerequisiteNotMetError (missing Programming Fundamentals)
    attempt_registration(uni, 101, "Python")         # Success (fills Python's 2nd and final seat)
    attempt_registration(uni, 999, "Python")         # StudentNotFoundError
    attempt_registration(uni, 101, "Quantum AI")     # CourseNotFoundError
    attempt_registration(uni, 102, "Data Science")   # Success (fills Data Science's 1 seat)
    attempt_registration(uni, 101, "Data Science")   # CourseFullError (capacity 1 already taken)

    uni.display_catalog()
```

## ### Hospital Appointment System

### Goal:
Create a program that manages hospital appointments.

### Steps:
- Store patients and doctors.
- Validate patient and doctor information.
- Check doctor availability.
- Prevent duplicate appointment times.
- Create custom exceptions for important business errors.
- Handle invalid input without crashing.

### Example:
Patient ID: P102
Doctor: D05
Time: 10:30 AM
→ Appointment booked successfully

```python
from dataclasses import dataclass, field
from datetime import datetime


# --- Custom Business-Rule Exceptions ---
class AppointmentSystemError(Exception):
    """Base exception for all hospital appointment errors."""


class PatientNotFoundError(AppointmentSystemError):
    """Raised when a patient ID is not registered in the hospital system."""

    def __init__(self, patient_id: str) -> None:
        self.patient_id = patient_id
        super().__init__(f"Patient ID '{patient_id}' was not found in the registry.")


class DoctorNotFoundError(AppointmentSystemError):
    """Raised when a doctor ID does not exist in the hospital system."""

    def __init__(self, doctor_id: str) -> None:
        self.doctor_id = doctor_id
        super().__init__(f"Doctor ID '{doctor_id}' was not found in the directory.")


class InvalidAppointmentTimeError(AppointmentSystemError):
    """Raised when the requested time format is invalid or outside clinic hours."""

    def __init__(self, raw_time: str, reason: str) -> None:
        self.raw_time = raw_time
        super().__init__(f"Invalid appointment time '{raw_time}': {reason}")


class DoctorUnavailableError(AppointmentSystemError):
    """Raised when a doctor is not on duty at the requested time slot."""

    def __init__(self, doctor_name: str, time_slot: str, available_slots: list[str]) -> None:
        slots_str = ", ".join(available_slots) if available_slots else "None"
        super().__init__(
            f"{doctor_name} is not available at {time_slot}. "
            f"Scheduled clinic slots: {slots_str}."
        )


class DuplicateAppointmentError(AppointmentSystemError):
    """Raised when a time slot is already booked with the doctor or the patient has a conflict."""

    def __init__(self, reason: str) -> None:
        super().__init__(f"Booking conflict: {reason}")


# --- Domain Models ---
@dataclass
class Patient:
    """Represents a registered hospital patient."""
    patient_id: str
    name: str
    age: int


@dataclass
class Doctor:
    """Represents a hospital doctor and their available daily time slots."""
    doctor_id: str
    name: str
    specialty: str
    available_slots: set[str] = field(default_factory=set)
    booked_slots: dict[str, str] = field(default_factory=dict)  # time_slot -> patient_id


@dataclass(frozen=True)
class Appointment:
    """Represents a confirmed hospital appointment."""
    appointment_id: str
    patient_id: str
    patient_name: str
    doctor_id: str
    doctor_name: str
    time_slot: str


class HospitalScheduler:
    """Manages patients, doctors, and conflict-free appointment scheduling."""

    def __init__(self) -> None:
        self.patients: dict[str, Patient] = {}
        self.doctors: dict[str, Doctor] = {}
        self.appointments: list[Appointment] = []

    @staticmethod
    def normalize_time(raw_time: str) -> str:
        """Validates and normalizes time strings (e.g., '10:30 am' -> '10:30 AM')."""
        cleaned = raw_time.strip().upper()
        if not cleaned:
            raise InvalidAppointmentTimeError(raw_time, "Time cannot be blank.")

        for fmt in ("%I:%M %p", "%I:%M%p", "%H:%M"):
            try:
                parsed = datetime.strptime(cleaned, fmt)
                return parsed.strftime("%I:%M %p")
            except ValueError:
                continue

        raise InvalidAppointmentTimeError(
            raw_time,
            "Expected format like '10:30 AM' or '14:30'.",
        )

    def add_patient(self, patient: Patient) -> None:
        pid = patient.patient_id.strip().upper()
        if not pid or not patient.name.strip():
            raise ValueError("Patient ID and name cannot be blank.")
        if patient.age < 0 or patient.age > 130:
            raise ValueError(f"Invalid patient age ({patient.age}).")
        patient.patient_id = pid
        self.patients[pid] = patient

    def add_doctor(self, doctor: Doctor) -> None:
        did = doctor.doctor_id.strip().upper()
        if not did or not doctor.name.strip():
            raise ValueError("Doctor ID and name cannot be blank.")
        doctor.doctor_id = did
        doctor.available_slots = {self.normalize_time(slot) for slot in doctor.available_slots}
        self.doctors[did] = doctor

    def display_doctors(self) -> None:
        print("\n" + "=" * 78)
        print(f"{'ID':<7} {'Doctor Name':<22} {'Specialty':<16} {'Open Slots'}")
        print("-" * 78)
        for doc in self.doctors.values():
            open_slots = sorted(
                slot for slot in doc.available_slots if slot not in doc.booked_slots
            )
            slots_text = ", ".join(open_slots) if open_slots else "Fully Booked"
            print(f"{doc.doctor_id:<7} {doc.name:<22} {doc.specialty:<16} {slots_text}")
        print("=" * 78)

    def book_appointment(self, patient_id: str, doctor_id: str, raw_time: str) -> Appointment:
        """Validates patient, doctor, schedule, and conflicts before confirming an appointment."""
        pid = patient_id.strip().upper()
        did = doctor_id.strip().upper()

        # 1. Validate patient existence
        if pid not in self.patients:
            raise PatientNotFoundError(pid or patient_id)
        patient = self.patients[pid]

        # 2. Validate doctor existence
        if did not in self.doctors:
            raise DoctorNotFoundError(did or doctor_id)
        doctor = self.doctors[did]

        # 3. Validate and normalize appointment time
        time_slot = self.normalize_time(raw_time)

        # 4. Check whether the doctor works during this time slot
        if time_slot not in doctor.available_slots:
            raise DoctorUnavailableError(
                doctor.name,
                time_slot,
                sorted(doctor.available_slots),
            )

        # 5. Prevent duplicate booking on the doctor's schedule
        if time_slot in doctor.booked_slots:
            raise DuplicateAppointmentError(
                f"{doctor.name} already has an appointment booked at {time_slot}."
            )

        # 6. Prevent the same patient from double-booking two doctors at the exact same time
        for existing in self.appointments:
            if existing.patient_id == pid and existing.time_slot == time_slot:
                raise DuplicateAppointmentError(
                    f"{patient.name} ({pid}) already has an appointment with {existing.doctor_name} at {time_slot}."
                )

        # 7. Commit booking atomically
        doctor.booked_slots[time_slot] = pid
        appointment = Appointment(
            appointment_id=f"APT-{len(self.appointments) + 1001}",
            patient_id=pid,
            patient_name=patient.name,
            doctor_id=did,
            doctor_name=doctor.name,
            time_slot=time_slot,
        )
        self.appointments.append(appointment)
        return appointment


def run_appointment_desk() -> None:
    """Runs automated verification tests followed by an interactive booking prompt."""
    hospital = HospitalScheduler()

    # Seed registered patients
    hospital.add_patient(Patient(patient_id="P101", name="Bilal Ahmed", age=34))
    hospital.add_patient(Patient(patient_id="P102", name="Sara Tariq", age=28))
    hospital.add_patient(Patient(patient_id="P103", name="Kamran Siddiqui", age=45))

    # Seed doctors with clinic hours
    hospital.add_doctor(
        Doctor(
            doctor_id="D05",
            name="Dr. Ayesha Malik",
            specialty="Cardiology",
            available_slots={"10:00 AM", "10:30 AM", "11:00 AM"},
        )
    )
    hospital.add_doctor(
        Doctor(
            doctor_id="D08",
            name="Dr. Usman Raza",
            specialty="Dermatology",
            available_slots={"10:30 AM", "02:00 PM"},
        )
    )

    hospital.display_doctors()

    # Demonstrate prompt example and business exceptions
    sample_requests = [
        ("P102", "D05", "10:30 AM"),  # Valid booking (matches prompt example)
        ("P101", "D05", "10:30 AM"),  # DuplicateAppointmentError (slot already taken)
        ("P102", "D08", "10:30 AM"),  # DuplicateAppointmentError (patient double-booked at 10:30 AM)
        ("P101", "D05", "03:00 PM"),  # DoctorUnavailableError (outside doctor's slots)
        ("P999", "D05", "11:00 AM"),  # PatientNotFoundError
        ("P101", "D99", "11:00 AM"),  # DoctorNotFoundError
        ("P101", "D05", "invalid"),   # InvalidAppointmentTimeError
    ]

    for pid, did, slot in sample_requests:
        print(f"\nPatient ID: {pid} | Doctor: {did} | Time: {slot}")
        try:
            appt = hospital.book_appointment(pid, did, slot)
        except PatientNotFoundError as err:
            print(f"-> Failed [Patient Error]   : {err}")
        except DoctorNotFoundError as err:
            print(f"-> Failed [Doctor Error]    : {err}")
        except InvalidAppointmentTimeError as err:
            print(f"-> Failed [Time Error]      : {err}")
        except DoctorUnavailableError as err:
            print(f"-> Failed [Schedule Error]  : {err}")
        except DuplicateAppointmentError as err:
            print(f"-> Failed [Conflict Error]  : {err}")
        except AppointmentSystemError as err:
            print(f"-> Failed [System Error]    : {err}")
        else:
            print(
                f"-> Appointment booked successfully! "
                f"(Ref: {appt.appointment_id} | {appt.patient_name} with {appt.doctor_name} at {appt.time_slot})"
            )

    hospital.display_doctors()


if __name__ == "__main__":
    run_appointment_desk()
```

## ### Digital Wallet System

### Goal:
Create a class-based digital wallet application.

### Steps:
- Validate wallet ID and PIN.
- Handle inactive accounts.
- Validate transfer amounts.
- Check balance.
- Enforce a transfer limit.
- Create multiple custom exceptions.
- Make sure failed transfers do not change wallet balances.

### Example:
Transfer: Rs. 2,000
→ Transfer successful

```python
import math
from dataclasses import dataclass, field
from datetime import datetime


# --- Custom Business-Rule Exceptions ---
class WalletSystemError(Exception):
    """Base exception for all digital wallet errors."""


class WalletNotFoundError(WalletSystemError):
    """Raised when a sender or receiver wallet ID does not exist."""

    def __init__(self, wallet_id: str) -> None:
        self.wallet_id = wallet_id
        super().__init__(f"Wallet ID '{wallet_id}' was not found.")


class InvalidPinError(WalletSystemError):
    """Raised when the entered PIN is malformed or does not match the wallet's PIN."""

    def __init__(self, reason: str = "Incorrect PIN entered.") -> None:
        super().__init__(f"Authentication failed: {reason}")


class InactiveWalletError(WalletSystemError):
    """Raised when either the sender or recipient wallet is frozen/inactive."""

    def __init__(self, wallet_id: str, owner_name: str) -> None:
        self.wallet_id = wallet_id
        super().__init__(f"Wallet '{wallet_id}' ({owner_name}) is currently inactive.")


class InvalidTransferAmountError(WalletSystemError):
    """Raised when the transfer amount is zero, negative, non-finite, or self-directed."""

    def __init__(self, reason: str) -> None:
        super().__init__(f"Invalid transfer: {reason}")


class TransferLimitExceededError(WalletSystemError):
    """Raised when a transfer exceeds the single-transaction or daily limit."""

    def __init__(self, amount: float, limit: float, limit_type: str = "single-transaction") -> None:
        self.amount = amount
        self.limit = limit
        super().__init__(
            f"Transfer of Rs. {amount:,.2f} exceeds the {limit_type} limit of Rs. {limit:,.2f}."
        )


class InsufficientBalanceError(WalletSystemError):
    """Raised when the sender's wallet balance cannot cover the transfer amount."""

    def __init__(self, balance: float, required: float) -> None:
        self.balance = balance
        self.required = required
        self.shortfall = required - balance
        super().__init__(
            f"Insufficient balance: Available Rs. {balance:,.2f}, "
            f"required Rs. {required:,.2f} (shortfall: Rs. {self.shortfall:,.2f})."
        )


# --- Domain Models ---
@dataclass(frozen=True)
class TransferReceipt:
    """Immutable audit receipt for a completed wallet transfer."""
    tx_id: str
    timestamp: str
    sender_id: str
    receiver_id: str
    amount: float
    sender_balance_after: float


@dataclass
class DigitalWallet:
    """Represents an individual user's digital wallet."""
    wallet_id: str
    owner_name: str
    pin: str
    balance: float
    is_active: bool = True
    single_tx_limit: float = 50_000.00
    daily_limit: float = 100_000.00
    daily_transferred: float = 0.0
    history: list[TransferReceipt] = field(default_factory=list)

    def verify_pin(self, entered_pin: str) -> None:
        cleaned_pin = entered_pin.strip()
        if not (cleaned_pin.isdigit() and len(cleaned_pin) == 4):
            raise InvalidPinError("PIN must be a 4-digit numeric code.")
        if cleaned_pin != self.pin:
            raise InvalidPinError("Incorrect PIN for this wallet.")


class WalletNetwork:
    """Manages registered wallets and executes atomic peer-to-peer transfers."""

    def __init__(self) -> None:
        self.wallets: dict[str, DigitalWallet] = {}
        self._tx_counter = 5000

    def register_wallet(self, wallet: DigitalWallet) -> None:
        wid = wallet.wallet_id.strip().upper()
        if not wid or not wallet.owner_name.strip():
            raise ValueError("Wallet ID and owner name cannot be blank.")
        if not (wallet.pin.strip().isdigit() and len(wallet.pin.strip()) == 4):
            raise ValueError("Initial wallet PIN must be a 4-digit number.")
        if wallet.balance < 0:
            raise ValueError("Initial balance cannot be negative.")
        wallet.wallet_id = wid
        self.wallets[wid] = wallet

    def transfer(
        self,
        sender_id: str,
        pin: str,
        receiver_id: str,
        amount: float,
    ) -> TransferReceipt:
        """
        Validates all rules before mutating any state.
        Guarantees atomicity: if any exception is raised, both balances remain untouched.
        """
        s_id = sender_id.strip().upper()
        r_id = receiver_id.strip().upper()

        # 1. Validate wallet IDs exist
        if s_id not in self.wallets:
            raise WalletNotFoundError(s_id or sender_id)
        if r_id not in self.wallets:
            raise WalletNotFoundError(r_id or receiver_id)

        if s_id == r_id:
            raise InvalidTransferAmountError("Sender and receiver wallet IDs cannot be the same.")

        sender = self.wallets[s_id]
        receiver = self.wallets[r_id]

        # 2. Validate sender PIN
        sender.verify_pin(pin)

        # 3. Validate active account status for both parties
        if not sender.is_active:
            raise InactiveWalletError(sender.wallet_id, sender.owner_name)
        if not receiver.is_active:
            raise InactiveWalletError(receiver.wallet_id, receiver.owner_name)

        # 4. Validate transfer amount format and positivity
        if isinstance(amount, bool) or not isinstance(amount, (int, float)):
            raise InvalidTransferAmountError("Amount must be a valid number.")
        if math.isnan(amount) or math.isinf(amount) or amount <= 0:
            raise InvalidTransferAmountError(f"Amount must be greater than Rs. 0 (received Rs. {amount}).")

        # 5. Enforce single-transaction and daily transfer limits
        if amount > sender.single_tx_limit:
            raise TransferLimitExceededError(amount, sender.single_tx_limit, "single-transaction")
        if sender.daily_transferred + amount > sender.daily_limit:
            remaining_daily = sender.daily_limit - sender.daily_transferred
            raise TransferLimitExceededError(
                amount,
                remaining_daily,
                "remaining daily",
            )

        # 6. Check sufficient balance
        if amount > sender.balance:
            raise InsufficientBalanceError(balance=sender.balance, required=amount)

        # 7. Atomic commit — runs ONLY after all validations pass
        transfer_amount = float(amount)
        sender.balance -= transfer_amount
        sender.daily_transferred += transfer_amount
        receiver.balance += transfer_amount

        self._tx_counter += 1
        receipt = TransferReceipt(
            tx_id=f"TXN-{self._tx_counter}",
            timestamp=datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
            sender_id=sender.wallet_id,
            receiver_id=receiver.wallet_id,
            amount=transfer_amount,
            sender_balance_after=sender.balance,
        )
        sender.history.append(receipt)
        receiver.history.append(receipt)
        return receipt

    def print_balances(self) -> None:
        print("\n" + "=" * 64)
        print(f"{'Wallet ID':<12} {'Owner':<20} {'Status':<10} {'Balance (Rs.)':>18}")
        print("-" * 64)
        for w in self.wallets.values():
            status = "Active" if w.is_active else "Inactive"
            print(f"{w.wallet_id:<12} {w.owner_name:<20} {status:<10} {w.balance:>18,.2f}")
        print("=" * 64)


def run_test_transfer(
    network: WalletNetwork,
    sender_id: str,
    pin: str,
    receiver_id: str,
    amount: float,
) -> None:
    """Executes a transfer attempt and prints clean diagnostic output."""
    print(f"\nTransfer: Rs. {amount:,.2f} ({sender_id} -> {receiver_id})")
    try:
        receipt = network.transfer(sender_id, pin, receiver_id, amount)
    except WalletNotFoundError as err:
        print(f"-> Failed [Wallet Error]   : {err}")
    except InvalidPinError as err:
        print(f"-> Failed [Security Error] : {err}")
    except InactiveWalletError as err:
        print(f"-> Failed [Account Error]  : {err}")
    except InvalidTransferAmountError as err:
        print(f"-> Failed [Amount Error]   : {err}")
    except TransferLimitExceededError as err:
        print(f"-> Failed [Limit Error]    : {err}")
    except InsufficientBalanceError as err:
        print(f"-> Failed [Balance Error]  : {err}")
    except WalletSystemError as err:
        print(f"-> Failed [System Error]   : {err}")
    else:
        print(
            f"-> Transfer successful! (Ref: {receipt.tx_id} | "
            f"New {receipt.sender_id} Balance: Rs. {receipt.sender_balance_after:,.2f})"
        )


if __name__ == "__main__":
    network = WalletNetwork()
    network.register_wallet(DigitalWallet("W-101", "Hamza Tariq", pin="4829", balance=25_000.00, single_tx_limit=15_000.00))
    network.register_wallet(DigitalWallet("W-102", "Ayesha Noor", pin="1937", balance=8_000.00))
    network.register_wallet(DigitalWallet("W-103", "Bilal Qureshi", pin="7741", balance=12_000.00, is_active=False))

    print("Initial Wallet Balances:")
    network.print_balances()

    # 1. Valid transfer matching prompt example (Rs. 2,000)
    run_test_transfer(network, "W-101", "4829", "W-102", 2_000.00)

    # 2. Failed transfers — balances must NOT change
    run_test_transfer(network, "W-101", "0000", "W-102", 1_000.00)   # InvalidPinError
    run_test_transfer(network, "W-101", "4829", "W-103", 1_500.00)   # InactiveWalletError (receiver inactive)
    run_test_transfer(network, "W-101", "4829", "W-102", -500.00)    # InvalidTransferAmountError
    run_test_transfer(network, "W-101", "4829", "W-102", 20_000.00)  # TransferLimitExceededError (> 15,000 limit)
    run_test_transfer(network, "W-102", "1937", "W-101", 50_000.00)  # InsufficientBalanceError (balance is 10,000)
    run_test_transfer(network, "W-999", "4829", "W-102", 500.00)     # WalletNotFoundError

    print("\nFinal Wallet Balances (Only the valid Rs. 2,000 transfer applied):")
    network.print_balances()
```

## ### Retail Inventory and Sales System

### Task:
Create a menu-driven inventory and sales application.

The program should allow the user to:
- Add stock
- View inventory
- Sell products
- Search for products

Handle invalid input, missing products, invalid quantities, and insufficient stock. Use `raise`, custom exceptions, `else`, and `finally` where appropriate.

The application should continue running after recoverable errors.

```python
from datetime import datetime


class InventoryError(Exception):
    pass


class ProductNotFoundError(InventoryError):
    pass


class InvalidQuantityError(InventoryError):
    pass


class InsufficientStockError(InventoryError):
    pass


class RetailStore:
    def __init__(self):
        self.inventory = {
            "P-101": {"name": "Wireless Mouse", "price": 3200.0, "stock": 15},
            "P-102": {"name": "Mechanical Keyboard", "price": 11500.0, "stock": 8},
            "P-103": {"name": "USB-C Hub", "price": 5800.0, "stock": 4},
        }
        self.total_sales = 0.0

    def parse_qty(self, raw_qty):
        try:
            qty = int(raw_qty.strip())
        except ValueError:
            raise InvalidQuantityError(f"'{raw_qty}' valid number nahi hai.")
        if qty <= 0:
            raise InvalidQuantityError("Quantity 1 ya us se zyada honi chahiye.")
        return qty

    def add_stock(self):
        sku = input("Product SKU (e.g. P-101): ").strip().upper()
        if not sku:
            raise InventoryError("SKU khali nahi ho sakta.")
        qty = self.parse_qty(input("Quantity to add: "))

        if sku in self.inventory:
            self.inventory[sku]["stock"] += qty
            print(f"Stock updated! {self.inventory[sku]['name']}: {self.inventory[sku]['stock']} units.")
        else:
            name = input("New Product Name: ").strip()
            if not name:
                raise InventoryError("Product name khali nahi ho sakta.")
            try:
                price = float(input("Unit Price (PKR): ").strip())
            except ValueError:
                raise InventoryError("Price valid number honi chahiye.")
            if price <= 0:
                raise InventoryError("Price 0 se zyada honi chahiye.")
            self.inventory[sku] = {"name": name, "price": price, "stock": qty}
            print(f"New product '{name}' ({sku}) added!")

    def view_inventory(self):
        print("\n--- Current Inventory ---")
        print(f"{'SKU':<8} {'Product':<22} {'Price':>10} {'Stock':>7}")
        for sku, item in self.inventory.items():
            print(f"{sku:<8} {item['name']:<22} {item['price']:>10,.2f} {item['stock']:>7}")

    def sell_product(self):
        sku = input("Product SKU to sell: ").strip().upper()
        raw_qty = input("Quantity to sell: ")
        try:
            if sku not in self.inventory:
                raise ProductNotFoundError(f"Product '{sku}' inventory mein nahi hai.")
            qty = self.parse_qty(raw_qty)
            item = self.inventory[sku]
            if qty > item["stock"]:
                raise InsufficientStockError(f"Sirf {item['stock']} units available hain.")
            item["stock"] -= qty
            bill = item["price"] * qty
            self.total_sales += bill
        except (ProductNotFoundError, InvalidQuantityError, InsufficientStockError) as err:
            print(f"Sale Failed: {err}")
        else:
            time_now = datetime.now().strftime("%H:%M:%S")
            print(f"Sale Successful [{time_now}]! {qty}x {item['name']} = PKR {bill:,.2f}")
        finally:
            print("[Status] Sale step complete. Returning to menu.")

    def search_product(self):
        term = input("Search SKU ya Name: ").strip().lower()
        if not term:
            raise InventoryError("Search keyword khali nahi ho sakta.")
        found = [
            (sku, item) for sku, item in self.inventory.items()
            if term in sku.lower() or term in item["name"].lower()
        ]
        if not found:
            raise ProductNotFoundError(f"'{term}' se match karta koi product nahi mila.")
        for sku, item in found:
            print(f"  [{sku}] {item['name']} - PKR {item['price']:,.2f} (Stock: {item['stock']})")


def main():
    store = RetailStore()
    while True:
        print("\n1. Add Stock | 2. View Inventory | 3. Sell | 4. Search | 5. Exit")
        choice = input("Option (1-5): ").strip()
        try:
            if choice == "1":
                store.add_stock()
            elif choice == "2":
                store.view_inventory()
            elif choice == "3":
                store.sell_product()
            elif choice == "4":
                store.search_product()
            elif choice == "5":
                print(f"Total Session Sales: PKR {store.total_sales:,.2f}. Goodbye!")
                break
            else:
                raise InventoryError("Sirf 1 se 5 tak option select karein.")
        except InventoryError as err:
            print(f"Error: {err}")


main()
```

## ### Food Delivery Application

### Goal:
Create a complete console-based food delivery application using exception handling.

### Steps:
- Register customers.
- Select a restaurant.
- Display menu items.
- Take item quantities.
- Validate stock.
- Validate minimum order amount.
- Validate delivery area.
- Process payment.
- Create custom exception classes for business-rule failures.
- Use `try`, specific `except` blocks, `else`, `finally`, and `raise`.
- Prevent invalid operations from partially changing important application data.

### Example:
Customer: Ali
Restaurant: ABC Foods
Item: Burger
Quantity: 2
→ Order placed successfully

```python
class DeliveryError(Exception):
    pass


class CustomerError(DeliveryError):
    pass


class RestaurantNotFoundError(DeliveryError):
    pass


class MenuItemNotFoundError(DeliveryError):
    pass


class InvalidQuantityError(DeliveryError):
    pass


class OutOfStockError(DeliveryError):
    pass


class MinimumOrderError(DeliveryError):
    pass


class DeliveryAreaError(DeliveryError):
    pass


class PaymentFailedError(DeliveryError):
    pass


class FoodDeliveryApp:
    def __init__(self):
        self.customers = {}
        self.valid_areas = {"pechs", "gulshan", "dha", "clifton", "nazimabad"}
        self.restaurants = {
            "ABC Foods": {
                "min_order": 500.0,
                "menu": {
                    "Burger": {"price": 450.0, "stock": 10},
                    "Pizza": {"price": 1200.0, "stock": 5},
                    "Fries": {"price": 200.0, "stock": 15},
                },
            },
            "Karachi Biryani": {
                "min_order": 400.0,
                "menu": {
                    "Chicken Biryani": {"price": 350.0, "stock": 12},
                    "Beef Pulao": {"price": 450.0, "stock": 6},
                    "Raita": {"price": 80.0, "stock": 20},
                },
            },
        }
        self.orders = []

    def register_customer(self):
        name = input("Customer Name: ").strip()
        if not name or not any(c.isalpha() for c in name):
            raise CustomerError("Valid customer name enter karein.")
        area = input(f"Delivery Area ({', '.join(sorted(self.valid_areas))}): ").strip()
        if area.lower() not in self.valid_areas:
            raise DeliveryAreaError(f"Sorry, '{area}' mein delivery available nahi hai.")
        self.customers[name.lower()] = {"name": name, "area": area.upper()}
        print(f"Customer '{name}' registered successfully!")

    def find_key(self, mapping, user_text, exc_cls, err_msg):
        for key in mapping:
            if key.lower() == user_text.strip().lower():
                return key
        raise exc_cls(err_msg)

    def place_order(self):
        try:
            # 1. Validate Customer & Area
            cust_input = input("Customer Name: ").strip()
            cust_key = self.find_key(
                self.customers, cust_input, CustomerError, f"Customer '{cust_input}' registered nahi hai."
            )
            customer = self.customers[cust_key]

            # 2. Select Restaurant & Show Menu
            print("Restaurants:", ", ".join(self.restaurants.keys()))
            rest_input = input("Select Restaurant: ").strip()
            rest_name = self.find_key(
                self.restaurants, rest_input, RestaurantNotFoundError, f"Restaurant '{rest_input}' nahi mila."
            )
            rest = self.restaurants[rest_name]

            print(f"\n--- {rest_name} Menu (Min Order: Rs. {rest['min_order']:,.0f}) ---")
            for item, info in rest["menu"].items():
                print(f"  {item:<18} Rs. {info['price']:>6,.0f} (Stock: {info['stock']})")

            # 3. Take Cart Items (Atomic staging—stock abhi deduct nahi hoga)
            cart = {}
            while True:
                item_in = input("\nItem name (ya 'done' likhein): ").strip()
                if item_in.lower() == "done":
                    break
                item_name = self.find_key(
                    rest["menu"], item_in, MenuItemNotFoundError, f"'{item_in}' menu mein nahi hai."
                )
                try:
                    qty = int(input(f"Quantity for {item_name}: ").strip())
                except ValueError:
                    raise InvalidQuantityError("Quantity valid number honi chahiye.")
                if qty <= 0:
                    raise InvalidQuantityError("Quantity 1 ya zyada honi chahiye.")

                total_qty = cart.get(item_name, 0) + qty
                if total_qty > rest["menu"][item_name]["stock"]:
                    raise OutOfStockError(f"'{item_name}' ka sirf {rest['menu'][item_name]['stock']} stock hai.")
                cart[item_name] = total_qty

            if not cart:
                raise InvalidQuantityError("Order mein kam az kam 1 item hona chahiye.")

            # 4. Validate Minimum Order Amount
            total_bill = sum(rest["menu"][i]["price"] * q for i, q in cart.items())
            if total_bill < rest["min_order"]:
                raise MinimumOrderError(
                    f"Total Rs. {total_bill:,.0f} hai; minimum order Rs. {rest['min_order']:,.0f} zaroori hai."
                )

            # 5. Process Payment
            print(f"\nTotal Bill: Rs. {total_bill:,.2f}")
            method = input("Payment Method (cash/card/raast): ").strip().lower()
            if method not in {"cash", "card", "raast"}:
                raise PaymentFailedError(f"Unsupported payment method: '{method}'.")
            try:
                paid = float(input("Enter Payment Amount (Rs.): ").strip())
            except ValueError:
                raise PaymentFailedError("Payment amount valid number honi chahiye.")
            if paid < total_bill:
                raise PaymentFailedError(f"Payment kam hai! Rs. {total_bill:,.2f} zaroori hain.")

            # 6. Atomic Commit (Sab checks pass hone ke baad hi stock kam hoga)
            for item_name, qty in cart.items():
                rest["menu"][item_name]["stock"] -= qty
            self.orders.append((customer["name"], rest_name, cart, total_bill))

        except (
            CustomerError,
            DeliveryAreaError,
            RestaurantNotFoundError,
            MenuItemNotFoundError,
            InvalidQuantityError,
            OutOfStockError,
            MinimumOrderError,
            PaymentFailedError,
        ) as err:
            print(f"\n-> Order Failed: {err}")
        else:
            change = paid - total_bill
            print(f"\n-> Order placed successfully for {customer['name']} from {rest_name}!")
            print(f"   Paid: Rs. {paid:,.2f} | Change: Rs. {change:,.2f}")
        finally:
            print("[System] Order session finished. Stock & data safe.")


def main():
    app = FoodDeliveryApp()
    while True:
        print("\n1. Register Customer | 2. Place Order | 3. View Orders | 4. Exit")
        choice = input("Choice (1-4): ").strip()
        try:
            if choice == "1":
                app.register_customer()
            elif choice == "2":
                app.place_order()
            elif choice == "3":
                print("\nCompleted Orders:", len(app.orders))
                for idx, (cust, rest, items, amt) in enumerate(app.orders, 1):
                    print(f"  {idx}. {cust} | {rest} | {items} | Rs. {amt:,.2f}")
            elif choice == "4":
                print("Thank you for using Food Delivery App!")
                break
            else:
                raise DeliveryError("1 se 4 ke darmiyan option select karein.")
        except DeliveryError as err:
            print(f"Error: {err}")


main()
```
