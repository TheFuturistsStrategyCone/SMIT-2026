### Q7. ATM Withdrawal — Basic

A simple ATM asks a customer to enter a withdrawal amount. The amount must be numeric, positive, and no greater than the available balance.

Set an example account balance in a variable. Ask for the withdrawal amount, handle invalid input, and raise an appropriate built-in exception for a non-positive amount or an amount greater than the balance. Deduct money only when all checks pass.

```python
def atm_withdrawal() -> None:
    """Simulates a basic ATM withdrawal with input and balance validation."""
    balance: float = 50000.00  # Example starting balance (e.g., PKR 50,000)
    print(f"Current Balance: PKR {balance:,.2f}")

    try:
        raw_amount = input("Enter withdrawal amount: ").strip()
        amount = float(raw_amount)

        if amount <= 0:
            raise ValueError("Withdrawal amount must be greater than zero.")

        if amount > balance:
            raise OverflowError("Insufficient funds: Withdrawal amount exceeds available balance.")

        balance -= amount
        print(f"Withdrawal successful: PKR {amount:,.2f}")
        print(f"Remaining Balance:     PKR {balance:,.2f}")

    except OverflowError as e:
        print(f"Transaction declined: {e}")
    except ValueError as e:
        if "could not convert string to float" in str(e):
            print("Invalid input: Please enter a valid numeric amount.")
        else:
            print(f"Transaction declined: {e}")


if __name__ == "__main__":
    atm_withdrawal()
```

### Q14. Weak Password Rejection

A website requires passwords to contain at least 8 characters, including at least one digit. A password that does not meet the rules must be rejected.

Create a custom exception named `WeakPasswordError`. Ask the user for a password, check both rules, raise the custom exception when a rule fails, and handle it to explain what needs to be corrected. Do not display the actual password in the success message.

```python
class WeakPasswordError(Exception):
    """Custom exception raised when a password does not meet security rules."""


def validate_password() -> None:
    """Validates password length and digit requirements without revealing the password."""
    try:
        password = input("Create a new password: ")
        errors: list[str] = []

        if len(password) < 8:
            errors.append("must be at least 8 characters long")

        if not any(char.isdigit() for char in password):
            errors.append("must contain at least one digit (0-9)")

        if errors:
            raise WeakPasswordError(
                f"Password is too weak: it {' and '.join(errors)}."
            )

        print("Success: Password accepted and saved securely.")

    except WeakPasswordError as e:
        print(f"Registration Failed: {e}")


if __name__ == "__main__":
    validate_password()
```

### Q16. Hotel Room Booking

A hotel has a limited number of rooms. A booking request must be for at least one room and cannot exceed the rooms currently available.

Create two custom exceptions: `InvalidRoomCountError` and `RoomsUnavailableError`. Validate the requested room count, raise the appropriate exception, handle each exception separately, and reduce the available room count only after a successful booking.

```python
class InvalidRoomCountError(Exception):
    """Custom exception raised when the requested room count is zero or negative."""


class RoomsUnavailableError(Exception):
    """Custom exception raised when the requested rooms exceed available rooms."""


def book_hotel_rooms() -> None:
    """Validates a hotel room booking request and updates available room inventory."""
    available_rooms: int = 8  # Sample available rooms in the hotel
    print(f"Rooms Currently Available: {available_rooms}")

    try:
        raw_rooms = input("Enter number of rooms to book: ").strip()
        requested_rooms = int(raw_rooms)

        if requested_rooms <= 0:
            raise InvalidRoomCountError(
                "Room count must be at least 1."
            )

        if requested_rooms > available_rooms:
            raise RoomsUnavailableError(
                f"Only {available_rooms} room(s) available; cannot book {requested_rooms}."
            )

        available_rooms -= requested_rooms
        print(f"Booking Confirmed: {requested_rooms} room(s) reserved.")
        print(f"Remaining Rooms:   {available_rooms}")

    except ValueError:
        print("Invalid Input: Please enter a whole number for the room count.")
    except InvalidRoomCountError as e:
        print(f"Booking Error: {e}")
    except RoomsUnavailableError as e:
        print(f"Availability Error: {e}")


if __name__ == "__main__":
    book_hotel_rooms()
```

### Q18. File Reading Safely

A report application asks the user for the name of a text file and tries to read its contents.

Use a `try` block to open and read the requested file. Handle `FileNotFoundError` and `PermissionError` with different messages. Use `finally` to display a message that the file-reading attempt has finished, whether it succeeds or fails. Do not assume that a particular file exists.

```python
def read_file_safely() -> None:
    """Prompts for a filename, reads its contents safely, and handles file errors."""
    filename = input("Enter the name of the text file to read: ").strip()

    try:
        if not filename:
            raise FileNotFoundError("No filename was entered.")

        with open(filename, mode="r", encoding="utf-8") as file:
            content = file.read()

        print("\n--- File Contents ---")
        print(content if content else "[The file is empty]")

    except FileNotFoundError:
        print(f"File Not Found: Could not locate '{filename}'. Please check the file name and path.")
    except PermissionError:
        print(f"Permission Denied: You do not have permission to read '{filename}'.")
    except IsADirectoryError:
        print(f"Invalid Target: '{filename}' is a folder/directory, not a text file.")
    finally:
        print("\nFile-reading attempt has finished.")


if __name__ == "__main__":
    read_file_safely()
```

### Q20. Library Borrowing Rules

A library member may borrow up to three books at a time. A member who already has three borrowed books cannot borrow another one. The requested number of books must be a positive integer.

Create custom exceptions named `BorrowLimitExceededError` and `InvalidBookCountError`. Validate the request, raise the correct exception, handle each case separately, and update the member's borrowed-book count only after a valid request.

```python
MAX_BORROW_LIMIT = 3


class InvalidBookCountError(Exception):
    """Custom exception raised when the requested book count is zero or negative."""


class BorrowLimitExceededError(Exception):
    """Custom exception raised when borrowing exceeds the maximum allowed limit."""


def borrow_library_books() -> None:
    """Validates a book borrowing request and updates the member's borrowed count."""
    currently_borrowed: int = 1  # Sample number of books already borrowed by the member
    remaining_allowance = MAX_BORROW_LIMIT - currently_borrowed

    print(f"Books Currently Borrowed: {currently_borrowed} / {MAX_BORROW_LIMIT}")

    try:
        raw_count = input("Enter number of books to borrow: ").strip()
        requested_books = int(raw_count)

        if requested_books <= 0:
            raise InvalidBookCountError(
                "Requested book count must be a positive whole number (at least 1)."
            )

        if currently_borrowed + requested_books > MAX_BORROW_LIMIT:
            raise BorrowLimitExceededError(
                f"Borrow limit exceeded: You already have {currently_borrowed} book(s) "
                f"and can only borrow up to {remaining_allowance} more."
            )

        currently_borrowed += requested_books
        print(f"Request Approved: {requested_books} book(s) issued.")
        print(f"Updated Borrowed Count: {currently_borrowed} / {MAX_BORROW_LIMIT}")

    except ValueError:
        print("Invalid Input: Please enter a whole number (no text or decimals).")
    except InvalidBookCountError as e:
        print(f"Invalid Request: {e}")
    except BorrowLimitExceededError as e:
        print(f"Limit Error: {e}")


if __name__ == "__main__":
    borrow_library_books()
```

### Q21. ATM with Multiple Custom Errors

Build a small ATM withdrawal program. The withdrawal must be greater than zero, must not exceed the account balance, and must not exceed a daily withdrawal limit.

Create custom exception classes named `InvalidAmountError`, `InsufficientBalanceError`, and `DailyLimitExceededError`. Ask for a withdrawal amount and validate each rule. Raise the relevant custom exception and handle each one with a different message. Update the balance and daily total only after all validations pass.

```python
DAILY_WITHDRAWAL_LIMIT = 50000.00  # Maximum daily withdrawal limit (PKR)


class InvalidAmountError(Exception):
    """Custom exception raised when the withdrawal amount is zero or negative."""


class InsufficientBalanceError(Exception):
    """Custom exception raised when the withdrawal amount exceeds account balance."""


class DailyLimitExceededError(Exception):
    """Custom exception raised when the withdrawal breaches the daily limit."""


def atm_withdrawal_system() -> None:
    """Validates an ATM withdrawal against amount, balance, and daily limit rules."""
    account_balance: float = 120000.00     # Sample current account balance
    withdrawn_today: float = 15000.00     # Sample amount already withdrawn today
    remaining_daily_limit = DAILY_WITHDRAWAL_LIMIT - withdrawn_today

    print("--- ATM Account Status ---")
    print(f"Available Balance:      PKR {account_balance:,.2f}")
    print(f"Withdrawn Today:        PKR {withdrawn_today:,.2f}")
    print(f"Remaining Daily Limit:  PKR {remaining_daily_limit:,.2f}\n")

    try:
        raw_amount = input("Enter withdrawal amount (PKR): ").strip()
        amount = float(raw_amount)

        if amount <= 0:
            raise InvalidAmountError(
                "Withdrawal amount must be greater than zero."
            )

        if amount > account_balance:
            raise InsufficientBalanceError(
                f"Requested PKR {amount:,.2f}, but available balance is PKR {account_balance:,.2f}."
            )

        if withdrawn_today + amount > DAILY_WITHDRAWAL_LIMIT:
            raise DailyLimitExceededError(
                f"Exceeds daily limit of PKR {DAILY_WITHDRAWAL_LIMIT:,.2f}. "
                f"You can withdraw up to PKR {remaining_daily_limit:,.2f} more today."
            )

        # Update state only after all validations pass
        account_balance -= amount
        withdrawn_today += amount

        print("\n--- Transaction Successful ---")
        print(f"Dispensed Amount:       PKR {amount:,.2f}")
        print(f"Updated Balance:        PKR {account_balance:,.2f}")
        print(f"Updated Daily Total:    PKR {withdrawn_today:,.2f}")

    except ValueError:
        print("Input Error: Please enter a valid numeric amount.")
    except InvalidAmountError as e:
        print(f"Invalid Amount: {e}")
    except InsufficientBalanceError as e:
        print(f"Balance Error: {e}")
    except DailyLimitExceededError as e:
        print(f"Daily Limit Error: {e}")


if __name__ == "__main__":
    atm_withdrawal_system()
```

```python
class InvalidTransferError(Exception):
    """Custom exception raised for invalid accounts, self-transfers, or non-positive amounts."""


class InsufficientFundsError(Exception):
    """Custom exception raised when the sender lacks sufficient balance for the transfer."""


def bank_transfer() -> None:
    """Validates and executes a fund transfer between two distinct bank accounts."""
    accounts: dict[str, float] = {
        "ACC-1001": 75000.00,
        "ACC-1002": 30000.00,
    }

    print("--- Initial Account Balances ---")
    for acc_id, bal in accounts.items():
        print(f"{acc_id}: PKR {bal:,.2f}")
    print()

    sender_id = input("Enter sender account ID: ").strip().upper()
    receiver_id = input("Enter receiver account ID: ").strip().upper()

    try:
        if sender_id not in accounts:
            raise InvalidTransferError(f"Sender account '{sender_id}' does not exist.")

        if receiver_id not in accounts:
            raise InvalidTransferError(f"Receiver account '{receiver_id}' does not exist.")

        if sender_id == receiver_id:
            raise InvalidTransferError("Sender and receiver accounts must be different.")

        raw_amount = input("Enter transfer amount (PKR): ").strip()
        amount = float(raw_amount)

        if amount <= 0:
            raise InvalidTransferError("Transfer amount must be greater than zero.")

        if amount > accounts[sender_id]:
            raise InsufficientFundsError(
                f"Account '{sender_id}' has PKR {accounts[sender_id]:,.2f}, "
                f"which is insufficient to transfer PKR {amount:,.2f}."
            )

        # Atomic update: executes only after every validation succeeds
        accounts[sender_id] -= amount
        accounts[receiver_id] += amount
        print(f"\nSuccess: Transferred PKR {amount:,.2f} from {sender_id} to {receiver_id}.")

    except ValueError:
        print("\nInput Error: Please enter a valid numeric transfer amount.")
    except InvalidTransferError as e:
        print(f"\nTransfer Error: {e}")
    except InsufficientFundsError as e:
        print(f"\nFunds Error: {e}")
    finally:
        print("\n--- Account Balances After Attempt ---")
        for acc_id, bal in accounts.items():
            print(f"{acc_id}: PKR {bal:,.2f}")


if __name__ == "__main__":
    bank_transfer()
```

### Q28. Bulk Employee Payroll Processing

A company stores several employee records in a list. Each record contains an employee name, hours worked, and hourly rate. One invalid record should not stop the payroll calculation for all other employees.

Create a custom exception named `InvalidPayrollRecordError`. Validate each employee record, raise this exception when required data is missing or values are invalid, and handle the error inside the record-processing loop. Calculate payroll for valid employees and report which records were skipped and why.

```python
MAX_WEEKLY_HOURS = 80.0


class InvalidPayrollRecordError(Exception):
    """Custom exception raised when an employee payroll record has missing or invalid data."""


def process_bulk_payroll(records: list[dict]) -> None:
    """Processes a batch of employee records, isolating invalid ones without halting execution."""
    processed_payroll: list[dict] = []
    skipped_records: list[tuple[int, str, str]] = []

    for index, record in enumerate(records, start=1):
        emp_name = str(record.get("name", "")).strip() or f"Record #{index}"

        try:
            # 1. Validate employee name
            if not str(record.get("name", "")).strip():
                raise InvalidPayrollRecordError("Missing or blank employee name.")

            # 2. Validate presence and numeric conversion of hours & rate
            if "hours" not in record or record["hours"] is None:
                raise InvalidPayrollRecordError("Missing working hours.")
            if "rate" not in record or record["rate"] is None:
                raise InvalidPayrollRecordError("Missing hourly rate.")

            try:
                hours = float(record["hours"])
                rate = float(record["rate"])
            except (ValueError, TypeError) as exc:
                raise InvalidPayrollRecordError(
                    "Hours and hourly rate must be valid numbers."
                ) from exc

            # 3. Validate business rules
            if not (0 <= hours <= MAX_WEEKLY_HOURS):
                raise InvalidPayrollRecordError(
                    f"Hours ({hours:g}) must be between 0 and {int(MAX_WEEKLY_HOURS)}."
                )

            if rate <= 0:
                raise InvalidPayrollRecordError(
                    f"Hourly rate ({rate:g}) must be greater than zero."
                )

            gross_pay = hours * rate
            processed_payroll.append({
                "name": emp_name,
                "hours": hours,
                "rate": rate,
                "gross_pay": gross_pay,
            })

        except InvalidPayrollRecordError as e:
            skipped_records.append((index, emp_name, str(e)))

    # Summary Report
    print("=== PROCESSED PAYROLL ===")
    if processed_payroll:
        for entry in processed_payroll:
            print(
                f"- {entry['name']}: {entry['hours']:g} hrs @ PKR {entry['rate']:,.2f}/hr "
                f"-> Gross Pay: PKR {entry['gross_pay']:,.2f}"
            )
        total_payout = sum(item["gross_pay"] for item in processed_payroll)
        print(f"Total Valid Payout: PKR {total_payout:,.2f}")
    else:
        print("No valid employee records were processed.")

    print("\n=== SKIPPED RECORDS REPORT ===")
    if skipped_records:
        for rec_num, name, reason in skipped_records:
            print(f"- Row {rec_num} ({name}): Skipped -> {reason}")
    else:
        print("All records were valid!")


if __name__ == "__main__":
    sample_records = [
        {"name": "Ayesha Siddiqui", "hours": 40, "rate": 1500},       # Valid
        {"name": "", "hours": 35, "rate": 1200},                      # Invalid: Missing name
        {"name": "Bilal Ahmed", "hours": -5, "rate": 1400},           # Invalid: Negative hours
        {"name": "Zainab Raza", "hours": "forty", "rate": 1600},      # Invalid: Non-numeric hours
        {"name": "Usman Tariq", "hours": 45, "rate": 0},              # Invalid: Zero hourly rate
        {"name": "Fatima Noor", "hours": 38.5, "rate": 1800},         # Valid
    ]

    process_bulk_payroll(sample_records)
```

### Q29. Inventory Purchase with Safe Transaction

A warehouse purchase operation receives a product ID and requested quantity. The product must exist, the quantity must be positive, and enough stock must be available. A failed purchase must leave inventory unchanged.

Create custom exceptions named `ProductNotFoundError`, `InvalidQuantityError`, and `InsufficientStockError`. Implement validation before changing stock. Use `try`, `except`, `else`, and `finally` meaningfully: report success only in `else`, handle expected failures in `except`, and display an audit/attempt-finished message in `finally`. Demonstrate that stock remains unchanged after a failed purchase.

```python
class ProductNotFoundError(Exception):
    """Custom exception raised when the requested product ID does not exist in inventory."""


class InvalidQuantityError(Exception):
    """Custom exception raised when the requested quantity is zero or negative."""


class InsufficientStockError(Exception):
    """Custom exception raised when requested units exceed available warehouse stock."""


def process_purchase(
    inventory: dict[str, dict],
    product_id: str,
    raw_quantity: str,
) -> None:
    """Validates and executes a warehouse purchase transaction safely."""
    clean_id = product_id.strip().upper()
    print(f"\n>>> Purchase Request | Product: '{clean_id}' | Quantity: '{raw_quantity}'")

    try:
        # 1. Check if product exists
        if clean_id not in inventory:
            raise ProductNotFoundError(
                f"Product ID '{clean_id}' was not found in the warehouse catalogue."
            )

        # 2. Convert and validate positive integer quantity
        try:
            quantity = int(raw_quantity.strip())
        except (ValueError, AttributeError) as exc:
            raise InvalidQuantityError(
                "Requested quantity must be a valid whole number."
            ) from exc

        if quantity <= 0:
            raise InvalidQuantityError(
                f"Requested quantity ({quantity}) must be greater than zero."
            )

        # 3. Check available stock
        available_stock = inventory[clean_id]["stock"]
        if quantity > available_stock:
            raise InsufficientStockError(
                f"Insufficient stock for '{inventory[clean_id]['name']}': "
                f"requested {quantity}, but only {available_stock} unit(s) available."
            )

        # 4. Update stock only after all validations succeed
        inventory[clean_id]["stock"] -= quantity
        total_cost = quantity * inventory[clean_id]["price"]

    except ProductNotFoundError as e:
        print(f"[FAILED - Product Error] {e}")
    except InvalidQuantityError as e:
        print(f"[FAILED - Quantity Error] {e}")
    except InsufficientStockError as e:
        print(f"[FAILED - Stock Error] {e}")
    else:
        # Runs ONLY when the try block completes without any exception
        print(
            f"[SUCCESS] Purchased {quantity} x '{inventory[clean_id]['name']}' "
            f"(Total: PKR {total_cost:,.2f}). Remaining stock: {inventory[clean_id]['stock']}."
        )
    finally:
        # Always runs for audit logging whether the transaction succeeded or failed
        if clean_id in inventory:
            print(f"[AUDIT] Attempt finished. Current '{clean_id}' stock: {inventory[clean_id]['stock']} unit(s).")
        else:
            print("[AUDIT] Attempt finished. Inventory unchanged.")


if __name__ == "__main__":
    warehouse_inventory = {
        "PRD-101": {"name": "Wireless Barcode Scanner", "stock": 15, "price": 6500.00},
        "PRD-102": {"name": "Thermal Receipt Printer", "stock": 8, "price": 12000.00},
    }

    print("Initial Warehouse Inventory:", warehouse_inventory)

    # 1. Valid purchase (stock drops from 15 -> 10)
    process_purchase(warehouse_inventory, "PRD-101", "5")

    # 2. Failed purchase: Insufficient stock (stock must remain 10)
    process_purchase(warehouse_inventory, "PRD-101", "20")

    # 3. Failed purchase: Invalid quantity (stock must remain 8)
    process_purchase(warehouse_inventory, "PRD-102", "-2")

    # 4. Failed purchase: Product not found
    process_purchase(warehouse_inventory, "PRD-999", "3")

    print("\nFinal Warehouse Inventory:", warehouse_inventory)
```

Keep the program running after an invalid transaction, and show the final account balance when the user chooses to exit.

```python
class BankingError(Exception):
    """Base custom exception for all banking domain errors."""


class InvalidAmountError(BankingError):
    """Raised when a deposit or withdrawal amount is non-numeric, zero, or negative."""


class InsufficientBalanceError(BankingError):
    """Raised when a withdrawal amount exceeds the current account balance."""


class BankAccount:
    """Represents a simple bank account with deposit, withdrawal, and balance operations."""

    def __init__(self, account_holder: str, initial_balance: float = 0.0) -> None:
        self.account_holder = account_holder
        self.balance = initial_balance

    @staticmethod
    def _parse_positive_amount(raw_amount: str) -> float:
        """Converts input string to float and ensures it is strictly greater than zero."""
        try:
            amount = float(raw_amount.strip())
        except ValueError as exc:
            raise InvalidAmountError(
                "Please enter a valid numeric amount."
            ) from exc

        if amount <= 0:
            raise InvalidAmountError(
                "Amount must be greater than zero."
            )

        return amount

    def deposit(self, raw_amount: str) -> None:
        """Validates and adds a positive amount to the account balance."""
        amount = self._parse_positive_amount(raw_amount)
        self.balance += amount
        print(f"Success: Deposited PKR {amount:,.2f}.")
        print(f"Updated Balance: PKR {self.balance:,.2f}")

    def withdraw(self, raw_amount: str) -> None:
        """Validates amount and available funds before deducting from the account."""
        amount = self._parse_positive_amount(raw_amount)

        if amount > self.balance:
            raise InsufficientBalanceError(
                f"Cannot withdraw PKR {amount:,.2f}; available balance is PKR {self.balance:,.2f}."
            )

        self.balance -= amount
        print(f"Success: Withdrew PKR {amount:,.2f}.")
        print(f"Remaining Balance: PKR {self.balance:,.2f}")

    def check_balance(self) -> None:
        """Displays the current account balance."""
        print(f"Current Balance ({self.account_holder}): PKR {self.balance:,.2f}")


def run_banking_app() -> None:
    """Runs the interactive menu loop until the user chooses to exit."""
    account = BankAccount(account_holder="Ali Raza", initial_balance=25000.00)
    print(f"=== Welcome to Mini Banking System ({account.account_holder}) ===")

    while True:
        print("\nChoose an option:")
        print("1. Deposit Money")
        print("2. Withdraw Money")
        print("3. View Balance")
        print("4. Exit")

        choice = input("Enter your choice (1-4): ").strip()

        try:
            if choice == "1":
                raw_deposit = input("Enter deposit amount (PKR): ")
                account.deposit(raw_deposit)

            elif choice == "2":
                raw_withdraw = input("Enter withdrawal amount (PKR): ")
                account.withdraw(raw_withdraw)

            elif choice == "3":
                account.check_balance()

            elif choice == "4":
                print("\n--- Session Ended ---")
                print(f"Final Account Balance: PKR {account.balance:,.2f}")
                print("Thank you for banking with us!")
                break

            else:
                print("Invalid Choice: Please select an option from 1 to 4.")

        except InsufficientBalanceError as e:
            print(f"Withdrawal Declined: {e}")
        except InvalidAmountError as e:
            print(f"Invalid Input: {e}")
        except BankingError as e:
            # Fallback catch-all for any future child exceptions of BankingError
            print(f"Banking Error: {e}")


if __name__ == "__main__":
    run_banking_app()
```
