```python
# Ask the user to enter the current floor number and convert it to an integer
current_floor = int(input("Enter current floor: "))

# Use a while loop to count down as long as the floor number is greater than 0
while current_floor >= 1:
    # Print the current floor number
    print(current_floor)
    
    # Decrement the floor counter by 1 to simulate moving down
    current_floor -= 1

# Print the final message after exiting the loop
print("Ground floor reached!")
```

```python
# Ask the user for the total number of days and the daily saving amount
total_days = int(input("Enter number of days: "))
daily_amount = float(input("Enter daily saving amount: "))

# Initialize the counter for days and the accumulator for total savings
day_counter = 1
total_savings = 0.0

# Loop from day 1 up to N days
while day_counter <= total_days:
    # Add the daily saving amount to the running total
    total_savings += daily_amount
    
    # Move to the next day
    day_counter += 1

# Print the final total savings (formatted to handle decimal values cleanly)
print(f"Total savings after {total_days} days = {total_savings:g}")
```

```python
# Ask the user for the starting and ending seat numbers
start_seat = int(input("Start seat: "))
end_seat = int(input("End seat: "))

# Handle the case where the start seat is greater than the end seat
# by swapping them so the loop always moves from lowest to highest
if start_seat > end_seat:
    start_seat, end_seat = end_seat, start_seat

# Initialize the current seat pointer
current_seat = start_seat

# Loop through every number in the range (inclusive of the end seat)
while current_seat <= end_seat:
    # Check if the current seat number is even using the modulo operator (%)
    if current_seat % 2 == 0:
        print(current_seat)
    
    # Move to the next seat number
    current_seat += 1
```

```python
# Hard-code the correct password
CORRECT_PASSWORD = "secure@2024"

# Set the maximum allowed attempts and initialize variables
max_attempts = 3
attempts_used = 0
is_logged_in = False

# The loop continues while the user hasn't successfully logged in
# and they still have attempts remaining
while attempts_used < max_attempts and not is_logged_in:
    # Ask the user for their password password
    entered_password = input("Enter password: ")
    
    # Check if the entered password matches the correct one
    if entered_password == CORRECT_PASSWORD:
        is_logged_in = True
    else:
        # Increment the attempt counter for an incorrect guess
        attempts_used += 1
        attempts_remaining = max_attempts - attempts_used
        
        # Only print the remaining attempts if the account isn't about to be locked
        if attempts_remaining > 0:
            # Handle plural vs singular grammar for a polished look
            attempt_word = "attempt" if attempts_remaining == 1 else "attempts"
            print(f"Incorrect. {attempts_remaining} {attempt_word} remaining.")

# Final output based on the login status flag
if is_logged_in:
    print("Login successful. Welcome to your account.")
else:
    print("Account temporarily locked for security reasons.")
```

```python
# Read the tracking number as an integer from the user
tracking_number = int(input("Enter tracking number: "))

# Make a copy of the number to modify in the loop, keeping the original intact
temp_number = tracking_number

# Initialize a counter to track the number of digits
digit_count = 0

# Special case: If the input tracking number is 0, it has 1 digit
if temp_number == 0:
    digit_count = 1
else:
    # Use a while loop to peel off digits one by one
    while temp_number > 0:
        # Optional: capture the peeled-off digit using modulo if needed for logging
        # last_digit = temp_number % 10 
        
        # Use integer division (//) to remove the last digit
        temp_number = temp_number // 10
        
        # Increment the counter for the removed digit
        digit_count += 1

# Print the final total count
print(f"Number of digits in tracking ID: {digit_count}")
```

```python
# Read a positive integer serial number from the user
serial_number = int(input("Enter serial number: "))

# Initialize the variable to build the reversed number
reversed_serial = 0

# Process the number until all digits are extracted
while serial_number > 0:
    # Extract the last digit using the modulo operator
    last_digit = serial_number % 10
    
    # Append the digit to the reversed number by shifting existing digits left
    reversed_serial = (reversed_serial * 10) + last_digit
    
    # Remove the last digit from the original number using integer division
    serial_number = serial_number // 10

# Print the final mathematically reversed serial number
print(f"Reversed serial number: {reversed_serial}")
```

```python
# Ask the user for the feedback sentence and the target character
feedback = input("Enter feedback: ")
target_char = input("Enter character to count: ")

# Initialize the index pointer and the accumulator counter
index = 0
occurrence_count = 0

# Loop through the string one character at a time using its length
while index < len(feedback):
    # Check if the character at the current index matches the target character
    if feedback[index] == target_char:
        occurrence_count += 1
    
    # Move to the next index character position
    index += 1

# Print the final result matching the exact example format
print(f"The character '{target_char}' appears {occurrence_count} time(s).")
```

```python
# Ask the user for the maximum room side length
max_side = int(input("Enter maximum side length N: "))

# Initialize an empty list to accumulate the calculated areas
room_areas = []

# Start the counter at a side length of 1
current_side = 1

# Loop from 1 up to and including N
while current_side <= max_side:
    # Calculate the square footage (side * side)
    area = current_side * current_side
    
    # Append the calculated area to the list
    room_areas.append(area)
    
    # Increment the side length for the next iteration
    current_side += 1

# Print the final completed list of areas
print(room_areas)
```

```python
# Ask the user for the number of rows of lights
total_rows = int(input("Enter number of rows: "))

# Initialize the outer row counter at 1
current_row = 1

# Outer while loop controls the total number of rows
while current_row <= total_rows:
    # Inner loop prints the exact number of stars for the current row
    # Row 1 gets 1 star, Row 2 gets 2 stars, etc.
    for star in range(current_row):
        print("*", end="")
    
    # Move to a new line after printing all stars in the current row
    print()
    
    # Increment the outer row counter
    current_row += 1
```

```python
# Initialize the aisle counter at 1
aisle = 1

# Loop through all aisles up to 30
while aisle <= 30:
    # Check if the aisle is a multiple of 3 (under maintenance)
    if aisle % 3 == 0:
        # CRITICAL STEP: We must increment the counter BEFORE using continue
        # to prevent an infinite loop on the skipped number.
        aisle += 1
        continue
    
    # Print the aisle numbers that the robot must visit
    print(aisle)
    
    # Increment the counter for the standard paths
    aisle += 1
```

```python
# Ask the user for the bay size multiplier
bay_size = int(input("Enter bay size: "))

# Start checking numbers from 101 upward
spot_number = 101

# Create an infinite loop that runs until explicitly broken
while True:
    # Check if the current spot number is cleanly divisible by the bay size
    if spot_number % bay_size == 0:
        # Print the success message matching the example output
        print(f"First available spot multiple of {bay_size} after 100 is {spot_number}")
        
        # Stop searching immediately and terminate the loop
        break
        
    # Increment the spot number to check the next slot
    spot_number += 1
```

```python
import math

# Ask the user for an integer product ID greater than 1
product_id = int(input("Enter product ID: "))

# Initialize the divisor at 2
divisor = 2

# Calculate the upper test limit (square root of the number) to make the code efficient
# Testing up to the square root is mathematically sufficient to prove primality
test_limit = int(math.sqrt(product_id))

# Check divisibility starting from 2 up to the test limit
while divisor <= test_limit:
    # If the product ID is perfectly divisible by the current divisor, it's not prime
    if product_id % divisor == 0:
        print(f"{product_id} is not a prime ID.")
        break  # Exit the loop early because we found a factor
        
    divisor += 1
else:
    # The else block executes ONLY if the while loop completed fully without hitting a 'break'
    print(f"{product_id} is a prime ID (strong security code).")
```

```python
# Initialize the outer counter for periods (1 to 5)
period = 1

# Outer while loop controls the periods
while period <= 5:
    # Check if it is the assembly period
    if period == 3:
        # CRITICAL: Increment the counter before using continue
        # to prevent an infinite loop on period 3
        period += 1
        continue
    
    # Optional header to visually divide the periods clearly
    print(f"--- Timetable for Period {period} ---")
    
    # Inner loop prints slots 1 through 10 for the active period
    for slot in range(1, 11):
        result = period * slot
        print(f"Period {period} × Slot {slot} = {result}")
    
    # Print an empty line between period blocks for clean formatting
    print()
    
    # Increment the outer loop counter for the regular path
    period += 1
```

```python
### Initialize the hard-coded starting balance
account_balance = 5000

# Create an infinite loop to keep displaying the menu until the user exits
while True:
    # Print the interactive banking menu
    print("\n1. Check Balance")
    print("2. Withdraw")
    print("3. Exit")
    
    # Capture the user's choice
    choice = input("Choose: ")
    
    # Process Option 1: Check Balance
    if choice == "1":
        print(f"Current balance: {account_balance}")
        
    # Process Option 2: Withdraw Money
    elif choice == "2":
        amount_to_withdraw = int(input("Amount to withdraw: "))
        
        # Check if the user has enough money in their account
        if amount_to_withdraw > account_balance:
            print("Error: Insufficient funds for this withdrawal.")
        elif amount_to_withdraw <= 0:
            print("Error: Please enter a valid positive amount.")
        else:
            # Deduct the amount and update the balance bucket
            account_balance -= amount_to_withdraw
            print(f"Withdrawal successful. New balance: {account_balance}")
            
    # Process Option 3: Exit the ATM Application
    elif choice == "3":
        print("Thank you for using the ATM.")
        break  # Completely break out of the infinite loop
        
    # Handle unexpected inputs cleanly
    else:
        print("Invalid choice. Please select 1, 2, or 3.")
```

```python
# Initialize accumulators and tracking variables
score_sum = 0
score_count = 0
highest_score = -1     # Set lower than any valid score to start tracking max
lowest_score = 101     # Set higher than any valid score to start tracking min

# Start the sentinel-controlled loop
while True:
    # Read input from the user
    score = int(input("Enter score (-1 to finish): "))
    
    # Check for the sentinel value to break the loop
    if score == -1:
        break
        
    # Validate that the score falls within the allowed 0-100 range
    if score < 0 or score > 100:
        print("Invalid score. Please enter a value between 0 and 100.")
        continue

    # Update accumulators
    score_sum += score
    score_count += 1
    
    # Track the highest score found so far
    if score > highest_score:
        highest_score = score
        
    # Track the lowest score found so far
    if score < lowest_score:
        lowest_score = score

# After the loop, check if any valid scores were processed
if score_count > 0:
    # Compute the average score
    average_score = score_sum / score_count
    
    # Output the final statistics
    print(f"Students: {score_count}")
    print(f"Average: {average_score:.2f}")
    print(f"Highest: {highest_score}")
    print(f"Lowest: {lowest_score}")
else:
    print("No scores were entered.")
```

```python
# Hard-code the minimum stock threshold
MIN_THRESHOLD = 10

# Initialize an accumulator to track total alerts issued
alert_count = 0

# Create an infinite loop to monitor items until the user is finished
while True:
    # Ask for the product name first to check for the sentinel value
    product_name = input("Product name (or 'done'): ")
    
    # Check for the sentinel value to break the loop cleanly
    if product_name.lower() == "done":
        break
        
    # Ask for the current quantity of that product
    quantity = int(input("Quantity: "))
    
    # Trigger an alert if the quantity falls below the minimum threshold
    if quantity < MIN_THRESHOLD:
        print(f"ALERT: {product_name} stock is low ({quantity} units).")
        alert_count += 1

# Print the final summary after exiting the loop
print(f"Total alerts issued: {alert_count}")
```

```python
# Constants and initial tracking states
TOTAL_SEATS = 8
booked_seats = 0
passenger_list = []

print(f"--- Welcome to City Bus Bookings (Total Seats: {TOTAL_SEATS}) ---")

# The loop runs dynamically while there are empty seats available
while booked_seats < TOTAL_SEATS:
    # Ask the passenger for their name or entry of the supervisor sentinel string
    passenger_name = input("Enter passenger name (or type 'close' to exit): ")
    
    # Handle the bonus challenge early exit sequence
    if passenger_name.strip().lower() == "close":
        print("\nBooking window has been closed early by the supervisor.")
        break
        
    # Check for empty input strings to prevent blank bookings
    if not passenger_name.strip():
        print("Invalid name. Please try again.")
        continue
        
    # Process successful seat registration
    passenger_list.append(passenger_name.strip())
    booked_seats += 1
    seats_left = TOTAL_SEATS - booked_seats
    
    print(f"Seat booked for {passenger_name}. Seats left: {seats_left}")

# Post-loop assessment: Determine if the bus filled up completely or closed early
if booked_seats == TOTAL_SEATS:
    print("\nThe bus is now full! Cannot accept more passengers.")
    
print(f"Total passengers boarded: {booked_seats}")
```

```python
# Initialize running totals for the daily tracking system
total_books_returned = 0
total_fees_collected = 0

# Start the infinite menu loop
while True:
    # Display the system options menu layout
    print("\n1. Record returned book")
    print("2. Show summary")
    print("3. Exit")
    print("4. Reset daily counters")  # Challenge feature addition
    
    # Capture the staff choice string
    choice = input("Choice: ")
    
    # Option 1: Process and record a returned book
    if choice == "1":
        book_title = input("Book title: ")
        days_late = int(input("Days late: "))
        
        # Guard clause: Ensure days late cannot be an accidental negative number
        if days_late < 0:
            print("Invalid input. Days late cannot be negative.")
            continue
            
        # Calculate individual late fee (10 currency units per day)
        current_fee = days_late * 10
        print(f"Late fee: {current_fee}")
        
        # Accumulate metrics into the global daily bucket states
        total_books_returned += 1
        total_fees_collected += current_fee
        
    # Option 2: Print out itemized performance metrics accumulated so far
    elif choice == "2":
        print(f"Books returned today: {total_books_returned}")
        print(f"Total late fees: {total_fees_collected}")
        
    # Option 3: Break out of the execution loop cleanly
    elif choice == "3":
        print("Library system closed. Have a nice day!")
        break
        
    # Challenge Option 4: Flush existing variables to zero for a new shift
    elif choice == "4":
        total_books_returned = 0
        total_fees_collected = 0
        print("Daily counters successfully reset to zero.")
        
    # Catch-all conditional layer to keep wrong options from failing execution
    else:
        print("Warning: Invalid option. Please select 1, 2, 3, or 4.")
```
