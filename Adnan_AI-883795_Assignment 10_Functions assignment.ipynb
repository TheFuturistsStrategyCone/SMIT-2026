```python
def print_electricity_bill(units):
    """Calculates an electricity bill based on tiered slabs and prints the slip."""
    if units <= 100:
        bill = units * 10
    elif units <= 200:
        bill = (100 * 10) + ((units - 100) * 15)
    else:
        bill = (100 * 10) + (100 * 15) + ((units - 200) * 20)

    print(f"Units Used: {units}")
    print(f"Total Bill: {bill} PKR")


# Example usage:
print_electricity_bill(80)
print_electricity_bill(250)
```

```python
def highest_bid(bid1, bid2, bid3):
    """Returns the highest of three bids using if/elif/else without max()."""
    if bid1 >= bid2 and bid1 >= bid3:
        return bid1
    elif bid2 >= bid1 and bid2 >= bid3:
        return bid2
    else:
        return bid3


# Example usage:
print(highest_bid(5000, 7500, 6200))
print(highest_bid(300, 300, 250))
```

```python
def bmi_category(weight_kg, height_m):
    """Calculates BMI from weight (kg) and height (m) and returns the BMI category."""
    bmi = weight_kg / (height_m * height_m)

    if bmi < 18.5:
        return "Underweight"
    elif bmi < 25:
        return "Normal"
    elif bmi < 30:
        return "Overweight"
    else:
        return "Obese"


# Example usage:
print(bmi_category(70, 1.75))
print(bmi_category(95, 1.70))
print(bmi_category(50, 1.75))
```

```python
def word_frequency(sentence):
    """Counts the frequency of each word in a sentence (case-insensitive) and returns a dictionary."""
    words = sentence.lower().split()
    counts = {}

    for word in words:
        counts[word] = counts.get(word, 0) + 1

    return counts


# Example usage:
print(word_frequency("to be or not to be"))
print(word_frequency("Data is power and data is growth"))
```

```python
import math


def courier_charge(weight_kg, is_express):
    """Calculates courier charges based on parcel weight (rounding partial extra kg up) and express option."""
    if weight_kg <= 1:
        charge = 200
    else:
        extra_kg = math.ceil(weight_kg - 1)
        charge = 200 + extra_kg * 80

    if is_express:
        charge = charge * 1.5

    return charge


# Example usage:
print(courier_charge(0.5, False))
print(courier_charge(3.5, False))
print(courier_charge(3.5, True))
print(courier_charge(1, True))
```

```python
def gym_membership_price(months, age, is_student):
    """Calculates gym membership cost with sequential duration and age/student discounts."""
    price = 3000 * months

    # Layer 1: Duration discount (mutually exclusive)
    if months >= 12:
        price *= 0.80
    elif months >= 6:
        price *= 0.90

    # Layer 2: Senior or student discount on the reduced price (mutually exclusive)
    if age >= 60:
        price *= 0.75
    elif is_student:
        price *= 0.85

    return round(price, 2)


# Example usage:
print(gym_membership_price(3, 30, False))
print(gym_membership_price(12, 21, True))
print(gym_membership_price(6, 65, True))
```

```python
def atm_withdraw(balance, amount, withdrawn_today):
    """Validates an ATM withdrawal against amount rules, daily limits, and available balance."""
    if amount <= 0 or amount % 500 != 0:
        return "Invalid Amount"
    elif withdrawn_today + amount > 50000:
        return "Daily Limit Exceeded"
    elif amount > balance:
        return "Insufficient Balance"
    else:
        return balance - amount


# Example usage:
print(atm_withdraw(20000, 5000, 0))
print(atm_withdraw(20000, 1200, 0))
print(atm_withdraw(100000, 30000, 25000))
print(atm_withdraw(3000, 5000, 0))
```

```python
def restock_report(stock, sales, threshold):
    """Processes daily sales against a copy of stock and returns updated inventory, low-stock items, and failed sales."""
    updated_stock = stock.copy()
    failed_sales = []

    for sale in sales:
        item = sale["item"]
        qty = sale["qty"]

        if item not in updated_stock or qty > updated_stock[item]:
            failed_sales.append(item)
        else:
            updated_stock[item] -= qty

    low_stock = sorted([item for item, qty in updated_stock.items() if qty < threshold])

    return {
        "updated_stock": updated_stock,
        "low_stock": low_stock,
        "failed_sales": failed_sales,
    }


# Example usage:
shop_stock = {"Rice": 50, "Oil": 20, "Sugar": 15, "Tea": 8}
todays_sales = [
    {"item": "Rice", "qty": 10},
    {"item": "Oil", "qty": 25},
    {"item": "Sugar", "qty": 10},
    {"item": "Salt", "qty": 2},
]
print(restock_report(shop_stock, todays_sales, 10))
```

```python
def create_account(owner, balance):
    """Creates and returns a bank account dictionary."""
    return {
        "owner": owner,
        "balance": balance,
        "history": [],
    }


def deposit(account, amount):
    """Deposits a positive amount into the account and records it in history."""
    if amount <= 0:
        return "Invalid Amount"
    account["balance"] += amount
    account["history"].append(f"Deposit: {amount}")
    return "Deposit Successful"


def withdraw(account, amount):
    """Withdraws an amount if valid and covered by the account balance."""
    if amount <= 0:
        return "Invalid Amount"
    if amount > account["balance"]:
        return "Insufficient Balance"
    account["balance"] -= amount
    account["history"].append(f"Withdraw: {amount}")
    return "Withdraw Successful"


def transfer(sender, receiver, amount):
    """Transfers money from sender to receiver with a 1% fee deducted from sender."""
    if amount <= 0:
        return "Invalid Amount"
    fee = round(amount * 0.01, 2)
    total_deduction = amount + fee
    if total_deduction > sender["balance"]:
        return "Transfer Failed"

    sender["balance"] -= total_deduction
    receiver["balance"] += amount

    sender["history"].append(f"Transfer to {receiver['owner']}: {amount} (Fee: {fee})")
    receiver["history"].append(f"Transfer from {sender['owner']}: {amount}")
    return "Transfer Successful"


def print_statement(account):
    """Prints the account owner, current balance, and numbered transaction history."""
    print(f"Owner: {account['owner']}")
    print(f"Balance: {account['balance']}")
    print("History:")
    for idx, entry in enumerate(account["history"], start=1):
        print(f"{idx}. {entry}")


# Main program:
ali = create_account("Ali", 10000)
sara = create_account("Sara", 5000)

print(deposit(ali, 5000))
print(withdraw(ali, 20000))
print(withdraw(ali, 3000))
print(transfer(ali, sara, 2000))
print(deposit(sara, -100))

print_statement(ali)
print_statement(sara)
```

```python
def add_item(cart, name, price, qty):
    """Adds an item to the shopping cart or increases its quantity if already present."""
    if name in cart:
        cart[name]["qty"] += qty
    else:
        cart[name] = {"price": price, "qty": qty}
    return cart


def remove_item(cart, name):
    """Removes an item from the cart if present and returns a status message."""
    if name not in cart:
        return "Item not found"
    del cart[name]
    return "Item removed"


def calculate_subtotal(cart):
    """Calculates and returns the total cost of all items in the cart before discounts."""
    return sum(details["price"] * details["qty"] for details in cart.values())


def get_discount(subtotal, coupon):
    """Calculates the discount amount based on the coupon code and subtotal."""
    if coupon == "SAVE10":
        return round(subtotal * 0.10, 2)
    elif coupon == "FLAT500" and subtotal >= 3000:
        return 500
    else:
        return 0


def print_invoice(cart, coupon):
    """Prints the itemized shopping cart invoice including subtotal, discount, tax, and final amount."""
    for name, details in cart.items():
        item_total = details["price"] * details["qty"]
        print(f"{name} — Qty: {details['qty']} — Total: {item_total}")

    subtotal = calculate_subtotal(cart)
    discount = get_discount(subtotal, coupon)
    taxable_amount = subtotal - discount
    tax = round(taxable_amount * 0.05, 2)
    final_amount = round(taxable_amount + tax, 2)

    print(f"Subtotal: {subtotal}")
    print(f"Discount: {discount}")
    print(f"Tax (5%): {tax}")
    print(f"Final Amount: {final_amount}")


# Main program:
cart = {}

add_item(cart, "Shoes", 3000, 1)
add_item(cart, "Socks", 250, 3)
add_item(cart, "Cap", 600, 1)
add_item(cart, "Socks", 250, 1)

print(remove_item(cart, "Watch"))
print(remove_item(cart, "Cap"))

print_invoice(cart, "SAVE10")
```

```python
def calculate_average(marks):
    """Calculates and returns the average of a list of marks rounded to 1 decimal place."""
    return round(sum(marks) / len(marks), 1)


def get_grade(avg):
    """Returns the letter grade corresponding to the given average."""
    if avg >= 90:
        return "A"
    elif avg >= 80:
        return "B"
    elif avg >= 70:
        return "C"
    elif avg >= 60:
        return "D"
    else:
        return "F"


def build_result(name, marks):
    """Builds and returns a student's result dictionary with name, average, grade, and pass/fail status."""
    avg = calculate_average(marks)
    grade = get_grade(avg)
    status = "Pass" if all(m >= 40 for m in marks) else "Fail"

    return {
        "name": name,
        "average": avg,
        "grade": grade,
        "status": status,
    }


def find_topper(results):
    """Finds and returns the name of the student with the highest average (first in case of tie)."""
    if not results:
        return "No students"

    topper = results[0]
    for student in results[1:]:
        if student["average"] > topper["average"]:
            topper = student

    return topper["name"]


def print_result_sheet(results):
    """Prints the full class result sheet, topper, and overall pass rate."""
    if not results:
        print("No students")
        return

    print("--- Result Sheet ---")
    passed_count = 0

    for res in results:
        print(f"{res['name']} | Avg: {res['average']} | Grade: {res['grade']} | {res['status']}")
        if res["status"] == "Pass":
            passed_count += 1

    topper_name = find_topper(results)
    pass_rate = round((passed_count / len(results)) * 100, 1)

    print(f"Topper: {topper_name}")
    print(f"Pass Rate: {pass_rate}%")


# Main program:
students = [
    ("Ali", [85, 90, 78]),
    ("Sara", [95, 88, 92]),
    ("Omar", [40, 35, 50]),
    ("Hina", [72, 65, 80]),
]

results = []
for name, marks in students:
    results.append(build_result(name, marks))

print_result_sheet(results)
```

```python
def calculate_overtime_pay(hours):
    """Calculates overtime pay at 300 PKR per hour."""
    return hours * 300


def calculate_bonus(basic, performance):
    """Calculates performance bonus based on basic salary and rating."""
    rates = {
        "excellent": 0.20,
        "good": 0.10,
        "average": 0.05,
    }
    rate = rates.get(performance, 0)
    return basic * rate


def calculate_tax(gross):
    """Calculates progressive income tax across salary brackets."""
    if gross <= 50000:
        return 0
    elif gross <= 100000:
        return (gross - 50000) * 0.05
    else:
        return (50000 * 0.05) + ((gross - 100000) * 0.12)


def generate_payslip(employee):
    """Generates and returns a dictionary containing an employee's complete payslip breakdown."""
    overtime = round(calculate_overtime_pay(employee["overtime_hours"]), 2)
    bonus = round(calculate_bonus(employee["basic"], employee["performance"]), 2)
    gross = round(employee["basic"] + overtime + bonus, 2)
    tax = round(calculate_tax(gross), 2)
    insurance = 1000
    net = round(gross - tax - insurance, 2)

    return {
        "name": employee["name"],
        "overtime": overtime,
        "bonus": bonus,
        "gross": gross,
        "tax": tax,
        "insurance": insurance,
        "net": net,
    }


def print_payslip(slip):
    """Prints the 7 formatted lines of an employee payslip."""
    print(f"Employee: {slip['name']}")
    print(f"Overtime Pay: {slip['overtime']}")
    print(f"Performance Bonus: {slip['bonus']}")
    print(f"Gross Salary: {slip['gross']}")
    print(f"Income Tax: {slip['tax']}")
    print(f"Medical Insurance: {slip['insurance']}")
    print(f"Net Salary: {slip['net']}")


def run_payroll(employees):
    """Processes payroll for all employees, prints payslips, and returns the total net payroll."""
    total_payroll = 0

    for emp in employees:
        slip = generate_payslip(emp)
        print_payslip(slip)
        print("--------")
        total_payroll += slip["net"]

    total_payroll = round(total_payroll, 2)
    print(f"Total Payroll: {total_payroll}")
    return total_payroll


# Main program:
employees = [
    {"name": "Ali", "basic": 50000, "overtime_hours": 20, "performance": "excellent"},
    {"name": "Sara", "basic": 90000, "overtime_hours": 10, "performance": "good"},
    {"name": "Omar", "basic": 30000, "overtime_hours": 0, "performance": "poor"},
]

run_payroll(employees)
```

```python
def print_bill(order):
    """Prints the itemized restaurant bill including subtotal, discount, 8% GST, and total."""
    for entry in order:
        item_total = entry["price"] * entry["qty"]
        print(f"{entry['item']} — Qty: {entry['qty']} — Total: {item_total}")

    subtotal = calculate_subtotal(order)
    discount = apply_discount(subtotal)
    discounted_subtotal = subtotal - discount
    gst = round(discounted_subtotal * 0.08, 2)
    total_bill = round(discounted_subtotal + gst, 2)

    print(f"Subtotal: {subtotal}")
    print(f"Discount: {discount}")
    print(f"GST (8%): {gst}")
    print(f"Total Bill: {total_bill}")


# Example usage:
order_b = [
    {"item": "Burger", "qty": 2, "price": 450},
    {"item": "Pizza", "qty": 1, "price": 1200},
    {"item": "Cold Drink", "qty": 3, "price": 150},
]
print_bill(order_b)
```
