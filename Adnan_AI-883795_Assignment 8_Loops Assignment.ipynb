## Write a program to print numbers from 1 to 10 using a for loop.

```python
def print_number_range(start: int = 1, end: int = 10) -> None:
    """Prints a sequence of numbers from start to end (inclusive) using a for loop.

    Args:
        start: The beginning integer of the sequence. Defaults to 1.
        end: The ending integer of the sequence. Defaults to 10.
    """
    # range(start, end + 1) ensures the 'end' number is inclusive
    for number in range(start, end + 1):
        print(number)


if __name__ == "__main__":
    print_number_range()
```

## Write a program to print all even numbers from 1 to 50.

```python
def print_even_numbers(start: int = 1, end: int = 50) -> None:
    """Generates and prints even numbers from start to end (inclusive).
    
    Optimized by advancing the range sequence via a step configuration 
    rather than evaluating modulo operations on every iteration.
    """
    # Enforce or find the first even number to align our stride
    actual_start = start if start % 2 == 0 else start + 1
    
    # range(start, stop, step) 
    for number in range(actual_start, end + 1, 2):
        print(number)

if __name__ == "__main__":
    print_even_numbers()
```

## Write a program to calculate the sum of numbers from 1 to N.
### Example:

* Enter number: 5
* Output: 15

```python
def calculate_sum_loop(n: int) -> int:
    """Calculates the sum of numbers from 1 to N using an iterative loop.
    
    Time Complexity: O(N)
    Space Complexity: O(1)
    """
    if n < 0:
        raise ValueError("Input integer must be non-negative.")
        
    total_sum = 0
    for number in range(1, n + 1):
        total_sum += number
    return total_sum

if __name__ == "__main__":
    try:
        user_input = int(input("Enter number: "))
        result = calculate_sum_loop(user_input)
        print(f"Output: {result}")
    except ValueError as e:
        print(f"Invalid input: {e}")
```

## Write a program to count how many digits exist in a number using a loop.
### Example:

* Input: 45892
* Output: 5

```python
def count_digits_loop(number: int) -> int:
    """Counts the number of digits in an integer using an iterative while loop.
    
    Time Complexity: O(log10(N))
    Space Complexity: O(1)
    """
    # Normalize negative numbers to prevent an infinite loop
    number = abs(number)
    
    # Handle the base case edge condition explicitly
    if number == 0:
        return 1
        
    digit_count = 0
    while number > 0:
        number //= 10  # Strips the last digit
        digit_count += 1
        
    return digit_count

if __name__ == "__main__":
    try:
        user_input = int(input("Input: "))
        result = count_digits_loop(user_input)
        print(f"Output: {result}")
    except ValueError:
        print("Error: Please provide a valid integer.")
```

## Write a program to calculate the factorial of a number.
### Example:

* Input : 5
* Output 120

```python
def calculate_factorial_loop(n: int) -> int:
    """Calculates the factorial of a number using an iterative loop.
    
    Time Complexity: O(N)
    Space Complexity: O(1)
    """
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers.")
    
    # Base cases: 0! = 1 and 1! = 1
    result = 1
    for i in range(2, n + 1):
        result *= i
        
    return result

if __name__ == "__main__":
    try:
        user_input = int(input("Input: "))
        output = calculate_factorial_loop(user_input)
        print(f"Output: {output}")
    except ValueError as e:
        print(f"Error: {e}")
```

## Write a program to calculate:

* base ^ exponent
* without using **

```python
def calculate_power_loop(base: float, exponent: int) -> float:
    """Calculates base raised to the power of exponent using an iterative loop.
    
    Time Complexity: O(N) where N is the magnitude of the exponent.
    Space Complexity: O(1)
    """
    # Mathematical invariant: Any number to the power of 0 is 1
    if exponent == 0:
        return 1.0

    result = 1.0
    # Use the absolute value of the exponent for the multiplication loop
    abs_exponent = abs(exponent)
    
    for _ in range(abs_exponent):
        result *= base
        
    # If the exponent was negative, invert the final product (e.g., 2^-3 = 1 / (2^3))
    return result if exponent > 0 else 1.0 / result

if __name__ == "__main__":
    try:
        user_base = float(input("Enter base: "))
        user_exponent = int(input("Enter exponent: "))
        
        output = calculate_power_loop(user_base, user_exponent)
        print(f"Output: {output}")
    except ValueError:
        print("Error: Invalid inputs. Base must be a number, exponent must be an integer.")
```

## Write a program to count vowels in a string.
### Example:

```python
def count_vowels_loop(text: str) -> int:
    """Counts the number of vowels in a string using an explicit loop.
    
    Time Complexity: O(N) where N is the length of the string.
    Space Complexity: O(1)
    """
    # Using a set ensures O(1) instantaneous lookup efficiency
    VOWELS = {'a', 'e', 'i', 'o', 'u'}
    vowel_count = 0
    
    # Normalize input string to lowercase to handle casing variance safely
    for char in text.lower():
        if char in VOWELS:
            vowel_count += 1
            
    return vowel_count

if __name__ == "__main__":
    user_input = input("Input: ")
    output = count_vowels_loop(user_input)
    print(f"Output: {output}")
```

## Write a program to find the maximum value in a list without using max().

```python
from typing import List, Union

def find_maximum_loop(numbers: List[Union[int, float]]) -> Union[int, float]:
    """Finds the maximum value in a list using a linear scanning loop.
    
    Time Complexity: O(N) - Passes through the list exactly once.
    Space Complexity: O(1) - Constant memory overhead.
    """
    # Defensive programming guardrail: Reject empty inputs to prevent IndexErrors
    if not numbers:
        raise ValueError("Cannot calculate the maximum value of an empty collection.")
        
    # Seed the tracker with the initial element index 
    current_max = numbers[0]
    
    # Slice the array from index 1 onward to avoid redundant self-comparison
    for number in numbers[1:]:
        if number > current_max:
            current_max = number
            
    return current_max

if __name__ == "__main__":
    # Test suite vector demonstrating handling of mixed negative numbers
    test_dataset = [-12, -3, -45, -1, -7, -22]
    
    try:
        result = find_maximum_loop(test_dataset)
        print(f"Dataset: {test_dataset}")
        print(f"Output Maximum Value: {result}")
    except ValueError as e:
        print(f"Execution Error: {e}")
```

## Write a program to print:

    *
    **
    ***
    ****
    *****

```python
def print_triangle_nested_loop(rows: int = 5) -> None:
    """Generates a left-aligned star triangle pattern using nested loops.
    
    Time Complexity: O(N²) - Scaled quadratically by the row depth.
    Space Complexity: O(1) - Prints directly to the output stream.
    """
    for row in range(1, rows + 1):
        # The inner loop prints exactly 'row' number of stars
        for _ in range(row):
            print("*", end="")
        # Trigger a newline after completing the current row stream
        print()

if __name__ == "__main__":
    print_triangle_nested_loop()
```

## 🏪 1. Shopping Cart System

### Goal:
    Create a shopping cart program that allows users to add items to their cart.
    Stop adding items when the user types "done".

### Steps:

    Ask the user to enter product names.
    
    Add each item to a list.
    
    If user types "done" → use break.
    
    Print all items added to the cart.

### Example:

    Enter item: Shirt  
    Enter item: Shoes  
    Enter item: done  
    Items in your cart: ['Shirt', 'Shoes']

```python
from typing import List

def run_shopping_cart() -> List[str]:
    """Executes an interactive shopping cart terminal session.
    
    Collects item names from user inputs into an internal collection 
    until a terminal 'done' break signal is received.
    """
    cart: List[str] = []
    print("--- Shopping Cart System Initialization ---")
    print("Type 'done' at any point to finalize your selections.\n")
    
    while True:
        # Capture input and strip leading/trailing whitespace boundaries
        user_input = input("Enter item: ").strip()
        
        # Guard clause: Check for the loop termination command
        if user_input.lower() == "done":
            break
            
        # Validation: Prevent users from inserting completely empty rows
        if not user_input:
            print("Warning: Item name cannot be blank. Please try again.")
            continue
            
        cart.append(user_input)
        
    return cart

if __name__ == "__main__":
    items_ordered = run_shopping_cart()
    print(f"Items in your cart: {items_ordered}")
```

## Student Marks Entry

### Goal:
    Record student marks and skip invalid entries.

### Steps:

    Create an empty list marks = [].
    
    Ask user to enter marks for 5 students.
    
    If marks < 0 or > 100 → print “Invalid marks” and continue.
    
    If user enters -1 → stop taking input (break).
    
    Print valid marks and their average.

### Example:

    Enter marks: 90  
    Enter marks: 105 → Invalid marks  
    Enter marks: 80  
    Enter marks: -1 → stop  
    Valid Marks: [90, 80]  
    Average = 85.0

```python
from typing import List, Optional

def collect_student_marks(target_count: int = 5) -> List[float]:
    """Collects valid student marks from console input.
    
    Validates range entries, skips exceptions using continue, 
    and handles early break signals natively.
    """
    marks: List[float] = []
    valid_entries_count = 0

    print(f"--- Student Marks Entry ({target_count} Students Required) ---")
    print("Enter '-1' at any point to stop entry early.\n")

    while valid_entries_count < target_count:
        try:
            user_input = input("Enter marks: ").strip()
            # Handle empty carriage returns gracefully
            if not user_input:
                continue
                
            score = float(user_input)

            # Check for the terminal early exit break condition
            if score == -1:
                print("-> stop")
                break

            # Range validation gatekeeper
            if score < 0 or score > 100:
                print("Invalid marks")
                continue

            # Record clean validated numerical entry
            marks.append(score)
            valid_entries_count += 1

        except ValueError:
            print("Invalid input format. Please enter a valid number.")
            continue

    return marks

def display_metrics(marks_list: List[float]) -> None:
    """Computes and formats the structural statistical metrics from collected marks."""
    print(f"Valid Marks: {marks_list}")
    
    if marks_list:
        average = sum(marks_list) / len(marks_list)
        print(f"Average = {average:.1f}")
    else:
        print("Average = 0.0 (No valid data recorded)")

if __name__ == "__main__":
    collected_marks = collect_student_marks(target_count=5)
    display_metrics(collected_marks)
```

## Class Attendance Tracker

### Goal:
    Track only present students using continue.

### Steps:

    Have a list of students:
    
    students = ["Ali", "Sara", "Ahmed", "Ayesha"]
    
    
    Loop through each student.
    
    Ask "Is <name> present (yes/no)?"
    
    If “no” → use continue to skip.
    
    Add only present students to a present_list.

### Example:

    Is Ali present? yes  
    Is Sara present? no  
    Is Ahmed present? yes  
    Is Ayesha present? yes  
    Present Students: ['Ali', 'Ahmed', 'Ayesha']

```python
from typing import List

def track_attendance(roster: List[str]) -> List[str]:
    """Iterates through a student roster and tracks who is present.
    
    Uses 'continue' to filter out absent entries and isolate active participants.
    """
    present_list: List[str] = []
    
    print("--- Class Attendance Tracker Initialization ---")
    print("Please respond with 'yes' or 'no' for each student.\n")
    
    for student in roster:
        while True:
            # Capture the response and clean up trailing spaces/casing
            status = input(f"Is {student} present (yes/no)? ").strip().lower()
            
            # Input validation boundary
            if status in {"yes", "no"}:
                break
            print("Invalid response. Please type exactly 'yes' or 'no'.")
            
        # Architectural constraint: Skip absent students using 'continue'
        if status == "no":
            continue
            
        # Append only confirmed present students to our tracking array
        present_list.append(student)
        
    return present_list

if __name__ == "__main__":
    # Fixed production registry roster
    class_roster = ["Ali", "Sara", "Ahmed", "Ayesha"]
    
    active_students = track_attendance(class_roster)
    print(f"\nPresent Students: {active_students}")
```

```python
def simulate_traffic_light() -> None:
    """Simulates traffic light behavior using a nested conditional matrix.
    
    Validates input boundaries and routes the appropriate driving action.
    """
    print("--- Traffic Light Simulation System ---")
    
    # Normalize data stream to eliminate case variance and trailing spaces
    light_color = input("Enter light color: ").strip().lower()
    
    # Outer Condition: Structural validation gatekeeper
    VALID_COLORS = {"red", "yellow", "green"}
    
    if light_color in VALID_COLORS:
        # Nested Conditions: Business logic execution mapping
        if light_color == "red":
            print("Stop! Wait for green light.")
            
        elif light_color == "yellow":
            print("Caution! Slow down and prepare to stop.")
            
        elif light_color == "green":
            print("Go! The intersection is clear to proceed.")
            
    else:
        # Fallback handling for corrupted system inputs
        print(f"System Error: '{light_color}' is an invalid traffic signal color.")

if __name__ == "__main__":
    simulate_traffic_light()
```

## 🏨 8. Hotel Room Booking

### Goal:
    Manage simple room booking.

### Task:

    Create a list of rooms [101, 102, 103, 104, 105]
    
    Ask user to choose a room.
    
    If room already booked, print “Room not available.”

### Example:

```python
from typing import List

def run_room_booking() -> None:
    """Manages an interactive hotel room booking loop.
    
    Tracks available inventory and terminates dynamically when all rooms are booked.
    """
    # Use an explicit list to track active inventory pool
    available_rooms: List[int] = [101, 102, 103, 104, 105]
    
    print("--- Hotel Property Management Engine Initialized ---")
    
    # Loop continues dynamically until the available pool is completely empty
    while available_rooms:
        print(f"\nAvailable rooms: {available_rooms}")
        
        try:
            # Capture input and convert to an integer for exact data matching
            chosen_room = int(input("Choose room: ").strip())
            
            # Inventory validation lookup
            if chosen_room in available_rooms:
                available_rooms.remove(chosen_room)
                print(f"Room {chosen_room} booked successfully!")
            else:
                print("Room not available.")
                
        except ValueError:
            print("Invalid input format. Please enter a valid room number integer.")
            
    print("\n--- System Notification: All rooms have been successfully booked. ---")

if __name__ == "__main__":
    run_room_booking()
```

## 📅 9. Daily Task Reminder

### Goal:
    Create a daily to-do list.

### Task:

    Ask user to enter 5 tasks.
    
    Print all tasks using a loop.
    
    Then allow user to remove a completed task.

### Example:

    Enter task 1: Study
    Enter task 2: Gym
    Today's Tasks: ['Study', 'Gym', 'Read']
    Enter completed task: Gym
    Remaining Tasks: ['Study', 'Read']

```python
from typing import List

def manage_daily_tasks(num_tasks: int = 5) -> None:
    """Collects, displays, and mutates a daily to-do list roster.
    
    Ensures safe array modification when removing completed assignments.
    """
    tasks: List[str] = []
    
    print(f"--- Daily Task Reminder System ({num_tasks} Tasks Required) ---\n")
    
    # Task Collection Phase
    while len(tasks) < num_tasks:
        current_index = len(tasks) + 1
        task_input = input(f"Enter task {current_index}: ").strip()
        
        # Guard clause: Prevent blank tasks from corrupting the list
        if not task_input:
            print("Warning: Task cannot be empty. Please enter a valid entry.")
            continue
            
        tasks.append(task_input)
        
    # Task Display Phase
    print(f"\nToday's Tasks: {tasks}")
    
    # Task Deletion / Completion Phase
    if tasks:
        completed_task = input("Enter completed task: ").strip()
        
        # Defensive Programming: Check if the element exists prior to mutation
        if completed_task in tasks:
            tasks.remove(completed_task)
            print(f"Remaining Tasks: {tasks}")
        else:
            print(f"Notification: '{completed_task}' was not found in your active task list.")
            print(f"Unmodified Tasks: {tasks}")

if __name__ == "__main__":
    manage_daily_tasks(num_tasks=5)
```

## Fast Food Ordering with Stop Option

### Goal:
    Simulate ordering from a fast food menu.
    Stop ordering when the user says “stop”.

### Steps:


    menu = ["Burger", "Pizza", "Fries", "Drink"]
        
    Use a for loop with a large range (like for i in range(100)).
        
    Ask the user to input an item.
        
    If item not in menu → continue.
        
    If item == “stop” → break.
        
    Store valid items and show total order.

### Example:

    Enter item: Burger  
    Enter item: Ice Cream → Not on menu  
    Enter item: Fries  
    Enter item: stop  
    Your order: ['Burger', 'Fries']

```python
from typing import List, Set

def run_fast_food_ordering() -> None:
    """Simulates an interactive fast-food ordering transaction session.
    
    Implements a capped allocation range loop with conditional break parameters
    to isolate and aggregate verified menu selections.
    """
    # Core menu dataset array
    menu: List[str] = ["Burger", "Pizza", "Fries", "Drink"]
    
    # Architecture Optimization: Convert list to set for instantaneous O(1) membership lookups
    menu_lookup: Set[str] = {item.lower() for item in menu}
    
    order_bucket: List[str] = []
    
    print("--- Fast Food POS Engine Initialized ---")
    print(f"Available Menu: {menu}")
    print("Type 'stop' at any point to finalize your receipt.\n")
    
    # Constrained loop execution boundary matching prompt requirements
    for _ in range(100):
        raw_input = input("Enter item: ").strip()
        normalized_input = raw_input.lower()
        
        # Termination condition rule
        if normalized_input == "stop":
            break
            
        # Menu exception filtering block
        if normalized_input not in menu_lookup:
            print(f"'{raw_input}' is not on the menu.")
            continue
            
        # Re-map the lowercase input back to clean title-case for corporate inventory storage
        clean_item_name = raw_input.capitalize()
        order_bucket.append(clean_item_name)
        
    print(f"\nYour order: {order_bucket}")

if __name__ == "__main__":
    run_fast_food_ordering()
```

## ATM Simulation

### Goal:
    Simulate simple ATM operations using for and stop after 3 tries.

### Steps:

    Set correct PIN = 1234.
    
    Use for i in range(3) for 3 attempts.
    
    If wrong PIN → print “Try again” and continue.
    
    If correct PIN → print menu and break.

### Example:

    Enter PIN: 1111 → Wrong PIN. Try again.
    Enter PIN: 2222 → Wrong PIN. Try again.
    Enter PIN: 1234 → Access Granted!

```python
def run_atm_simulation() -> None:
    """Simulates an ATM credential verification gateway.
    
    Constrains access limits using a sequential range loop capped at 3 attempts,
    halting processing immediately upon a positive match.
    """
    CORRECT_PIN: str = "1234"
    max_attempts: int = 3
    authenticated: bool = False

    print("--- ATM Authentication Terminal ---")
    
    for attempt in range(max_attempts):
        # Capture input as a string to preserve leading zeros if present
        user_input = input("Enter PIN: ").strip()
        
        # Validation evaluation block
        if user_input != CORRECT_PIN:
            print("Wrong PIN. Try again.")
            continue
            
        # Authentication success gate
        print("Access Granted!")
        authenticated = True
        break

    # Security Lockout enforcement block
    if not authenticated:
        print("\n[SECURITY ALERT]: 3 incorrect attempts recorded. Your card has been retained.")

if __name__ == "__main__":
    run_atm_simulation()
```

## Bank Transaction Log

### Goal:
    Record 5 transactions and skip invalid amounts.

### Steps:

    Ask the user to enter deposit amounts (positive integers).
    
    If amount <= 0 → continue.
    
    If amount > 10000 → stop using break.

Example:

    Enter amount: 200

## Electricity Bill Tracker

### Goal:
    Take user input for monthly bills, stop if user enters 0.

### Steps:

    Loop 12 times for months.
    
    If bill = 0 → break
    
    Add valid bills to list.
    
    Print total and average.

### Example:

    Enter bill for month 1: 1500  
    Enter bill for month 2: 1800  
    Enter bill for month 3: 0 → Stop  
    Total = 3300, Average = 1650.0

```python
from typing import List

def run_electricity_bill_tracker(max_months: int = 12) -> None:
    """Tracks monthly electricity utility consumption statements across a fiscal calendar.
    
    Processes numerical inputs, aggregates total financial obligations,
    and calculates statistical averages.
    """
    monthly_bills: List[float] = []
    
    print(f"--- Fiscal Electricity Bill Aggregator ({max_months}-Month Cycle) ---")
    print("Enter '0' at any point to stop entry early.\n")
    
    # Loop exactly 12 times representing the calendar months
    for month_index in range(1, max_months + 1):
        try:
            # Capture the input and cast to a float to accommodate cents/decimals accurately
            user_input = input(f"Enter bill for month {month_index}: ").strip()
            
            # Guard clause: handle empty carriage returns safely
            if not user_input:
                print("Warning: Input cannot be empty. Please enter a valid numerical figure.")
                continue
                
            bill_amount = float(user_input)
            
            # 1. Negative amount filter safety check
            if bill_amount < 0:
                print("Error: Utility bills cannot be negative amounts. Please try again.")
                continue
                
            # 2. Terminal loop exit condition check
            if bill_amount == 0:
                print("-> Stop")
                break
                
            # 3. Commit valid record to our memory ledger list
            monthly_bills.append(bill_amount)
            
        except ValueError:
            print("Invalid input format. Please enter a valid numeric value.")
            continue
            
    # --- Statistics Compilation Phase ---
    total_consumption = sum(monthly_bills)
    
    # Defensive Boundary Verification: Check if the list contains entries to avoid dividing by zero
    if monthly_bills:
        average_consumption = total_consumption / len(monthly_bills)
    else:
        average_consumption = 0.0
        
    print(f"\nTotal = {total_consumption:.0f}, Average = {average_consumption:.1f}")

if __name__ == "__main__":
    run_electricity_bill_tracker(max_months=12)

```

## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python
### import sys
from typing import Optional


class BankAccount:
    """Enterprise-grade service engine managing single-account financial operations.

    Handles data validation boundaries, balance state integrity mutations,
    and runtime security exceptions.
    """

    def __init__(self) -> None:
        self._account_number: Optional[str] = None
        self._account_holder: Optional[str] = None
        self._balance: float = 0.0

    def create_account(self, account_number: str, holder_name: str) -> None:
        """Initializes account parameters within the state ledger."""
        if not account_number or not holder_name:
            print("[ERROR]: Account details cannot be empty or blank.")
            return

        self._account_number = account_number
        self._account_holder = holder_name
        self._balance = 0.0
        print(
            f"\n[SUCCESS]: Account registered for {self._account_holder} (Acc No: {self._account_number})!"
        )

    def deposit(self, amount: float) -> None:
        """Validates and mutates state ledger by adding capital assets."""
        if not self._is_account_active():
            return

        if amount <= 0:
            print("[INVALID]: Deposit amount must be a positive integer or decimal.")
            return

        self._balance += amount
        print(f"[SUCCESS]: Deposited PKR {amount:,.2f}. Current Balance: PKR {self._balance:,.2f}")

    def withdraw(self, amount: float) -> None:
        """Evaluates liquid solvency and deducts capital resources from state ledger."""
        if not self._is_account_active():
            return

        if amount <= 0:
            print("[INVALID]: Withdrawal amount must be a positive integer or decimal.")
            return

        # Solvency verification gatekeeper
        if amount > self._balance:
            print(
                f"[DENIED]: Insufficient funds. Available Liquidity: PKR {self._balance:,.2f}"
            )
            return

        self._balance -= amount
        print(f"[SUCCESS]: Withdrew PKR {amount:,.2f}. Current Balance: PKR {self._balance:,.2f}")

    def check_balance(self) -> None:
        """Streams verified real-time financial tracking figures."""
        if not self._is_account_active():
            return

        print(f"\n--- Account Statement ---")
        print(f"Holder Name: {self._account_holder}")
        print(f"Account No : {self._account_number}")
        print(f"Net Balance: PKR {self._balance:,.2f}")

    def _is_account_active(self) -> bool:
        """Internal security guard checking for object lifecycle initialization."""
        if self._account_number is None:
            print("[DENIED]: Access Error. No active account found. Please Create an Account first.")
            return False
        return True


def execute_banking_terminal() -> None:
    """Orchestrates the terminal-driven menu loop session interface."""
    atm_service = BankAccount()

    print("========================================")
    print("      Welcome to Apex Bank Limited      ")
    print("========================================")

    while True:
        print("\n=== MAIN INTERFACE MENU ===")
        print("1. Create Account")
        print("2. Deposit Money")
        print("3. Withdraw Money")
        print("4. Check Balance Statement")
        print("5. Exit Secure Terminal")

        choice = input("\nSelect menu option (1-5): ").strip()

        try:
            if choice == "1":
                acc_num = input("Enter customized account number: ").strip()
                name = input("Enter account holder legal name: ").strip()
                atm_service.create_account(acc_num, name)

            elif choice == "2":
                amount = float(input("Enter deposit amount (PKR): ").strip())
                atm_service.deposit(amount)

            elif choice == "3":
                amount = float(input("Enter withdrawal amount (PKR): ").strip())
                atm_service.withdraw(amount)

            elif choice == "4":
                atm_service.check_balance()

            elif choice == "5":
                print("\n[INFO]: Terminating secure session. Thank you for choosing Apex Bank.")
                break

            else:
                print("[INVALID]: Selection out of bounds. Please pick an option from 1 to 5.")

        except ValueError:
            print("[CRITICAL ERROR]: Numeric conversion failure. Please input valid numbers only.")


if __name__ == "__main__":
    execute_banking_terminal()
```

## School Marks Report
### Goal:

    Take input for multiple students’ marks in 3 subjects and calculate their total and average.

### Steps:

    Outer loop → for each student (3 students).
    
    Inner loop → take marks for 3 subjects.
    
    Store each student’s total in a list.
    
    Print all totals and class average.

### Example:
    Enter marks for Student 1:
    Subject 1: 80
    Subject 2: 75
    Subject 3: 90
    Enter marks for Student 2:
    Subject 1: 60
    Subject 2: 70
    Subject 3: 65
    → Output: [245, 195]

```python
from typing import List

def generate_marks_report(total_students: int = 3, total_subjects: int = 3) -> None:
    """Collects individual subject marks for multiple students using nested loops.
    
    Validates range parameters, isolates totals in a list ledger, 
    and calculates final class statistics.
    """
    student_totals: List[float] = []

    print(f"--- Academic Performance Reporting System ---")
    print(f"Configured for {total_students} Students across {total_subjects} Subjects.\n")

    # Outer Loop: Manage individual student entries
    for student_idx in range(1, total_students + 1):
        print(f"Enter marks for Student {student_idx}:")
        current_student_sum = 0.0
        subject_idx = 1
        
        # Inner Loop: Sequentially gather subject data points
        while subject_idx <= total_subjects:
            try:
                user_input = input(f"  Subject {subject_idx}: ").strip()
                
                # Intercept empty carriage returns
                if not user_input:
                    continue
                    
                score = float(user_input)
                
                # Business Validation Gatekeeper: Enforce realistic testing metrics
                if score < 0 or score > 100:
                    print("  [INVALID]: Marks must be between 0 and 100. Please re-enter.")
                    continue
                
                current_student_sum += score
                subject_idx += 1  # Advance the subject pointer safely upon valid input
                
            except ValueError:
                print("  [INVALID]: Numeric format exception. Please enter a valid number.")
                continue
                
        # Persist the processed total inside the data ledger array
        # Using int conversion if it's a whole number to match example output requirements
        formatted_total = int(current_student_sum) if current_student_sum.is_integer() else current_student_sum
        student_totals.append(formatted_total)

    # --- Metrics Compilation Phase ---
    print("\n--- Final Performance Metrics ---")
    print(f"Output: {student_totals}")
    
    if student_totals:
        class_average = sum(student_totals) / len(student_totals)
        print(f"Class Average Total Score: {class_average:.2f}")


if __name__ == "__main__":
    # Standard profile execution matching prompt requirements
    generate_marks_report(total_students=3, total_subjects=3)
```

## Shop Sales Tracker
### Goal:

    Track daily sales of 3 products over 5 days and calculate the total revenue.

### Steps:

    Outer loop → iterate over 5 days.
    
    Inner loop → take sales amount for 3 products each day.
    
    Store daily total in a list.
    
    If any product sale = 0 → continue (skip that product).
    
    Print all daily totals and grand total.

### Example:
    Day 1:
    Product 1: 200
    Product 2: 150
    Product 3: 0 → skipped
    Day 2:
    Product 1: 300
    Product 2: 250
    Product 3: 400
    → Daily Totals = [350, 950], Grand Total = 1300

```python
# Create an empty list to store the total sales for each day
daily_totals = []

print("--- Shop Sales Tracker ---")

# Outer loop: Runs for 5 days
for day in range(1, 6):
    print(f"Day {day}:")
    day_sum = 0  # Reset the daily sum for the new day
    
    # Inner loop: Runs for 3 products each day
    product = 1
    while product <= 3:
        user_input = input(f"  Product {product}: ").strip()
        
        # Guard clause: If the user inputs nothing, ask again
        if not user_input:
            continue
            
        sales_amount = int(user_input)
        
        # Step 1: If sales amount is 0, skip the product using 'continue'
        if sales_amount == 0:
            print("    → skipped")
            product += 1  # Move to the next product
            continue      # Skip the rest of the code and restart loop
            
        # Step 2: Add valid sales to the daily sum
        day_sum += sales_amount
        product += 1  # Move to the next product

    # Save the day's total to our list
    daily_totals.append(day_sum)

# Calculate the final grand total of all days
grand_total = sum(daily_totals)

# Display the final results
print("\n--- Final Report ---")
print(f"Daily Totals = {daily_totals}")
print(f"Grand Total  = {grand_total}")
```

## Invoice Tax Calculator
### Goal:

Take multiple items in multiple invoices and calculate tax per invoice.

### Steps:

Outer loop → for 3 invoices.

Inner loop → 4 items each.

Input item price, add to list.

Calculate total + 10% tax.

Print invoice total.

### Example:
    Invoice 1:
    Items: [100, 200, 300, 400]
    Invoice 2:
    Items: [250, 150, 350, 250]
    → Totals: [1100, 1100]

```python
print("--- Invoice Tax Calculator ---")

# Create an empty list to store the final grand totals (with tax) for all invoices
invoice_grand_totals = []

# Outer loop: Iterates through each of the 3 invoices
for invoice_num in range(1, 4):
    print(f"\nInvoice {invoice_num}:")
    invoice_items = []  # Reset the list to hold the current invoice's items
    
    # Inner loop: Gathers prices for exactly 4 items per invoice
    item_count = 1
    while item_count <= 4:
        try:
            user_input = input(f"  Item {item_count} Price: ").strip()
            
            # Guard clause: handle empty returns safely
            if not user_input:
                continue
                
            price = float(user_input)
            
            # Input validation: Prices cannot be negative numbers
            if price < 0:
                print("    Invalid! Price cannot be negative. Please re-enter.")
                continue
                
            # Add valid price to our current item list
            invoice_items.append(price)
            item_count += 1  # Move to the next item
            
        except ValueError:
            print("    Invalid entry! Please enter a valid number.")
            continue
            
    # Calculate the raw cost of items on this invoice
    raw_subtotal = sum(invoice_items)
    
    # Calculate the final total including a 10% tax charge (Multiplier: 1.10)
    final_total_with_tax = raw_subtotal * 1.10
    
    # Save the final formatted result into our summary history tracking ledger
    invoice_grand_totals.append(round(final_total_with_tax, 2))
    
    # Print the specific summary data for this invoice
    print(f"  Items Recorded: {invoice_items}")
    print(f"  → Total (with 10% tax): {final_total_with_tax:.2f}")

# Display final system logs matching your example matrix layout
print("\n--- Final Accounting Summary ---")
print(f"→ Totals: {invoice_grand_totals}")
```

                print("    Invalid! Marks must be between 0 and 100.")
                continue
                
            # Step 1: If marks are below 33, print fail warning immediately
            if marks < 33:
                print(f"    → Failed in Subject {subject}")
                failed_any = True
                
            student_marks.append(marks)
            subject += 1  # Move to next subject
            
        except ValueError:
            print("    Invalid entry! Please enter a whole number.")
            continue
            
    # Step 2: Calculate total score and determine Grade if they didn't fail
    total_score = sum(student_marks)
    average_score = total_score / 4
    
    # Simple grading conditions
    if failed_any:
        final_status = "Result: FAIL (Failed component detected)"
    elif average_score >= 80:
        final_status = "Grade: A"
    elif average_score >= 60:
        final_status = "Grade: B"
    elif average_score >= 45:
        final_status = "Grade: C"
    else:
        final_status = "Grade: D"
        
    # Print student summary report
    print(f"  Marks Record: {student_marks} | Total: {total_score} | {final_status}")

print("\n--- All Reports Generated Successfully ---")


## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python
#### print("--- Holiday Package Evaluator ---")

# Outer loop: Iterates through each of the 3 holiday packages
for package_num in range(1, 4):
    print(f"\nPackage {package_num}:")
    valid_ratings = []  # Reset the list to store the active ratings for this package
    
    # Inner loop: Collects ratings for 3 features (Hotel, Transport, Food)
    feature_count = 1
    while feature_count <= 3:
        # Assign a clear label to each feature index to make the prompt readable
        if feature_count == 1:
            feature_name = "Hotel"
        elif feature_count == 2:
            feature_name = "Transport"
        else:
            feature_name = "Food"
            
        try:
            user_input = input(f"  Rating for {feature_name} (0-5): ").strip()
            
            # Guard clause: handle empty entries safely
            if not user_input:
                continue
                
            rating = int(user_input)
            
            # Input validation: Ensure the rating falls strictly inside the 0 to 5 boundary
            if rating < 0 or rating > 5:
                print("    Invalid! Rating must be between 0 and 5. Please re-enter.")
                continue
                
            # Step 1: If rating is 0, skip this feature completely using 'continue'
            if rating == 0:
                print("    → skipped")
                feature_count += 1  # Move to the next feature index
                continue            # Jump back to the top of the while loop
                
            # Step 2: Add valid non-zero ratings to our calculation list
            valid_ratings.append(rating)
            feature_count += 1
            
        except ValueError:
            print("    Invalid entry! Please enter a whole number from 0 to 5.")
            continue
            
    # Step 3: Calculate and print average rating based only on evaluated components
    if valid_ratings:
        package_average = sum(valid_ratings) / len(valid_ratings)
        print(f"→ Package {package_num} Ratings: {valid_ratings} | Avg = {package_average:.2f}")
    else:
        print(f"→ Package {package_num} Ratings: None | Avg = 0.0 (No features evaluated)")

print("\n--- All Packages Evaluated Successfully ---")
```
